## 1. range()

### 문제 1-1

`range()`의 세 가지 쓰임을 차례로 연습한다.

1. `range(4)`로 `1번 라인 가동 점검` ~ `4번 라인 가동 점검`을 출력한다.
2. `range(1, 11)`로 1부터 10까지의 합을 `total`에 누적해 출력한다.
3. `range(9, 18, 3)`으로 `9시 온도 측정`, `12시 온도 측정`, `15시 온도 측정`을 출력한다.

**실행 결과**

```text
1번 라인 가동 점검
2번 라인 가동 점검
3번 라인 가동 점검
4번 라인 가동 점검
1부터 10까지의 합: 55
9시 온도 측정
12시 온도 측정
15시 온도 측정
```

공식 문서: [range() 함수](https://docs.python.org/ko/3.13/tutorial/controlflow.html#the-range-function)

````python
# 여기에 작성한다.
for i in range(4):
    print(f"{i+1} 라인 가동 점검")
total =0
for i in range(1,11):
    total+=i
print(f"1부터 10까지의 합 :{total}")
for i in range(9,18,3):
    print(f"{i}시 온도 측정")
````

## 2. 컴프리헨션

### 문제 2-1

문자열로 들어온 단가 목록을 리스트 컴프리헨션으로 처리한다.

1. 모두 정수로 바꾼 리스트 `prices`를 만들어 출력한다.
2. 10000 이상인 단가만 모은 리스트를 출력한다.
3. 부가세 10%를 더한 금액(`round()`로 정수) 리스트를 출력한다.

```python
prices_text = ["15000", "8200", "23500", "4900", "12000"]
```

**실행 결과**

```text
[15000, 8200, 23500, 4900, 12000]
[15000, 23500, 12000]
[16500, 9020, 25850, 5390, 13200]
```

공식 문서: [리스트 컴프리헨션](https://docs.python.org/ko/3.13/tutorial/datastructures.html#list-comprehensions)

````python
prices_text = ["15000", "8200", "23500", "4900", "12000"]

prices = [int(v) for v in prices_text]
print(prices)
print([v for v in prices if v >10000])
print([v+int(v*0.1) for v in prices])
# 여기에 작성한다.
````

### 문제 2-2

부품 코드에 공백과 대소문자가 섞여 같은 부품이 여러 번 적혀 있다. 집합 컴프리헨션으로 공백을 없애고 대문자로 바꾼 코드 집합을 만든 뒤, 정렬한 목록과 부품 종류 수를 출력한다.

```python
parts = [" bt-10", "BT-10", "nt-07 ", "Nt-07", "ws-22", " BT-10 "]
```

**실행 결과**

```text
['BT-10', 'NT-07', 'WS-22']
부품 종류: 3개
```

공식 문서: [리스트 컴프리헨션](https://docs.python.org/ko/3.13/tutorial/datastructures.html#list-comprehensions) · [집합](https://docs.python.org/ko/3.13/tutorial/datastructures.html#sets)

## 3. 딕셔너리 반복

### 문제 3-1

지역별 매출 딕셔너리를 `items()`로 반복한다.

1. `서울: 350,000원` 형태로 한 줄씩 출력한다.
2. 다시 반복하면서 매출이 300000 미만인 지역만 `부산: 목표 미달` 형태로 출력한다.

```python
region_sales = {"서울": 350000, "부산": 280000, "대구": 210000}
```

**실행 결과**

```text
서울: 350,000원
부산: 280,000원
대구: 210,000원
부산: 목표 미달
대구: 목표 미달
```

공식 문서: [루프 테크닉](https://docs.python.org/ko/3.13/tutorial/datastructures.html#looping-techniques) · [딕셔너리 뷰 객체](https://docs.python.org/ko/3.13/library/stdtypes.html#dictionary-view-objects)

````python
region_sales = {"서울": 350000, "부산": 280000, "대구": 210000}

# 여기에 작성한다.

for region,account in region_sales.items():
    print(f"{region}: {account:,}원")


for region,account in region_sales.items():
    if account < 300000:
        print(f"{region}: 목표 미만")
````

## 4. get(키, 0)으로 개수·합계 세기

### 문제 4-1

검사에서 나온 불량 유형 목록이다. 빈 딕셔너리 `counts`에 `get(키, 0)`으로 유형별 건수를 누적한 뒤, `스크래치: 3건` 형태로 한 줄씩 출력한다.

```python
defect_types = ["스크래치", "찍힘", "스크래치", "변색", "찍힘", "스크래치", "이물"]
```

**실행 결과**

```text
스크래치: 3건
찍힘: 2건
변색: 1건
이물: 1건
```

공식 문서: [dict.get()](https://docs.python.org/ko/3.13/library/stdtypes.html#dict.get)

````python
defect_types = ["스크래치", "찍힘", "스크래치", "변색", "찍힘", "스크래치", "이물"]
counts = {}
for i in defect_types:
    counts[i] = counts.get(i,0)+1
# 여기에 작성한다.
for j in counts.keys():
    print(f"{j}: {counts.get(j)}건")
````

### 문제 4-2

도시별 출하 기록이 `(도시, 수량)` 튜플의 리스트로 주어진다. 빈 딕셔너리 `totals`에 `get(키, 0)`으로 도시별 출하 수량 합계를 누적한 뒤, `울산: 184개` 형태로 출력한다.

```python
shipments = [("울산", 120), ("창원", 85), ("울산", 64), ("구미", 150), ("창원", 40)]
```

**실행 결과**

```text
울산: 184개
창원: 125개
구미: 150개
```

공식 문서: [dict.get()](https://docs.python.org/ko/3.13/library/stdtypes.html#dict.get)

````python
shipments = [("울산", 120), ("창원", 85), ("울산", 64), ("구미", 150), ("창원", 40)]
totals = {}
for i in shipments:
    totals[i[0]] = totals.get(i[0],0)+ i[1]
# 여기에 작성한다

for j in totals.keys():
    print(f"{j}: {totals.get(j)}개")
````

## 5. enumerate()와 zip()

### 문제 5-1

5일간의 생산량이다. 목표는 하루 1000개다.

1. `enumerate()`로 순번을 1부터 붙여 `1일차: 980개` 형태로 출력한다.
2. 다시 반복하면서 목표에 못 미친 날만 `1일차 목표 미달 (980개)` 형태로 출력한다.

```python
weekly = [980, 1120, 860, 1045, 1010]
```

**실행 결과**

```text
1일차: 980개
2일차: 1,120개
3일차: 860개
4일차: 1,045개
5일차: 1,010개
1일차 목표 미달 (980개)
3일차 목표 미달 (860개)
```

공식 문서: [enumerate()](https://docs.python.org/ko/3.13/library/functions.html#enumerate)

````python
weekly = [980, 1120, 860, 1045, 1010]

for i,v in enumerate(weekly,1):
    print(f"{i}일차 {v:,}개")
# 여기에 작성한다.


for i,v in enumerate(weekly,1):
    if v<1000:
        print(f"{i}일차 목표 미달 ({v:,}개)")
````
