## 1. 제품별 금액과 합계 (벡터화와 브로드캐스팅)

```python
quantity = np.array([5, 3, 7])
unit_price = np.array([12000, 18000, 4500])
```

1. `quantity * unit_price`로 제품별 금액 배열을 만든다.
2. `sum()`으로 전체 금액을 구한다.
3. **부가세율은 10%**로 한다. 제품별 금액에 1.1을 곱해 부가세 포함 금액 배열을 만들고, `astype(int)`로 정수로 바꾼다. 이때 **소수점 이하 값은 버리며, 반올림하지 않는다.** 숫자 하나가 모든 요소에 적용되는 것을 확인한다.
4. 제품별 금액이 50000 이상인 제품 수를 구한다.

공식 문서: [Broadcasting](https://numpy.org/doc/stable/user/basics.broadcasting.html) · [ndarray.astype()](https://numpy.org/doc/stable/reference/generated/numpy.ndarray.astype.html)

````python
quantity = np.array([5, 3, 7])
unit_price = np.array([12000, 18000, 4500])
# 여기에 작성한다.
````

```python
input_qty = np.array([250, 180, 320, 210, 275])
defects = np.array([5, 9, 4, 12, 6])
```

1. 설비별 양품 수량(투입 수량 - 불량 수량)을 출력한다.

3. 불량률이 2 이상이고(`&`) 4 미만인 값만 골라 출력한다.
4. 불량률이 4 이상이면 `"점검"`, 2 이상이면 `"주의"`, 나머지는 `"정상"`인 배열을 `np.where()`를 겹쳐 써서 만든다.

공식 문서: [Boolean 인덱싱](https://numpy.org/doc/stable/user/basics.indexing.html#boolean-array-indexing) · [np.where()](https://numpy.org/doc/stable/reference/generated/numpy.where.html)

## 3. 설비별 기준값 적용 (브로드캐스팅)

3개 설비(M01·M02·M03)의 4회 측정값이 행(측정 회차) × 열(설비) 형태로 주어진다. 설비마다 기준값이 다르며, 기준값은 1행 3열 배열이다.

```python
measurements = np.array([[78, 83, 70], [81, 84, 72], [77, 86, 73], [83, 82, 71]])
limits = np.array([[80, 85, 73]])
names = np.array(["M01", "M02", "M03"])
```

1. `measurements`와 `limits`의 `shape`을 출력하고, 두 배열을 계산할 수 있는 이유를 브로드캐스팅 규칙으로 주석에 적는다.
2. `measurements > limits`로 기준 초과 여부의 불리언 배열을 만든다.
3. 설비별 기준 초과 횟수를 `axis=0`으로 계산한다.
4. 불리언 인덱싱으로 기준 초과가 한 번이라도 있는 설비 이름을 출력한다.

주의: 기준 초과는 `>`를 사용하고, 기준과 같은 값은 초과가 아니다.

공식 문서: [Broadcasting](https://numpy.org/doc/stable/user/basics.broadcasting.html) · [Boolean 인덱싱](https://numpy.org/doc/stable/user/basics.indexing.html#boolean-array-indexing)

````python
measurements = np.array([[78, 83, 70], [81, 84, 72], [77, 86, 73], [83, 82, 71]])
limits = np.array([[80, 85, 73]])
names = np.array(["M01", "M02", "M03"])
# 여기에 작성한다.
````

## 4. 설비 가동 시간 비교 (axis·정렬·argmax)

설비 4대(M01~M04)의 5일간 가동 시간(시간)이다. 행은 설비, 열은 날짜다.

```python
hours = np.array([
    [7.5, 8.0, 6.5, 7.0, 8.0],
    [6.0, 5.5, 7.0, 6.5, 6.0],
    [8.0, 7.5, 8.0, 7.5, 7.0],
    [5.0, 6.5, 6.0, 6.0, 6.0],
])
names = np.array(["M01", "M02", "M03", "M04"])
```

1. 설비별 5일 합계를 구한다. 행이 설비이므로 `axis` 방향에 주의한다.
2. 1번 결과를 큰 순서로 정렬해 출력한다.
3. `np.argmax()`와 `np.argmin()`으로 가동 시간이 가장 긴 설비와 가장 짧은 설비의 이름을 출력한다.
4. 설비별 합계가 네 설비 합계의 평균 이상인 설비 이름을 불리언 인덱싱으로 출력한다.
5. 날짜별 전체 가동 시간을 구하고, 가장 적은 날의 위치 번호를 출력한다.

공식 문서: [np.sort()](https://numpy.org/doc/stable/reference/generated/numpy.sort.html) · [np.argmax()](https://numpy.org/doc/stable/reference/generated/numpy.argmax.html) · [np.argmin()](https://numpy.org/doc/stable/reference/generated/numpy.argmin.html)

````python
hours = np.array([
    [7.5, 8.0, 6.5, 7.0, 8.0],
    [6.0, 5.5, 7.0, 6.5, 6.0],
    [8.0, 7.5, 8.0, 7.5, 7.0],
    [5.0, 6.5, 6.0, 6.0, 6.0],
])
names = np.array(["M01", "M02", "M03", "M04"])
# 여기에 작성한다.
````

## 5. 결측 센서와 점검 대상 구분

3개 설비의 3회 측정값이다. 행은 측정 회차, 열은 설비다. 센서 오류로 빠진 값은 `np.nan`이다.

```python
raw = np.array([[70., 88., 91.], [75., np.nan, 84.], [81., 86., 79.]])
```

1. `np.isnan()`으로 결측값 위치를 확인한다.
2. `np.isnan(raw).sum(axis=0)`으로 설비별 결측 개수를 구한다.
3. `np.nanmean(raw, axis=0)`으로 결측값을 뺀 설비별 평균을 구한다. 결측이 있는 설비는 남은 값만으로 평균을 냈는지 확인한다.
4. 평균이 **85 이상**이면 `"점검"`, 아니면 `"정상"`으로 분류한다.

공식 문서: [np.isnan()](https://numpy.org/doc/stable/reference/generated/numpy.isnan.html) · [np.nanmean()](https://numpy.org/doc/stable/reference/generated/numpy.nanmean.html) · [np.where()](https://numpy.org/doc/stable/reference/generated/numpy.where.html)
