## 문제 1. 주문상품 기록 집계하기

주문상품 기록이 `(지역, 제품, 수량, 단가)` 튜플의 리스트로 주어진다.

1. `get(키, 0)`으로 지역별 매출 합계(수량 × 단가)를 누적해 `서울: 958,000원` 형태로 출력한다.
2. 같은 반복문에서 제품별 판매 수량도 누적해 `모니터: 6개` 형태로 출력한다.
3. 지역별 합계를 `items()`로 반복하며 가장 큰 지역을 찾아 `매출 1위 지역: 부산 (1,050,000원)` 형태로 출력한다.

```python
order_items = [
    ("서울", "모니터", 2, 280000), ("부산", "키보드", 5, 42000),
    ("서울", "마우스", 10, 23000), ("대구", "모니터", 1, 280000),
    ("부산", "모니터", 3, 280000), ("서울", "키보드", 4, 42000),
    ("대구", "마우스", 6, 23000),
]
```

**실행 결과**

```text
서울: 958,000원
부산: 1,050,000원
대구: 418,000원
모니터: 6개
키보드: 9개
마우스: 16개
매출 1위 지역: 부산 (1,050,000원)
```

공식 문서: [dict.get()](https://docs.python.org/ko/3.13/library/stdtypes.html#dict.get) · [루프 테크닉](https://docs.python.org/ko/3.13/tutorial/datastructures.html#looping-techniques)

````python
order_items = [
    ("서울", "모니터", 2, 280000), ("부산", "키보드", 5, 42000),
    ("서울", "마우스", 10, 23000), ("대구", "모니터", 1, 280000),
    ("부산", "모니터", 3, 280000), ("서울", "키보드", 4, 42000),
    ("대구", "마우스", 6, 23000),
]
totals = {}
for items in order_items:
    totals[items[0]]= totals.get(items[0],0) + (items[2]* items[3])

print(totals)

totals_cnt = {}

for cnt in order_items:
  totals_cnt[cnt[1]] = totals_cnt.get(cnt[1],0) + cnt[2]
print(totals_cnt)
x= sorted(totals,key=lambda x:totals.get(x),reverse=True)
print(x[0],totals.get(x[0]))
````

## 문제 2. 측정 시각과 기준 초과 찾기

설비 진동값을 0시부터 3시간마다 하루 8번 측정했다. 진동 기준은 3.5mm/s다.

1. `range()`로 측정 시각 리스트 `hours`를 만들어 출력한다. (`list(range(...))`)
2. `zip()`으로 시각과 진동값을 묶어 `0시: 2.1mm/s` 형태로 출력한다.
3. 리스트 컴프리헨션과 `zip()`으로 진동값이 기준 이상인 시각만 모아 출력한다.
4. 기준 이상 비율(기준 이상 횟수 ÷ 전체 측정 횟수)을 백분율 소수 첫째 자리까지 출력한다.

```python
vibration = [2.1, 2.4, 3.8, 4.6, 4.1, 2.9, 2.2, 3.6]
limit = 3.5
```

**실행 결과**

```text
[0, 3, 6, 9, 12, 15, 18, 21]
0시: 2.1mm/s
3시: 2.4mm/s
6시: 3.8mm/s
9시: 4.6mm/s
12시: 4.1mm/s
15시: 2.9mm/s
18시: 2.2mm/s
21시: 3.6mm/s
기준 이상 시각: [6, 9, 12, 21]
기준 이상 비율: 50.0%
```

공식 문서: [range() 함수](https://docs.python.org/ko/3.13/tutorial/controlflow.html#the-range-function) · [zip()](https://docs.python.org/ko/3.13/library/functions.html#zip) · [리스트 컴프리헨션](https://docs.python.org/ko/3.13/tutorial/datastructures.html#list-comprehensions)

````python
vibration = [2.1, 2.4, 3.8, 4.6, 4.1, 2.9, 2.2, 3.6]
limit = 3.5
hours = list(range(0,24,3))
for i,j in zip(hours,vibration):
    print(i,j)
tmp2 = [hours for hours,vibration in zip(hours,vibration) if vibration>limit]
print(tmp2)


print(f"{len(tmp2)/len(vibration)*100:.1f}%")
````

## 문제 4. 라인별 생산 리포트

라인 이름, 계획 수량, 실적 수량이 같은 순서의 리스트로 주어진다.

1. `enumerate(zip(...), start=1)`로 순번과 세 값을 함께 꺼내 `1. A라인 계획 1,000 / 실적 1,040 / 달성률 104.0%` 형태로 출력한다. (`enumerate()`가 꺼낸 묶음은 반복문 안에서 다시 언패킹한다)
2. 같은 반복문에서 달성률이 100% 미만인 라인 이름을 리스트에 모아 출력한다.
3. 전체 달성률(실적 합계 ÷ 계획 합계)을 출력한다.
4. `lines`에 `"E라인"`을 추가한 뒤, `lines`의 길이와 `zip()`으로 묶인 묶음 수를 출력해 길이가 다를 때 어떻게 되는지 확인한다.

```python
lines = ["A라인", "B라인", "C라인", "D라인"]
plan = [1000, 1200, 900, 1100]
actual = [1040, 1150, 915, 980]
```

**실행 결과**

```text
1. A라인 계획 1,000 / 실적 1,040 / 달성률 104.0%
2. B라인 계획 1,200 / 실적 1,150 / 달성률 95.8%
3. C라인 계획 900 / 실적 915 / 달성률 101.7%
4. D라인 계획 1,100 / 실적 980 / 달성률 89.1%
목표 미달 라인: ['B라인', 'D라인']
전체 달성률: 97.3%
lines 길이: 5, zip 묶음 수: 4
```

공식 문서: [enumerate()](https://docs.python.org/ko/3.13/library/functions.html#enumerate) · [zip()](https://docs.python.org/ko/3.13/library/functions.html#zip)

````python
lines = ["A라인", "B라인", "C라인", "D라인"]
plan = [1000, 1200, 900, 1100]
actual = [1040, 1150, 915, 980]

merge = enumerate(zip(lines,plan,actual),start= 1)
for i,v in merge:
    print(f"{i},{v[0]},{v[1]},     {v[2]/v[1]*100:,.2f}")
# 여기에 작성한다.


lines.append("adasd")
print(len(lines))

merge = enumerate(zip(lines,plan,actual),start= 1)
for i,v in merge:
    print(f"{i},{v[0]},{v[1]},     {v[2]/v[1]*100:,.2f}")
# 여기에 작성한다.


print(merge)
````

## 문제 5. 전일 대비 증감과 누적 생산량

6일간의 생산량이다.

1. `range(1, len(daily))`로 둘째 날부터 반복하며 전일 대비 증감(오늘 - 어제)을 `2일차: 전일 대비 45개` 형태로 출력한다.
2. 같은 반복문에서 생산량이 늘어난 날의 수를 세고, 가장 크게 늘어난 날과 증가량을 찾는다.
3. 늘어난 날의 수와 가장 크게 늘어난 날을 출력한다.
4. 날마다 누적 생산량을 담은 리스트를 만들어 출력한다.

```python
daily = [820, 865, 790, 910, 945, 880]
```

**실행 결과**

```text
2일차: 전일 대비 45개
3일차: 전일 대비 -75개
4일차: 전일 대비 120개
5일차: 전일 대비 35개
6일차: 전일 대비 -65개

누적 생산량: [820, 1685, 2475, 3385, 4330, 5210]
```

공식 문서: [range() 함수](https://docs.python.org/ko/3.13/tutorial/controlflow.html#the-range-function) · [for 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#for-statements)

````python
daily = [820, 865, 790, 910, 945, 880]
daily_delta = []
daily2 =[daily[0]]
for i in range(1,len(daily)):
  daily_delta.append(daily[i]-daily[i-1])
  print(f"{i+1}일차: 전일 대비 {daily[i]-daily[i-1]}개")
  daily2.append(daily[i]+ daily2[i-1])
# 여기에 작성한다.
count=0
for i in range(len(daily_delta)):
  if(daily_delta[i]>0):
    count+=1
print(f"생산량이 늘어난 날 {count}일")

print(f"가장 크게 늘어난 날: {daily_delta.index(sorted(daily_delta,reverse= True)[0])+2}일차 ({sorted(daily_delta,reverse= True)[0]}개 증가)")
print(list(map(lambda x:x+sum(daily[:x]),range(len(daily)))))

print(daily2)
````
