### 문제 1-1. 주문 문자열을 나누어 총금액 계산

주문 정보가 하나의 문자열로 저장되어 있다. **문자열 메서드와 형변환**을 사용해 다음을 처리한다.

1. `|` 기준으로 나누고, 각 항목 앞뒤 공백을 제거한다.
2. 상품코드와 상품명을 출력한다.
3. 단가와 수량을 정수로 바꾼 후 총금액을 계산한다.
4. 총금액에 천 단위 쉼표를 붙인다.

**제공 데이터**

```python
record = "  BK-207 | 무선마우스 | 18000 | 3  "
```

**실행 결과**

```text
상품코드: BK-207
상품명: 무선마우스
총금액: 54,000원
```


````python
record = "  BK-207 | 무선마우스 | 18000 | 3  "

# 여기에 작성한다.
record = record.split('|')
record = list(map(lambda x: x.strip(),record))
print(f"상품코드: {record[0]}")
print(f"상품명: {record[1]}")
print(f"총금액: {int(record[2])*int(record[3]):,.0f}원")

````

### 문제 1-4. 과목 점수 문자열로 성적표 만들기

과목과 점수가 `;`와 `=`로 연결된 문자열을 처리한다.

1. 각 과목과 점수를 분리해 **딕셔너리**에 저장한다(점수는 정수로 변환한다).
2. 과목 수, 총점, 평균을 계산한다.
3. 가장 높은 점수를 받은 과목과 점수를 찾는다. (최고점은 중복되지 않는다.)
4. 평균은 소수 첫째 자리까지 출력한다.

**제공 데이터**

```python
report = "국어=82;영어=91;수학=75"
```

**실행 결과**

```text
과목 수: 3
총점: 248점
평균: 82.7점
최고점 과목: 영어 (91점)
```


````python
report = "국어=82;영어=91;수학=75"

# 여기에 작성한다.
report = report.replace('=',';')
print(report)
report = report.split(';')
grade = dict(zip([report[i] for i in range(len(report)) if i%2 ==0 ],[int(report[i]) for i in range(len(report)) if i%2 ==1]))
print(f"과목수 {len(grade)}")
print(f"총점 {sum(grade.values())}점")
print(f"평균 {sum(grade.values())/len(grade):.1f}점")
grade1 = sorted(grade.items(), key = lambda row: row[1] ,reverse=True)

print(f"최고점 과목: {grade1[0][0]} ({grade1[0][1]}점)")
````

### 문제 2-1. 매출과 환불 내역 분리하기

거래 내역은 양수(매출), 음수(환불), 0(거래 없음)으로 기록되어 있다.

1. 양수만 모아 `sales` 리스트에 저장한다.
2. 음수의 **절댓값**만 모아 `refunds` 리스트에 저장한다.
3. 두 리스트와 매출·환불 합계를 출력한다.
4. 0은 어느 리스트에도 넣지 않는다.

**제공 데이터**

```python
amounts = [12000, -5000, 33000, 0, -2000, 14000]
```

**실행 결과**

```text
매출 내역: [12000, 33000, 14000]
환불 내역: [5000, 2000]
매출 합계: 59,000원
환불 합계: 7,000원
```


````python
amounts = [12000, -5000, 33000, 0, -2000, 14000]

# 여기에 작성한다.
sales = [x for x in amounts if x>0]
refunds = [-x for x in amounts if x<0]
print(f"매출 내역: {sales}")
print(f"환불 내역: {refunds}")
print(f"매출 합계: {sum(sales):,}원")
print(f"환불 합계: {sum(refunds):,}원")

````

### 문제 2-2. 입고·출고 목록으로 상품명 정리

현재 상품 목록에 입고와 출고 정보를 반영한다.

1. `incoming`의 상품을 하나씩 확인한다. **이미 목록에 있으면 추가하지 않는다.**
2. `outgoing`의 상품을 하나씩 확인한다. **목록에 있는 상품만 삭제한다.**
3. 마지막 상품 목록을 가나다순으로 정렬해 출력한다.

원본 목록을 직접 수정하거나 복사해서 사용해도 된다.

**제공 데이터**

```python
items = ["키보드", "마우스", "모니터", "스피커"]
incoming = ["웹캠", "키보드"]
outgoing = ["마우스", "헤드셋"]
```

**실행 결과**

```text
['모니터', '스피커', '웹캠', '키보드']
```


````python
items = ["키보드", "마우스", "모니터", "스피커"]
incoming = ["웹캠", "키보드"]
outgoing = ["마우스", "헤드셋"]

# 여기에 작성한다.
for i in incoming:
    if i not in items:
        items.append(i)
for i in outgoing:
    if i  in items:
        items.remove(i)

print(sorted(items))
````

### 문제 2-3. 순서를 유지하며 중복 코드 제거

제품 코드가 입력된 **첫 번째 순서는 유지**하면서 중복만 제거한다.

1. 빈 리스트 `unique_codes`를 만든다.
2. `for` 문과 `in`을 이용하여 처음 나타난 코드만 넣는다.
3. 정리한 리스트와 제거된 코드 개수를 출력한다.

`set()`으로 곧바로 바꾸면 순서가 유지된다고 보장할 수 없으므로 이 문제에서는 사용하지 않는다.

**제공 데이터**

```python
codes = ["A01", "B02", "A01", "C03", "B02", "D04"]
```

**실행 결과**

```text
중복 제거: ['A01', 'B02', 'C03', 'D04']
제거된 개수: 2
```


````python
codes = ["A01", "B02", "A01", "C03", "B02", "D04"]

# 여기에 작성한다.
unique_codes = []
ss = []
for i in codes:
    if i not in unique_codes:
        unique_codes.append(i)
    else:
        ss.append(i)

print(f"중복 제거: {unique_codes}")
print(f"제거된 개수:{len(ss)}")
````

각 거래가 `(지점명, 매출)` 형태의 **튜플**로 저장되어 있다.

1. 거래를 순서대로 확인하고 가장 매출이 높은 **단일 거래**를 찾는다.
2. `강남점`의 거래 매출을 모두 더한다.
3. 결과를 천 단위 쉼표를 붙여 출력한다. 최고 매출 거래는 하나뿐이다.

**제공 데이터**

```python

                ("송파점", 150000), ("강남점", 95000)]
```

**실행 결과**

```text

```

### 문제 3-2. 등급별 개수 세기

`grades`에 나타난 등급을 딕셔너리로 집계한다.

1. `for` 문을 돌면서 등급별 건수를 센다.
2. `dict.get(등급, 0)`을 사용해 처음 나오는 등급도 처리한다.
3. 결과는 반드시 `A`, `B`, `C`, `D` 순서로 출력한다.

**제공 데이터**

```python
grades = ["A", "B", "A", "C", "B", "A", "D"]
```

**실행 결과**

```text
A: 3건
B: 2건
C: 1건
D: 1건
```


````python
grades = ["A", "B", "A", "C", "B", "A", "D"]

# 여기에 작성한다.
counts = {}
for i in grades:
    counts[i] = counts.get(i,0)+1

for a,b in counts.items():
    print(f"{a}: {b}건")
````

### 문제 3-3. 판매량을 반영한 재고 딕셔너리

상품별 현재 재고와 오늘 판매량이 딕셔너리로 주어진다.

1. 각 상품의 현재 재고에서 판매량을 뺀다.
2. 상품 이름과 남은 재고를 출력한다.
3. 원래 재고 딕셔너리에 있던 상품 순서대로 출력한다.

모든 상품은 판매 후 재고가 0 이상이라고 가정한다.

**제공 데이터**

```python
stock = {"연필": 20, "공책": 12, "지우개": 15}
sold = {"연필": 5, "공책": 3, "지우개": 2}
```

**실행 결과**

```text
연필: 15개
공책: 9개
지우개: 13개
```


````python
stock = {"연필": 20, "공책": 12, "지우개": 15}
sold = {"연필": 5, "공책": 3, "지우개": 2}

# 여기에 작성한다.
for i in sold:
    stock[i] = stock.get(i) - sold.get(i)

for a,b in stock.items():
    print(f"{a}: {b}개")
````

### 문제 3-5. 상태코드 변환과 완료 건수 계산

문자 상태코드를 사람이 읽기 쉬운 상태명으로 변환한다.

- `P` → `진행`, `D` → `완료`, `H` → `보류`
- 딕셔너리에 없는 코드는 `미정`으로 처리한다.
- 변환 결과를 `status_names` 리스트에 순서대로 저장한다.
- `완료` 상태의 건수를 출력한다.

`get()`과 리스트의 `append()`를 사용한다.

**제공 데이터**

```python

codes = ["P", "D", "X", "H", "D"]
```

**실행 결과**

```text

```


````python
status_map = {"P": "진행", "D": "완료", "H": "보류"}
codes = ["P", "D", "X", "H", "D"]

# 여기에 작성한다.
for i in range(len(codes)):
    if codes[i] not in status_map:
        codes[i] = '미정'
    else:
        codes[i] = status_map.get(codes[i])


print(codes)
print(f"완료 건수:{codes.count('완료')}건")
````

- **8시간 초과분**은 기존 시급의 **1.5배**로 계산한다.

- `hours=6`, `hours=10`으로 각각 호출한다.

- `"인원"`: 점수 개수
- `"평균"`: 점수 평균
- `"최고"`: 최고점
- `"최저"`: 최저점
- 빈 리스트면 `None`을 반환한다.

함수를 `scores`로 호출하여 예시와 같은 형식으로 출력하고, 빈 리스트로 호출한 결과도 출력한다.

**제공 데이터**

```python
scores = [95, 72, 88, 85]
```

**실행 결과**

```text
인원: 4명
평균: 85.0점
최고/최저: 95점 / 72점
빈 값: None
```


````python
scores = [95, 72, 88, 85]


def summarize(scores):
    count = len(scores)
    if count ==0:
        return None
    avg = sum(scores)/count
    high = max(scores)
    low = min(scores)
    return {
        "인원":count,
        "평균":avg,
        "최고":high,
        "최저":low
    }

s=summarize(scores)

print(f"인원: {s.get("인원")}명")
print(f"평균 {s.get("평균"):,.1f}점")
print(f"최고/최저점 {s.get("최고")}점/{s.get("최저")}점")
print(summarize(""))


# 여기에 작성한다.
````

3. 공백만 들어 있는 항목은 버린다.
4. **처음 나온 순서**를 유지하면서 중복을 제거한 리스트를 반환한다.

함수 호출 결과와 정리된 코드 개수를 출력한다.

**제공 데이터**

```python
codes = [" a01 ", "A01", " b02 ", "", "B02", " c03"]
```

**실행 결과**

```text
정리한 코드: ['A01', 'B02', 'C03']
코드 개수: 3개
```


````python
codes = [" a01 ", "A01", " b02 ", "", "B02", " c03"]

# 여기에 작성한다.
def normalize_codes(codes):
    ans = []
    for i in range(len(codes)):
        if len(codes[i])!=0:
            if codes[i].strip().upper() not in ans:
             ans.append(codes[i].strip().upper())
    return ans

print(f"정리한 코드: {normalize_codes(codes)}\n코드 개수: {len(normalize_codes(codes))}개")
````

### 문제 5-4. 문자열 두 개로 안전한 나눗셈

두 문자열을 받아 계산 결과를 문자열로 반환하는 함수 `safe_divide(a_text, b_text)`를 작성한다.

- 두 문자열을 `int()`로 변환하고 `a ÷ b`를 계산한다.
- 정상 계산은 결과를 **소수 둘째 자리까지** 문자열로 반환한다.
- 정수로 변환할 수 없으면 `숫자가 아니다`를 반환한다.
- 0으로 나누면 `0으로 나눌 수 없다`를 반환한다.
- `try`와 `except`를 사용한다.

주어진 `pairs`를 반복하면서 `a / b = 결과` 형식으로 출력한다.

**제공 데이터**

```python
pairs = [("120", "4"), ("50", "0"), ("abc", "2")]
```

**실행 결과**

```text
120 / 4 = 30.00
50 / 0 = 0으로 나눌 수 없다
abc / 2 = 숫자가 아니다
```


````python
pairs = [("120", "4"), ("50", "0"), ("abc", "2")]

# 여기에 작성한다.
def safe_divide(a_text,b_text):
    ans = 0
    try:
        ans=int(a_text)/int(b_text)
    except ValueError:
         return"숫자가 아니다"
    except ZeroDivisionError:
        return "0으로 나눌수 없다"

    return str(round(ans,2))

for i in pairs:
    print(f"{i[0]} / {i[1]} = {safe_divide(i[0],i[1])}")
````

### 문제 5-5. 쉼표로 입력한 점수를 함수로 분석

`input()`으로 점수들을 쉼표(`,`)로 구분해 한 줄로 입력받는다. 함수를 이용해 분석한다.

1. `analyze_scores(text)` 함수에서 입력 문자열을 쉼표로 나누고 각 항목을 `int()`로 변환한다. 항목 앞뒤에 공백이 있을 수도 있다.
2. **점수 개수, 평균, 80점 이상 개수**를 튜플로 반환한다.
3. 반환된 튜플을 받아 평균을 소수 둘째 자리로 출력한다.
4. 0~100 범위의 올바른 정수 점수가 적어도 하나 입력된다고 가정한다.

**입력 예시 1** (위에서부터 순서대로 입력)

```text
78, 90,85,62
```

**실행 결과 1**

```text
점수 개수: 4개
평균: 78.75점
80점 이상: 2명
```

**입력 예시 2** (위에서부터 순서대로 입력)

```text
100,100
```

**실행 결과 2**

```text
점수 개수: 2개
평균: 100.00점
80점 이상: 2명
```


````python
# 여기에 작성한다.
def analyze_scores(text):
    text = list(map(lambda x:int(x.strip()),text.split(',')))
    return text
numbers = input()
ans = analyze_scores(numbers)
print(f"점수 개수:{len(ans)}개\n평균 {sum(ans)/len(ans):,.2f}점 \n80점 이상: {len(list(filter(lambda x:x>=80,ans)))}명")
````
