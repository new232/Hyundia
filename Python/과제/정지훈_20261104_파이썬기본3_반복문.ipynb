## 1. 조건문

### 문제 1-1

구매금액에 따라 등급을 분류하고 `구매금액 180,000원 → 등급: 우수` 형태로 출력한다.

- 300000 이상: VIP
- 150000 이상: 우수
- 그 외: 일반

```python
purchase_amount = 180000
```

**실행 결과**

```text
구매금액 180,000원 → 등급: 우수
```

공식 문서: [if 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#if-statements)

````python
purchase_amount = 180000
grade= ""
# 여기에 작성한다.
if purchase_amount>= 300000:
    grade = "VIP"
elif purchase_amount>=150000:
    grade = "우수"
else:
     "일반"
print(f"구매금액:{purchase_amount}-> 등금:{grade}")
    
````

humidity = 67

### 문제 1-3

주문 금액과 회원 여부로 배송비와 할인 금액을 정한 뒤, 결제 금액(주문 금액 - 할인 + 배송비)까지 출력한다.

- 배송비: 주문 금액이 50000 이상이거나 회원이면 0원, 아니면 3000원
- 할인: 회원이면서 주문 금액이 30000 이상이면 주문 금액의 5% (`round()`로 정수), 아니면 0원

```python
amount = 48000
is_member = True
```

**실행 결과**

```text
배송비: 0원
할인: 2,400원
결제 금액: 45,600원
```

공식 문서: [if 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#if-statements) · [불 연산](https://docs.python.org/ko/3.13/library/stdtypes.html#boolean-operations-and-or-not)

````python
amount = 48000
is_member = True
develop = 0
discount = 0
if amount >= 50000 or is_member == True:
    develop = 0
else:
    develop = 3000;
if amount>= 30000:
    discount=0.05


print("배송비:",develop,"원")
print(f"할인{discount*amount:,.0f}원")
print(f"결제 금액: {amount*(1-discount)+develop:,.0f}원")
# 여기에 작성한다.
````

### 문제 1-4

오늘 매출 리스트가 비어 있으면 `오늘 매출 데이터가 없다.`를 출력하고, 값이 있으면 평균 매출을 출력하는 조건문을 작성한다. 비교 연산자 없이 리스트 자체의 참·거짓으로 판단한다.

빈 리스트로 한 번 실행한 뒤, `today_sales = [85000, 64000, 91000]`으로 바꿔 같은 조건문을 한 번 더 실행한다.

today_sales = []

**실행 결과**

```text
오늘 매출 데이터가 없다.
평균 매출: 80,000원
```

공식 문서: [진릿값 검사](https://docs.python.org/ko/3.13/library/stdtypes.html#truth-value-testing)

## 2. for 문

### 문제 2-1

아래 매출 리스트에서 150000 이상인 값의 개수와 합계를 반복문으로 계산해 출력한다.

```python
sales = [92000, 145000, 88000, 220000, 167000, 133000, 205000]
```

**실행 결과**

```text
150,000원 이상: 3건, 합계 592,000원
```

공식 문서: [for 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#for-statements) · [if 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#if-statements)

````python
sales = [92000, 145000, 88000, 220000, 167000, 133000, 205000]

# 여기에 작성한다.
count = 0
total = 0
for i in sales:
    if i >150000:
        count += 1
        total += i
print(f"150,000이상 {count}건, 합계 {total:,}원")
````

### 문제 2-2

부품 무게(g) 측정값이 쉼표로 이어진 문자열로 들어왔다.

1. `split(",")`으로 나눈 뒤, `for` 문으로 하나씩 `float()`로 바꿔 빈 리스트 `weights`에 `append()`한다.
2. `weights`를 출력한다.
3. 합계(소수 첫째 자리)와 평균(소수 둘째 자리)을 출력한다.

```python
weights_text = "12.4,11.8,12.9,12.1,13.0"
```

**실행 결과**

```text
[12.4, 11.8, 12.9, 12.1, 13.0]
합계: 62.2g, 평균: 12.44g
```

공식 문서: [for 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#for-statements) · [float()](https://docs.python.org/ko/3.13/library/functions.html#float)

````python
weights_text = "12.4,11.8,12.9,12.1,13.0"
weights_spilt = weights_text.split(',')
weights = list(map(float,weights_spilt))
print(f"합계: {sum(weights)}g, 평균: {sum(weights)/len(weights):.2f}g")
# 여기에 작성한다.
````

## 3. while 문

### 문제 3-1

목표 생산량은 1000개이고, 하루에 135개씩 생산한다. `while` 문으로 누적 생산량이 목표 이상이 될 때까지 하루씩 더해, 며칠째에 목표를 달성하는지 출력한다.

```python
target = 1000
daily_output = 135
```

**실행 결과**

```text
8일째 누적 1,080개로 목표 달성
```

공식 문서: [while 문](https://docs.python.org/ko/3.13/reference/compound_stmts.html#the-while-statement)

````python
target = 1000
daily_output = 135

# 여기에 작성한다.
days = 1
while daily_output < target:
    days += 1
    daily_output = daily_output+ 135

print(f"{days}누적 {daily_output:,}개로 목표 달성")
````

## 4. break와 continue

### 문제 4-1

설비 온도 측정값에는 측정되지 않은 값(`None`)이 섞여 있다.

1. `None`은 `continue`로 건너뛴다.
2. 90 이상인 값이 나오면 `92.8도 감지, 측정 중단`을 출력하고 `break`로 반복을 끝낸다.
3. 그 전까지의 정상 측정값으로 건수와 평균(소수 둘째 자리)을 출력한다.

```python
readings = [71.2, None, 73.5, 74.1, None, 92.8, 72.0]
```

**실행 결과**

```text
92.8도 감지, 측정 중단
정상 측정 3건, 평균 72.93도
```

공식 문서: [break 문과 continue 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#break-and-continue-statements)

````python
readings = [71.2, None, 73.5, 74.1, None, 92.8, 72.0]

# 여기에 작성한다.
count = 0
total = 0
for i in readings:
    if i is None:
        continue
    elif i>90:
        print(f"{i}도 감지, 측정 중단")
        break
    else:
        count += 1
        total += i
print(f"정상 측정 {count}건, 평균{total/count:.2f}도")
````
