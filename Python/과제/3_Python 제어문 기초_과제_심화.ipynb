## 문제 1. 센서 로그 정리하기

온도 센서 로그가 문자열 리스트로 들어왔다. 빈 문자열은 값이 기록되지 않은 것이고, `"err"`는 센서 오류다.

1. 빈 문자열은 진릿값으로 확인해 빈값 건수를 세고 건너뛴다.
2. `"err"`는 오류 건수를 세고 건너뛴다.
3. 나머지는 `float()`로 바꾼다. 80 이상이면 `경고: 81.2도` 형태로 출력하고 경고 건수를 센다. 80 미만이면 정상 리스트 `normal`에 담는다.
4. 빈값·오류·경고 건수와 정상 리스트를 출력한다.
5. 정상 리스트에 값이 있을 때만 평균(소수 첫째 자리)을 출력하고, 없으면 `정상 값이 없다.`를 출력한다.

```python
log = ["78.4", "", "81.2", "err", "76.9", "85.5", "", "79.3"]
```

**실행 결과**

```text
경고: 81.2도
경고: 85.5도
빈값 2건, 오류 1건, 경고 2건
정상 값: [78.4, 76.9, 79.3]
정상 평균: 78.2도
```

공식 문서: [진릿값 검사](https://docs.python.org/ko/3.13/library/stdtypes.html#truth-value-testing) · [break 문과 continue 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#break-and-continue-statements) · [float()](https://docs.python.org/ko/3.13/library/functions.html#float)

````python
log = ["78.4", "", "81.2", "err", "76.9", "85.5", "", "79.3"]
emp_cnt=0
err_cnt=0
dan_cnt =0
normal= []
for i in log:
    if len(i) ==0:
        emp_cnt+=1
        continue
    elif i == 'err':
        err_cnt += 1
        continue
    elif float(i) > 80:
        print(f"비상{float(i)}도")
        dan_cnt += 1
    else:
         normal.append(float(i))

print(f"빈값 {emp_cnt}간, 오류 {err_cnt}건, 위험 {dan_cnt}건")
print(f"정상 값: {normal}")
print(f"정상 평균: {sum(normal)/len(normal):.1f}도")
# 여기에 작성한다.
````

## 문제 5. 종합 문제: 일별 매출 분석

일주일 매출과 하루 목표 매출이 주어진다. 최댓값 함수는 아직 배우지 않았으므로 반복문으로 직접 찾는다. 아래 계산은 리스트에 값이 있을 때만 하고, 비어 있으면 `매출 데이터가 없다.`를 출력한다.

1. 목표 이상인 날의 매출만 리스트에 모아 달성 일수와 함께 출력한다.
2. 첫 값을 최고 매출로 두고 반복문으로 더 큰 값을 찾아, 최고 매출과 그날이 몇 번째 날인지 출력한다.
3. 평균 매출로 등급을 정해 `평균 매출: 409,000원 → 보통` 형태로 출력한다. (430000 이상 우수, 380000 이상 보통, 그 외 개선 필요)

```python
daily_sales = [412000, 389000, 455000, 301000, 478000, 396000, 432000]
target = 400000
```

**실행 결과**

```text
목표 달성: 4일 [412000, 455000, 478000, 432000]
최고 매출: 5일차 478,000원
평균 매출: 409,000원 → 보통
```

공식 문서: [for 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#for-statements) · [if 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#if-statements) · [진릿값 검사](https://docs.python.org/ko/3.13/library/stdtypes.html#truth-value-testing)

````python
daily_sales = [412000, 389000, 455000, 301000, 478000, 396000, 432000]
target = 400000

count  = 0

count = sum(map(lambda x: x > target, daily_sales))
lis = list(filter(lambda x:x>target,daily_sales))
print(f"{count},{lis}")
max = sorted(daily_sales, reverse=True)[0]
print(daily_sales.index(max)+1,max)
avg = sum(daily_sales) / len(daily_sales)

result = "우수" if avg > 430000 else "보통" if avg > 380000 else "개선필요"

print(result)# 여기에 작성한다.
````
