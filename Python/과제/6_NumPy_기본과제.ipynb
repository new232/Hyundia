## 1. 1차원 배열과 기본 문법

### 문제 1-1. 평균 이상 조회

점수 배열에서 **평균**을 출력하고, **평균 이상인 점수만** 출력하세요.

사용 문법: `mean()`, 불리언 인덱싱.

```python
scores = np.array([85, 72, 91, 68, 88])

# 여기에 코드를 작성하세요.
print(scores[scores>scores.mean()])
```

### 문제 1-2. 합계와 최솟값·최댓값

판매 수량의 **합계, 평균, 최솟값, 최댓값**을 차례대로 출력하세요.

사용 문법: `sum()`, `mean()`, `min()`, `max()`.

```python
sales = np.array([10, 20, 30, 40, 50])

# 여기에 코드를 작성하세요.
print(f"합계: {sales.sum()}")
print(f"평균: {sales.mean()}")
print(f"최솟값: {sales.min()}")
print(f"최댓값: {sales.max()}")
```

### 문제 1-3. 결측값 확인

센서 측정값에서 **결측값 개수**와 **결측값을 제외한 평균**을 출력하세요. `np.nan` 값은 측정이 누락된 값입니다.

사용 문법: `np.isnan()`, `sum()`, `np.nanmean()`.

```python
readings = np.array([70.0, np.nan, 80.0, 90.0])

print(sum(np.isnan(readings)))
print(np.nansum(readings))
print(np.nanmean(readings))
# 여기에 코드를 작성하세요.
```

### 문제 1-4. 자료형 변환

숫자가 문자로 저장된 배열을 `astype(int)`로 정수 배열로 바꾸고 **변환된 배열과 합계**를 출력하세요.

```python
qty_text = np.array(["10", "20", "30"])

print(qty_text.astype(int),sum(qty_text.astype(int)))
# 여기에 코드를 작성하세요.
```

### 문제 1-5. 인덱싱과 슬라이싱

월요일부터 일요일까지의 판매 수량입니다. **첫날 값**, **마지막 날 값**, **평일(월~금) 값**을 출력하세요.

사용 문법: `[0]`, `[-1]`, 슬라이싱.

```python
weekly = np.array([12, 14, 13, 15, 11, 8, 6])
print(weekly[0], weekly[-1],weekly[:5])
# 여기에 코드를 작성하세요.
```

### 문제 1-6. 배열끼리 계산

설비별 **양품 수량(투입 수량 - 불량 수량)**과 **불량률(불량 수량 / 투입 수량 × 100)**을 출력하세요.

반복문 없이 배열끼리 계산합니다.

**출력 안내:** 불량률은 별도의 반올림 없이 계산한 값을 그대로 출력한다.


```python
input_qty = np.array([100, 120, 90])
defects = np.array([2, 6, 3])
print(f"양품 수량{input_qty- defects}")
print(f"불량률{defects/input_qty*100}")
# 여기에 코드를 작성하세요.
```

### 문제 1-7. 조건에 따른 분류

점수가 **80점 이상이면 `"우수"`**, 그 외에는 `"일반"`인 배열을 출력하세요.

사용 문법: `np.where()`.

```python
scores = np.array([85, 72, 91, 68])
A =np.where(scores >= 80, "우수","일반")
print(A)
# 여기에 코드를 작성하세요.
```

### 문제 1-8. 정렬

판매 수량을 `np.sort()`로 **오름차순 정렬**하여 출력하고, **원본 배열**도 출력하세요.

```python
sales = np.array([30, 10, 40, 20])
print(f"원본{sales}")
print(np.sort(sales))
# 여기에 코드를 작성하세요.
```

## 2. 2차원 배열 기초

### 문제 2-1. 2차원 배열에서 행과 열 조회

다음 배열은 **행: 측정 회차(3회), 열: 설비(2대)**입니다. `shape`, **첫 번째 행**, **두 번째 열**을 출력하세요.

```python
production = np.array([[100, 120], [110, 130], [105, 125]])
print(production.shape,production[0],production[:,1])
# 여기에 코드를 작성하세요.
```

### 문제 2-2. axis로 합계와 평균 구하기

아래 코드 셀에 제공된 `production` 배열에서 **설비별 합계(`axis=0`)**와 **측정 회차별 평균(`axis=1`)**을 출력하세요.

```python
production = np.array([[100, 120], [110, 130], [105, 125]])
print(production.sum(axis = 0))
print(production.mean(axis= 1))
# 여기에 코드를 작성하세요.
```
