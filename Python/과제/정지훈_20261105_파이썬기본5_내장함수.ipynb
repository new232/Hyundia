### 문제 2-3

매출 리스트를 받아 개수, 합계, 평균, 최솟값, 최댓값을 딕셔너리로 반환하는 함수 `summarize_sales`를 작성한다. 빈 리스트가 들어오면 `None`을 반환한다.

작성 후 아래 리스트로 호출해 `items()`로 한 줄씩 출력한다. 평균처럼 실수인 값은 소수 둘째 자리, 정수는 쉼표만 붙인다(`type(value) == float`로 구분). 마지막으로 빈 리스트로 호출한 결과도 출력한다.

```python
sales = [92000, 145000, 88000, 220000, 167000, 133000, 205000]
```

**실행 결과**

```text
개수: 7
합계: 1,050,000
평균: 150,000.00
최솟값: 88,000
최댓값: 220,000
None
```

공식 문서: [함수 정의하기](https://docs.python.org/ko/3.13/tutorial/controlflow.html#defining-functions) · [내장 함수](https://docs.python.org/ko/3.13/library/functions.html)

````python
sales = [92000, 145000, 88000, 220000, 167000, 133000, 205000]

# 여기에 작성한다.d
def summarice_sales(d):
    if len(d)==0:
        return None
    count = len(d)
    total = sum(d)
    avg = total/count
    min_d = min(d)
    max_d = max(d)
    return {
        "count":count,
        "total":total,
        "avg":avg,
        "min": min_d,
        "max": max_d
    }

s = summarice_sales(sales)
for key, value in s.items():
    if type(value) == float:  # 값의 자료형을 확인
        print(f"{key}: {value:,.2f}")
    elif type(value) == int:
        print(f"{key}: {value:,}")
    else:
        print(f"{key}: {value}")
print(summarice_sales([]))
````

### 문제 2-4

매출 리스트를 받아 최솟값, 최댓값, 두 값의 차이를 한 번에 반환하는 함수 `sales_range(values)`를 작성한다.

1. `[412000, 389000, 455000, 301000]`로 호출한 결과를 변수 하나(`result`)에 받아 값과 자료형을 출력한다.
2. 같은 호출 결과를 `low, high, gap`으로 언패킹해 `최저 301,000원, 최고 455,000원, 차이 154,000원` 형태로 출력한다.

**실행 결과**

```text
(301000, 455000, 154000)
<class 'tuple'>
최저 301,000원, 최고 455,000원, 차이 154,000원
```

공식 문서: [함수 정의하기](https://docs.python.org/ko/3.13/tutorial/controlflow.html#defining-functions) · [튜플과 시퀀스](https://docs.python.org/ko/3.13/tutorial/datastructures.html#tuples-and-sequences)

````python
# 여기에 작성한다.
def sales_range(values):
    high = max(values)
    low = min(values)
    gap = high-low
    return high,low,gap

asd = [412000, 389000, 455000, 301000]
result= sales_range(asd)
print(result)
print(type(result))
low, high, gap =result
print(f"최저{low:,}원 , 최고{high:,}원 , 차이 {gap:,}원")
````

점수를 받아 **합격/불합격**을 판정하는 함수를 작성한다.

- `pass_result(score)` 함수를 정의한다.
- 점수가 **60점 이상이면 `"합격"`**, 60점 미만이면 `"불합격"`을 반환한다.
- 아래 리스트의 점수를 순서대로 함수에 전달해 출력한다.

```python
scores = [58, 60, 82]
```

**실행 결과**

```text
58점: 불합격
60점: 합격
82점: 합격
```

공식 문서: [함수 정의하기](https://docs.python.org/ko/3.13/tutorial/controlflow.html#defining-functions)

### 문제 2-9

점수 리스트에서 기준 점수 **이상**인 학생 수를 계산하는 함수를 작성한다.

- `count_at_least(scores, cutoff)` 함수를 정의한다.
- 리스트에서 `cutoff` 이상인 값의 개수를 세어 **정수로 반환**한다.
- 아래 점수 리스트로 함수를 두 번 호출해 **80점 이상**과 **90점 이상** 학생 수를 각각 출력한다.

```python
scores = [65, 90, 80, 45, 95]
```

**실행 결과**

```text
80점 이상: 3명
90점 이상: 2명
```

힌트: 함수 안에서 `for`문으로 점수를 확인하고 조건을 만족할 때마다 개수를 1씩 증가시킬 수 있다.

공식 문서: [함수 정의하기](https://docs.python.org/ko/3.13/tutorial/controlflow.html#defining-functions)


````python
scores = [65, 90, 80, 45, 95]

# 여기에 작성한다.
def count_at_least(scores, cutoff):
    count = len(list(filter(lambda x:x>=cutoff,scores)))
    return count

print(f"90점 이상:{count_at_least(scores,90)}명")
print(f"80점 이상:{count_at_least(scores,80)}명")
````

````python
# 여기에 작성한다.
def shipping_fee(amount, remote):
    fee = amount
    dev = 0
    if amount >=50000:
        fee+=0
        dev = 0
    else:
        fee+= 3000
        dev = 3000
    if remote  == 'Y':
        fee += 2000
        dev+=2000
    return fee,dev

a = int(input())
b = input()
fee,dev = shipping_fee(a,b)
print(f"배달비: {dev}원")
print(f"총 결제 금액:{fee}원")
````

## 3. 예외 처리

### 문제 3-1

총 예산 1,000,000원을 수량으로 나눠 1개당 예산을 구한다. 수량은 문자열로 들어오고, 잘못된 값이 섞여 있다. `for` 문으로 하나씩 처리하면서 `try`·`except`·`else`를 사용한다.

- `int()`로 바꿀 수 없으면(`ValueError`): `1,250: 숫자로 변환할 수 없다`
- 수량이 0이면(`ZeroDivisionError`): `0: 0으로 나눌 수 없다`
- 오류가 없으면: `3200: 1개당 예산 312.50원`

```python
budget = 1000000
inputs = ["3200", "1,250", "0"]
```

**실행 결과**

```text
3200: 1개당 예산 312.50원
1,250: 숫자로 변환할 수 없다
0: 0으로 나눌 수 없다
```

공식 문서: [예외 처리하기](https://docs.python.org/ko/3.13/tutorial/errors.html#handling-exceptions)

````python
budget = 1000000
inputs = ["3200", "1,250", "0"]

# 여기에 작성한다.

for i in inputs:
    try:
        a=budget/int(i)
    except ValueError  :
        print(f"{i}: 숫자로 변환 할수 없다")
    except ZeroDivisionError:
        print(f"{i}: 0으로 나눌 수 없다")
    else:
        print(f"{i}: 1개당 예산 {a:.2f}원")



````

다음 문자열 리스트를 **상품 가격(원)**으로 변환해 출력한다. 잘못된 값은 분류해 안내한다.

- 아래 `prices`의 값을 `for`문으로 하나씩 처리한다.
- `int()`로 바꿀 수 없으면 `잘못된 가격: 값`으로 출력한다.
- 정수로 변환했더라도 음수이면 `음수 금액: 값`으로 출력한다.
- 0 이상인 정수는 `가격: 12,000원` 형태로 출력한다.

```python
prices = ["12000", "0", "-500", "열다섯"]
```

**실행 결과**


### 문제 3-5

사용자가 입력한 작업 시간(분)을 **시간과 분으로 분리해** 출력한다.

- `input()`으로 작업 시간(분)을 한 번 입력받아 `int()`로 변환한다.
- 숫자가 아닌 입력은 `정수 분을 입력한다`고 출력한다.
- 음수가 입력되면 `0 이상의 시간을 입력한다`고 출력한다.
- 정상적인 값이면 **시간은 `분 // 60`, 남은 분은 `분 % 60`**으로 계산해 출력한다.
- 각 입력 예시는 **한 번씩 따로 실행한 경우**이다.

**입력 예시 ①** `125` → **실행 결과**

```text
작업 시간: 2시간 5분
```

**입력 예시 ②** `-3` → **실행 결과**
a
```text
0 이상의 시간을 입력한다
```

**입력 예시 ③** `abc` → **실행 결과**

```text
정수 분을 입력한다
```

공식 문서: [예외 처리하기](https://docs.python.org/ko/3.13/tutorial/errors.html#handling-exceptions)


````python
try:
    minute = int(input("작업 시간(분): "))

    if minute < 0:
        print("0 이상의 시간을 입력한다")
    else:
        hour = minute // 60
        remain = minute % 60

        print(f"작업 시간: {hour}시간 {remain}분")

except ValueError:
    print("정수 분을 입력한다")
````
