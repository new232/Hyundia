# **6. 함수 (Function)**

같은 계산을 여러 번 해야 할 때마다 코드를 다시 작성하면 코드가 길어지고 실수가 발생하기 쉽다.

함수(function)는 특정 작업을 수행하는 코드를 하나의 이름으로 정의한 것이다. 필요한 값을 함수에 전달하여 호출할 수 있고, 처리 결과를 `return`으로 반환할 수 있다.

함수를 사용하면 다음과 같은 장점이 있다.

- 반복되는 코드를 줄일 수 있다
- 같은 계산을 여러 데이터에 재사용할 수 있다
- 코드를 기능별로 나누어 관리할 수 있다
- 코드의 의미와 처리 과정을 이해하기 쉬워진다

## **자주 사용하는 파이썬 내장 함수 (Built-in Functions)**

파이썬은 자주 사용하는 기능을 **내장 함수(Built-in Function)**로 제공한다. 내장 함수는 직접 `def`로 정의하거나 별도의 모듈을 불러오지 않고 호출할 수 있다.

앞에서 사용한 `print()`, `type()`, `round()`도 내장 함수이다. 숫자로 이루어진 리스트를 다룰 때는 다음 함수가 유용하다.

| 함수 | 기능 | 예시 |
|---|---|---|
| `len()` | 요소 개수 | `len(sales)` |
| `sum()` | 숫자 요소의 합계 | `sum(sales)` |
| `min()` | 최솟값 | `min(sales)` |
| `max()` | 최댓값 | `max(sales)` |
| `round()` | 지정한 자리로 반올림 | `round(12.345, 2)` |
| `type()` | 객체의 자료형 확인 | `type(sales)` |
| `print()` | 결과 출력 | `print(sales)` |

평균은 위 내장 함수를 조합하여 `sum(sales) / len(sales)`로 계산한다. 리스트가 비어 있으면 `len(sales)`가 0이 되므로 평균 계산 시 주의한다. 빈 리스트 처리 방법은 뒤에서 학습한다.


공식 문서: [Python 3.13 — 내장 함수](https://docs.python.org/ko/3.13/library/functions.html)


````python
# [예제] 내장 함수로 데이터 계산하기

sales = [100000, 150000, 200000]

# 데이터 개수
count = len(sales)
print(count)  # 3

# 합계
total = sum(sales)
print(total)  # 450000

# 평균
average = total / count
print(average)  # 150000.0

# 최솟값과 최댓값
minimum = min(sales)
maximum = max(sales)

print(minimum)  # 100000
print(maximum)  # 200000
````

```text
[구조]
 def 함수이름(매개변수):
    실행할 코드
    return 돌려줄_값

 def calculate_total(values):    ← 함수 정의. values는 전달받을 입력값
     return sum(values)          ← 호출한 곳으로 결과를 가지고 돌아가기

```
```python
sales = [100, 200, 300]

result = calculate_total(sales)

print(result)  # 600
```


* def 함수이름(매개변수):로 새로운 함수를 만든다.
* return은 함수의 실행을 끝내고 호출한 곳으로 결과값을 가지고 돌아간다.
* 만든 함수는 함수이름(인자) 형태로 호출한다.
```text
calculate_total(sales) 호출
        ↓
sales가 values에 전달됨
        ↓
sum(values) 계산
        ↓
return으로 계산 결과 반환
        ↓
반환된 값이 result에 대입됨
```

공식 문서: [함수 정의하기](https://docs.python.org/ko/3.13/tutorial/controlflow.html#defining-functions)

````python
def calculate_total(values):
    return sum(values)


sales = [100, 200, 300]

result = calculate_total(sales)

print(result)  # 600
````

## **3) 기본값 매개변수와 키워드 인자**

### **기본값 매개변수**

매개변수에 기본값을 지정하면 함수를 호출할 때 해당 인자를 생략할 수 있다.

```python
def count_target_sales(values, standard=100000):
    count = 0

    for value in values:
        if value >= standard:
            count += 1

    return count
```

기준 금액을 생략하면 기본값 `100000`을 사용한다.

```python
sales = [85000, 120000, 99000, 210000]

result = count_target_sales(sales)

print(result)  # 2
```

다른 기준을 사용하려면 두 번째 인자를 전달한다.

```python
result = count_target_sales(sales, 150000)

print(result)  # 1
```

### **키워드 인자**

함수를 호출할 때 매개변수 이름을 지정하여 값을 전달할 수 있다. 이를 **키워드 인자(Keyword Argument)**라고 한다.

키워드 인자는 값을 전달할 매개변수를 이름으로 지정하므로 순서를 바꾸어도 된다.

```python
result = count_target_sales( values=sales,  standard=150000)

print(result)  # 1
```


> 공식 문서: [기본 인자 값](https://docs.python.org/ko/3.13/tutorial/controlflow.html#default-argument-values) · [키워드 인자](https://docs.python.org/ko/3.13/tutorial/controlflow.html#keyword-arguments)




````python
# [예제] 기준 금액 이상인 매출이 몇 건인지 세는 함수
# values   : 매출 리스트
# standard : 기준 금액. 호출할 때 생략하면 기본값 100000을 쓴다
def count_target_sales(values, standard=100000):
    count = 0

    for value in values:
        if value >= standard:      # 기준 이상이면
            count += 1             # 1건 추가

    return count


sales = [85000, 120000, 99000, 210000]

print(count_target_sales(sales))                     # 2  (기본값 100000 사용)
print(count_target_sales(sales, 150000))             # 1  (두 번째 인자로 기준 변경)
print(count_target_sales(values=sales, standard=150000))   # 1  (키워드 인자)
print(count_target_sales(standard=150000, values=sales))
````

## **4) `print()`와 `return`의 차이**

`print()`와 `return`은 역할이 다르다.

- `print()`: 값을 화면에 출력한다.값이 호출한 곳으로 돌아가지 않으므로 그 결과를 다른 계산에 재사용할 수 없다.
- `return`: 함수의 처리 결과를 호출한 곳으로 반환한다. 전달받은 값은 변수에 저장하여 이후 다른 계산이나 코드에서 다시 사용할 수 있다.

```python
# 화면에 출력하지만 결과를 반환하지 않는 함수
def print_total(values):
    print(sum(values))


# 계산 결과를 반환하는 함수
def return_total(values):
    return sum(values)
```

### **함수의 4가지 형태**

함수는 입력(인자)이 있는지, 결과를 `return`으로 돌려주는지에 따라 네 가지 모양으로 나눌 수 있다.

| 형태 | 입력(인자) | 반환(`return`) | 하는 일 |
|---|---|---|---|
| 1) | 없음 | 없음 | 정해진 안내 문구를 화면에 보여 준다 |
| 2) | 있음 | 없음 | 받은 값을 계산해 화면에만 보여 준다 |
| 3) | 없음 | 있음 | 세율처럼 정해진 값을 돌려준다 |
| 4) | 있음 | 있음 | 받은 값을 계산해 돌려준다 |



### **`return`이 없는 함수는 `None`을 돌려준다**

```python
def show_sum(a, b):
    print(a + b)            # 화면에만 보여 준다

result = show_sum(3, 4)     # 화면에 7 이 보인다
print(result)               # None
```

공식 문서: [함수 정의하기](https://docs.python.org/ko/3.13/tutorial/controlflow.html#defining-functions)


````python
# [예제] 함수의 4가지 형태
# 1) 입력 없음 · 반환 없음: 정해진 일만 한다
def greet():
    print("작업 시작")

greet()

# 2) 입력 있음 · 반환 없음: 받은 값을 계산해 화면에만 보여 준다
def show_sum(a, b):
    print(a + b)

show_sum(3, 4)

# 3) 입력 없음 · 반환 있음: 정해진 값을 돌려준다
def get_tax_rate():
    return 0.1

print(20000 * get_tax_rate())

# 4) 입력 있음 · 반환 있음: 받은 값을 계산해 돌려준다
def add(a, b):
    return a + b

result = add(3, 4)
print(result * 10)

# return 이 없는 함수의 결과는 None 이다
result = show_sum(3, 4)          # 화면에 7 이 보인다
print(result)                    # None
````

## **5) 빈 리스트 처리**

값이 하나도 없는 빈 리스트(`[]`)의 평균을 구하려고 하면, 요소의 개수(`len`)가 `0`이 되어 0으로 나누기 오류(`ZeroDivisionError`)가 발생한다.

함수 내부에서 **계산을 수행하기 전에 먼저 "리스트가 비어 있는지" 확인**하고, 비어 있다면 `None`을 즉시 반환하도록 작성하면 프로그램이 갑자기 중단되는 것을 막을 수 있다.

---

###  **예외 방어 흐름**

```text
 def calculate_average(values):
     │
     ├─ 1. len(values) == 0 인가? ──➔ 참(True)이면 ──➔ return None (즉시 종료)
     │
     └─ 2. 거짓(False)이면 ─────────➔ sum(values) / len(values) 계산 후 반환
```

````python
# [실습 1] 빈 리스트 예외 처리가 적용된 평균 계산 함수
def calculate_average(values):
    # 1. 리스트가 비어 있는지 먼저 검사 (방어적 코드)
    if len(values) == 0:   #if not values:라고 적어도 동일
        return None  # 빈 리스트면 None을 돌려주고 함수 종료

    # 2. 요소가 있을 때만 평균 계산 실행
    return sum(values) / len(values)


# 정상 데이터 테스트
result1 = calculate_average([100, 200, 300])
print("정상 리스트 평균:", result1)  # 출력: 200.0

# 빈 리스트 데이터 테스트 (오류 없이 None 반환)
result2 = calculate_average([])
print("빈 리스트 평균:", result2)    # 출력: None

#반환된 `None` 처리하기
#함수가 `None`을 반환할 수 있다면 호출한 쪽에서 결과를 확인한 후 사용한다.


if result2 is None:
    print("평균을 계산할 데이터가 없습니다.")
else:
    print("평균:", result2)
````

````python
# [실습] 빈 리스트 처리
# 문제: 숫자 리스트에서 최댓값을 구하는 함수 max_value(values)를 작성하시오.
#       리스트에 데이터가 있으면 최댓값을 반환하고,
#       빈 리스트이면 None을 반환하도록 작성하시오.

#       numbers = [10, 30, 20]으로 함수를 호출하고,
#       빈 리스트 []로도 호출하여 결과를 출력하시오.

# [실행 결과]
def max_value(values):
    if len(values) == 0:
        return None
    else:
        return max(values)
# 30

numbers = [10, 30, 20]
print(max_value(numbers))

empty = []
print(max_value(empty))

# None
````

````python
# [실습] 빈 리스트 처리
# 문제: 설비별 불량 기록 중 비어 있는 설비가 있어도 오류 없이, 설비마다 최대 불량 수를 출력한다.

# 1. 설비별 불량 수 기록
defects = {"M01": [3, 1, 4], "M02": [], "M03": [2, 2]}

# 2. 불량 수 리스트를 받아 최댓값을 반환하는 함수 max_defects(values) 를 만든다.
def max_defects(values):
    if(len(values)==0):
        return None
    else:
        return max(values)


# 3. 딕셔너리를 반복하며 설비마다 max_defects 를 호출한다.
#    결과가 None이면 "M02: 가동 기록 없음", 아니면 "M01: 최대 불량 4개" 형식으로 출력한다.


for i in defects.keys():
    get_max = max_defects(defects.get(i))
    if get_max is None:
        print(f"{i}: 가동 기록 없음")
    else:
        print(f"{i} {get_max}")
# 출력: M01: 최대 불량 4개
# 출력: M02: 가동 기록 없음
# 출력: M03: 최대 불량 2개


````

## **6) 여러 결과를 한 번에 반환하기 (다중 반환과 튜플 언패킹)**

파이썬의 함수는 단 하나의 값뿐만 아니라 **여러 개의 결과값을 한 번에 반환**할 수 있다.

`return total, average, minimum, maximum`처럼 쉼표(`,`)로 구분하여 반환하면 파이썬은 이 값들을 **하나의 튜플(Tuple)로 묶어서 전달**한다. 이를 받는 쪽에서는 변수 여러 개를 나열하여 각 값을 나누어 담을 수 있다 (**튜플 언패킹**).

---

### **다중 반환 및 언패킹 구조**

```text
 return total, average, minimum, maximum  ➔ (합계, 평균, 최솟값, 최댓값) 튜플 생성

 tot, avg, min_v, max_v = summarize_sales(sales)  ➔ 각 변수에 순서대로 언패킹
```

````python
# [실습 1] 다중 반환 함수 정의 및 호출
def summarize_sales(values):
    # 빈 리스트 방어 예외 처리
    if not values:
        return None

    total = sum(values)
    average = total / len(values)
    minimum = min(values)
    maximum = max(values)

    # 4개의 값을 튜플 형태로 한 번에 반환
    return total, average, minimum, maximum


sales = [120000, 150000, 90000, 200000]
print(sales)
````

````python
# 1. 하나의 변수로 받기 (튜플 통째로 저장됨)
result = summarize_sales(sales)
print("통째로 받은 결과 (튜플):", result)
# 출력: (560000, 140000.0, 90000, 200000)

print("-" * 50)

# 2. 여러 변수로 나누어 받기 (튜플 언패킹 활용)
total_sales, avg_sales, min_sales, max_sales = summarize_sales(sales)

print(f"총 매출액: {total_sales:,}원")
print(f"평균 매출액: {avg_sales:,.1f}원")
print(f"최저 매출액: {min_sales:,}원")
print(f"최고 매출액: {max_sales:,}원")
````

````python
#빈 데이터의 반환 결과 확인하기
#빈 리스트를 전달하면 `summarize_sales()`는 `None`을 반환한다. 이때는 바로 언패킹하지 말고 결과를 먼저 확인해야 한다.

result = summarize_sales([])

if result is None:
    print("요약할 데이터가 없습니다.")
else:
    total, average, minimum, maximum = result
    print(total, average, minimum, maximum)


#다음과 같이 바로 언패킹하면 `None`을 나눠 받을 수 없어 `TypeError`가 발생한다.
# total, average, minimum, maximum = summarize_sales([])

````

````python
# [실습] 다중 반환과 튜플 언패킹
# 문제: 최저·최고 온도를 함께 반환하는 temperature_range(values) 를 만들어 [72.1, 75.4, 68.9, 79.8] 로 호출한 결과를 변수 하나(result)로 받아 튜플로 출력하고,
#  low, high 로 언패킹해 "최저 68.9도, 최고 79.8도" 로 출력한다.

def temperature_range(values):
    high = max(values)
    low = min(values)
    return high , low

# 1. 온도 리스트를 받아 최저 온도와 최고 온도를 함께 반환하는 함수 temperature_range(values) 를 만든다.


# 2. temperatures = [72.1, 75.4, 68.9, 79.8] 로 호출한 결과를 변수 하나 result 에 담아 출력한다.
# 출력: (68.9, 79.8)

temperatures = [72.1, 75.4, 68.9, 79.8] 
result = temperature_range(temperatures)
print(result[0],result[1])
# 3. 같은 호출 결과를 low, high 두 변수로 언패킹해 아래 형식으로 출력한다.
# 출력: 최저 68.9도, 최고 79.8도

````

## **7) 지역변수 (Local Variable)와 변수의 유효 범위**

함수 내부에서 만든 변수를 지역변수(Local Variable)라고 부른다.

함수 안에서 만든 변수는 함수 안에서만 사용할 수 있고, 함수 밖에서 지역변수의 이름을 직접 불러 사용할 수 없다.

함수 내부의 계산 결과를 함수 밖으로 전달하여 사용하고 싶다면 **반드시 `return` 키워드를 사용하여 값을 밖으로 넘겨주어야 한다.**

---

###  **지역변수의 유효 범위(Scope)와 전달 구조**

```text
 def calculate_sales():
     local_total = 300000   ➔ 함수 내부에서만 존재하는 지역변수
     return local_total     ➔ 밖으로 값을 가지고 호출한 곳으로 돌아감!
        │
        └──────┐
                      ▼
 result = calculate_sales()  ➔ 바깥의 변수(result)가 그 값을 넘겨받아 저장
```

````python
# [실습 1] 지역변수의 올바른 사용법 (return 활용)
def calculate_sales():
    local_total = 300000  # 지역변수 선언
    return local_total    # 값을 바깥으로 반환

# 함수를 호출하여 반환된 값을 바깥의 result 변수에 저장
result = calculate_sales()
print("반환된 매출액:", result)  # 출력: 300000


# [실습 2] 자주 하는 실수: 함수 밖에서 지역변수 직접 접근 시도
# print(local_total)
#  NameError: name 'local_total' is not defined
# (local_total은 함수가 끝나며 사라졌으므로 밖에서는 존재하지 않는 변수 취급을 받는다.)
````

## **8) 람다: 이름 없는 한 줄 함수**

`def`로 만드는 함수는 이름이 있고 여러 줄을 쓸 수 있다. 계산이 한 줄로 끝나고 한 번만 쓸 함수라면 `lambda`로 이름 없이 바로 만들 수 있다.

```python
def get_amount(row):
    return row[1]

get_amount = lambda row: row[1]     # 위와 같은 동작
```

```text
 lambda  매개변수 :  반환할 식
   │        │          └─ return 없이 식만 쓴다. 이 값이 그대로 반환된다
   │        └─ def와 같은 방식으로 적는다
   └─ "이름 없는 함수를 만든다"는 뜻
```

람다는 주로 **다른 함수에 "기준"을 넘길 때** 쓴다. 대표적인 예가 `sorted()`의 `key`다. `key`에 함수를 주면 그 함수의 반환값을 기준으로 정렬한다.


두 줄 이상이거나 여러 곳에서 다시 쓸 함수는 `def`로 만든다.

공식 문서: [람다 표현식](https://docs.python.org/ko/3.13/tutorial/controlflow.html#lambda-expressions) · [sorted()](https://docs.python.org/ko/3.13/library/functions.html#sorted)


````python
# [예제] 1. def와 lambda는 같은 쓰임
def add_vat(amount):
    return int(amount * 1.1)

add_vat_lambda = lambda amount: int(amount * 1.1)

print(add_vat(100000))          # 110000
print(add_vat_lambda(100000))   # 110000




# 2. sorted()의 key에는 "함수"를 넣는다. 먼저 이미 아는 함수 len을 넣어 본다
#    key=len → 각 값에 len()을 적용한 결과(글자 수)를 기준으로 정렬한다
regions = ["서울특별시", "부산", "대구광역시", "광주"]

print(sorted(regions))                # ['광주', '대구광역시', '부산', '서울특별시']  기본: 가나다순
print(sorted(regions, key=len))       # ['부산', '광주', '서울특별시', '대구광역시']  글자 수 짧은 순





# 3. 기준이 되는 함수가 없으면 람다로 그 자리에서 만든다
#    key=lambda row: row[1] → 각 튜플의 두 번째 값(금액)을 기준으로 정렬한다
orders = [("노트북", 2400000), ("마우스", 250000), ("키보드", 225000)]

print(sorted(orders, key=lambda row: row[1]))                 # 금액 오름차순
print(sorted(orders, key=lambda row: row[1], reverse=True))   # 금액 내림차순
print(sorted(orders, key=lambda row: row[0]))                 # 제품명 가나다순


# sorted(orders, key=lambda row: row[1]) 가 내부적으로 하는 일
#
# 1) sorted()는 orders의 항목을 하나씩 꺼내 람다에 넣는다. 꺼낸 항목이 row에 담긴다.
#      row = ("노트북", 2400000)  →  row[1] = 2400000
#      row = ("마우스", 250000)   →  row[1] = 250000
#      row = ("키보드", 225000)   →  row[1] = 225000
#
# 2) 람다가 돌려준 값(금액)을 "정렬 기준표"로 삼는다. 원래 항목은 그대로 두고 기준값만 비교한다.
#      기준값: 2400000, 250000, 225000  →  작은 순: 225000 < 250000 < 2400000
#
# 3) 기준값 순서대로 원래 항목(튜플 전체)을 다시 배열해 새 리스트로 돌려준다.
#      [("키보드", 225000), ("마우스", 250000), ("노트북", 2400000)]
#
# 즉 람다는 "각 항목에서 어떤기준으로 정렬할지만 알려 준다.
````

````python
# [실습] 일반 함수를 lambda 함수로 바꾸기
# 문제: def로 일반 함수를 만든 후, 같은 기능을 lambda로 작성한다.

# 1. 숫자 하나를 받아 두 배를 반환하는 double(x) 함수를 만든다.
#    double(5)를 호출해 출력한다.
# 출력: 10
def double1(x):
    return x*2

double2 = lambda x: x*2 

# 2. 위의 double 함수를 lambda로 변경하여 작성한다.
#    double(5)를 호출해 출력한다.
# 출력: 10
print(double1(5))
print(double2(5))

# 3. 두 숫자를 더한 결과를 반환하는 add(a, b) 함수를 만든다.
#    add(10, 20)을 호출해 출력한다.
# 출력: 30
def add(a,b):
    return a+b

# 4. 위의 add 함수를 lambda로 변경하여 작성한다.
#    add(10, 20)을 호출해 출력한다.
# 출력: 30
add2 = lambda a,b: str(a+b)

print(add(2,3))

print(add2(2,3))
````

## **예외 처리: `try`·`except`·`else`·`finally`**

프로그램 실행 중 문제가 발생하면 Python은 **예외(Exception)** 를 발생시킨다. 발생한 예외를 처리하지 않으면 오류 메시지가 출력되고 실행이 중단될 수 있다.

`try`·`except`를 사용하면 오류가 발생할 가능성이 있는 코드를 실행하고, 특정 예외가 발생했을 때 프로그램이 바로 중단되지 않도록 처리할 수 있다.

### 기본 구조

```python
try:
    실행해 볼 코드

except 오류이름:
    지정한 오류가 발생했을 때 실행할 코드

except Exception as e:
    앞에서 처리하지 않은 일반적인 오류를 처리할 코드

else:
    오류가 발생하지 않았을 때 실행할 코드

finally:
    오류 발생 여부와 관계없이 마지막에 실행할 코드
```

- **`try`**: 오류가 발생할 가능성이 있는 코드를 실행한다.
- **`except`**: 지정한 예외가 발생하면 실행한다.
- **`except Exception as e`**: 앞에서 처리하지 않은 대부분의 일반적인 예외를 처리한다.
- **`else`**: `try`에서 예외가 발생하지 않았을 때 실행한다. (생략 가능)
- **`finally`**: 예외 발생 여부와 관계없이 마지막에 실행한다. (생략 가능)

`except`에서는 `ValueError`, `TypeError`, `KeyError`, `ZeroDivisionError`, `ImportError` 등 실제 예외 이름을 사용한다.

### `Exception`으로 일반 예외 처리하기

`Exception`은 실행 중 발생하는 대부분의 일반적인 예외를 처리하는 데 사용할 수 있다.

예상되는 오류는 `ValueError`처럼 구체적으로 지정하는 것이 좋으며, `Exception`은 여러 종류의 오류를 포괄적으로 처리해야 할 때 사용한다.

`except Exception as e`에서 `e`에는 발생한 예외 객체가 저장되며, `print(e)`로 오류 내용을 확인할 수 있다.

단, 예상하지 못한 오류까지 무조건 무시하면 문제의 원인을 발견하기 어려우므로 주의해야 한다.

### `else`와 `finally`의 차이

| 구분 | 실행 조건 |
|---|---|
| `except` | 해당 예외가 발생했을 때 |
| `else` | `try`에서 예외가 발생하지 않았을 때 |
| `finally` | 예외 발생 여부와 관계없이 실행 |

**실행 흐름**

- 정상 실행: `try` → `else` → `finally`
- 예외 발생 및 처리: `try` → `except` → `finally`

`finally`는 주로 파일 닫기, 데이터베이스 연결 해제처럼 실행 결과와 관계없이 필요한 정리 작업에 사용한다.

### 실행 예제

```python
# "10", "0", "abc"로 값을 바꿔서 실행해 본다.
value = "10"

try:
    number = int(value)
    result = 100 / number

except ValueError:
    print("숫자로 변환할 수 없습니다.")

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")

except Exception as e:
    print("기타 오류:", e)

else:
    print("계산 결과:", result)

finally:
    print("계산 작업 완료")
```

**1 `value = "10"`인 경우**

```text
계산 결과: 10.0
계산 작업 완료
```

**2 `value = "0"`인 경우**

```text
0으로 나눌 수 없습니다.
계산 작업 완료
```

**3 `value = "abc"`인 경우**

```text
숫자로 변환할 수 없습니다.
계산 작업 완료
```

**핵심 정리**

- `except`: 오류가 발생하면 처리한다.
- `else`: 오류가 없을 때 실행한다.
- `finally`: 오류 발생 여부와 관계없이 실행한다.

예외 처리는 오류를 없애는 것이 아니라, **오류가 발생했을 때 어떻게 대응할지 정하는 방법**이다.

공식 문서: [Python 3.13 — 예외 처리하기](https://docs.python.org/ko/3.13/tutorial/errors.html#handling-exceptions) · [뒷정리 동작 정의하기(finally)](https://docs.python.org/ko/3.13/tutorial/errors.html#defining-clean-up-actions)

6. 평균 매출에 따른 성과 등급
   - 170000 이상: 우수
   - 120000 이상: 보통
   - 그 외: 개선 필요
7. 위 계산을 함수  create_sales_summary  로 작성

````python
sales = [
    120000,
    180000,
    95000,
    240000,
    160000,
    110000,
    200000
]
````

````python
# summary = create_sales_summary(sales)
def create_sales_summary(sales):
    count = len(sales)
    total = sum(sales)
    avg = total/count
    max_sales = max(sales)
    min_sales = min(sales)
    filter_count= len(list(filter(lambda x: x>150000,sales)))
    filter_sum = sum(list(filter(lambda x: x>150000,sales)))
    grade = (lambda sales: "우수" if sales>170000 else  "보통" if sales>120000 else "개선필요")(avg)















    return {
        "count"        : count,
        "total"        : total,
        "average"      : avg,
        "max_sales"    : max_sales,
        "min_sales"    : min_sales,
        "filter_count" : filter_count,
        "filtet_sum" : filter_sum,
        "grade" : grade
    }

# for key, value in summary.items():
#     print(key, value)

summary = create_sales_summary(sales)
print(summary)
print("------------------------")
for key, value in summary.items():
    if type(value) == float:  # 값의 자료형을 확인
        print(f"{key}: {value:,.2f}")
    elif type(value) == int:
        print(f"{key}: {value:,}")
    else:
        print(f"{key}: {value}")



````
