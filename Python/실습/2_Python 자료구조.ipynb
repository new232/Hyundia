# **1. 튜플(Tuple)**

튜플은 리스트처럼 여러 값을 순서대로 저장하는 자료형이다.

하지만 리스트와 달리 **한 번 생성한 튜플은 내부 요소를 수정하거나 추가·삭제할 수 없는 불변(Immutable) 객체**이다.

튜플은 일반적으로 소괄호 `()`를 사용하여 작성하며, 각 요소를 쉼표(`,`)로 구분한다.

```python
sales_range = (100000, 200000)

print(sales_range)        # (100000, 200000)
print(type(sales_range))  # <class 'tuple'>
```

튜플도 리스트처럼 인덱싱과 슬라이싱을 사용할 수 있다.

```python
sales_range = (100000, 200000)

print(sales_range[0])   # 100000
print(sales_range[-1])  # 200000
print(sales_range[:1])  # (100000,)
```

단, 튜플은 불변 객체이므로 요소를 직접 수정할 수 없다.

```python
sales_range[0] = 150000  # TypeError 발생
```

### **[참고] 튜플 생성 시 주의할 점**

튜플을 만드는 핵심은 소괄호가 아니라 **쉼표(`,`)**이다.

```python
a = (10, 20)   # 튜플
b = 10, 20     # 튜플
c = (10)       # 정수
d = (10,)      # 요소가 하나인 튜플

print(type(a))  # <class 'tuple'>
print(type(b))  # <class 'tuple'>
print(type(c))  # <class 'int'>
print(type(d))  # <class 'tuple'>
```

```python

# df.shape → (1000, 5)
```

````python
# 튜플 생성 및 인덱싱
sales_range = (100000, 200000)

print("전체 튜플:", sales_range)
print("자료형:", type(sales_range))  # <class 'tuple'>

# 리스트와 동일하게 인덱싱으로 값 조회 가능
print("sales_range[0]:", sales_range[0]) # 100000
print("sales_range[1]:", sales_range[1]) # 200000
````

## **언패킹(Unpacking)**

언패킹은 **리스트, 튜플 등 반복 가능한 객체(Iterable)에 들어 있는 여러 요소를 순서대로 각각의 변수에 할당하는 기능**이다.

데이터 분석과 파이썬 프로그래밍에서는 함수가 여러 결과를 튜플로 반환할 때, 각 결과를 개별 변수로 받아 처리하는 데 자주 사용한다.

### **1. 튜플 언패킹**

```text
(100000, 200000)
       ↓
min_sales, max_sales = (100000, 200000)
    ↓          ↓
 100000      200000
```

```python
sales_range = (100000, 200000)

min_sales, max_sales = sales_range

print(min_sales)  # 100000
print(max_sales)  # 200000
```

### **2. 리스트 언패킹**

리스트도 튜플과 동일한 방식으로 언패킹할 수 있다.

```python
sales = [100000, 200000]

min_sales, max_sales = sales

print(min_sales)  # 100000
print(max_sales)  # 200000
```

### **3. 언패킹의 기본 조건**

언패킹하려면 오른쪽에 반복 가능한 객체가 있어야 한다.

또한 일반적인 언패킹에서는 **왼쪽 변수 개수와 오른쪽 요소 개수가 일치해야 한다.** 개수가 일치하지 않으면 `ValueError`가 발생한다.

```python
sales_range = (100000, 200000)

# 정상적인 언패킹
a, b = sales_range

# 변수 개수가 부족한 경우
# a, = sales_range  # ValueError 발생

# 변수 개수가 많은 경우
# a, b, c = sales_range  # ValueError 발생
```

단, `*`를 사용하는 확장 언패킹에서는 여러 요소를 하나의 리스트로 받을 수 있다.

**공식 문서:** [Python 3.13 — 튜플과 시퀀스](https://docs.python.org/ko/3.13/tutorial/datastructures.html#tuples-and-sequences)

````python
# 1. 기본적인 튜플 언패킹
minimum_sales, maximum_sales = (100000, 200000)

print(minimum_sales)  # 100000
print(maximum_sales)  # 200000


# 2. 튜플 언패킹: 행과 열 개수
shape_info = (100, 5)  # (행 개수, 열 개수)
rows, cols = shape_info

print(f"행 개수: {rows}개, 열 개수: {cols}개")
# 행 개수: 100개, 열 개수: 5개


# 3. 리스트 언패킹
sales_list = [120000, 150000, 90000]

sales1, sales2, sales3 = sales_list

print(sales1)  # 120000
print(sales2)  # 150000
print(sales3)  # 90000


# 4. 확장 언패킹: *를 사용하여 여러 요소를 리스트로 받기
sales = [100000, 200000, 300000, 400000]

first, *middle, last = sales
test2,*test = sales
print(test)
print(first)   # 100000
print(middle)  # [200000, 300000]
print(last)    # 400000
````

````python
# [실습] 튜플 언패킹
# 문제: 튜플에 저장된 매출 범위와 데이터 크기 정보를 언패킹하여
#       각각의 변수에 할당하고 지정된 형식으로 출력하시오.

# 1. 이번 달 매출 범위(sales_range)에 (85000, 210000)을 할당한다.
sales_range = (85000, 210000)

# 2. 매출 범위를 minimum과 maximum에 각각 언패킹한다.
mininum, maximum = sales_range

# 3. 최소 매출과 최대 매출을 지정된 형식으로 출력한다.
# 출력: 최소: 85000, 최대: 210000
print(f"최소: {mininum}, 최대: {maximum} ")
# 4. 데이터 크기(shape)에 (1500, 8)을 할당한다.
shape = (1500, 8)

# 5. 데이터 크기를 rows와 cols에 각각 언패킹한다.
rows, cols  = shape

# 6. 행 개수와 열 개수를 지정된 형식으로 출력한다.
# 출력: 1500행 8열
print(f"출력: {rows}행 {cols}열")
````

## **튜플 패킹(Tuple Packing)**

튜플 패킹은 **여러 값을 쉼표(`,`)로 구분하여 하나의 튜플로 묶는 것**이다.

튜플은 소괄호 `()`를 생략해도 쉼표로 값을 구분하면 튜플로 생성된다.

반대로 **언패킹(Unpacking)**은 튜플에 들어 있는 여러 요소를 각각의 변수에 나누어 할당하는 것이다.

```python
# 패킹: 여러 값을 하나의 튜플로 묶기
size = 30, 50

print(size)        # (30, 50)
print(type(size))  # <class 'tuple'>

```


**공식 문서:** [Python 3.13 — 튜플과 시퀀스](https://docs.python.org/ko/3.13/tutorial/datastructures.html#tuples-and-sequences)

````python
# [예제] 튜플 패킹과 언패킹
# 값을 쉼표로 나열하면 소괄호가 없어도 튜플이 된다 (패킹)
size = 30, 50
print(size)            # (30, 50)
print(type(size))      # <class 'tuple'>

# 튜플을 풀어 변수 여러 개에 나누어 담는다 (언패킹)
width, height = size
print(width, height)   # 30 50
````

````python
# [실습] 튜플 패킹
# 문제: 상자의 가로·세로·높이 40, 25, 10 을 쉼표만으로 묶어 box 에 담아 출력하고 자료형을 확인한 뒤, 언패킹해 부피를 출력한다.
box= 40,25,10
print(type(box))
가로,세로,높이 = box


#  부피(가로 × 세로 × 높이)를 출력
print("가로:",가로 ,"세로",세로,"높이",높이)
print(가로*세로*높이)
````

## **튜플에서 사용할 수 있는 메서드**

튜플은 생성한 후 값을 추가·수정·삭제할 수 없는 자료형이다. 따라서 리스트의 `append()`, `remove()`, `pop()`과 같은 변경 메서드를 사용할 수 없다.

튜플에서는 주로 다음 두 메서드를 사용한다.

| 메서드 | 기능 |
|---|---|
| `count(값)` | 지정한 값이 튜플에 몇 번 있는지 반환 |
| `index(값)` | 지정한 값이 처음 나타나는 위치를 반환 |

`count()`와 `index()`는 튜플의 내용을 변경하지 않고 조회만 한다. 두 메서드는 리스트에서도 같은 방식으로 쓸 수 있다.

공식 문서: [공통 시퀀스 연산](https://docs.python.org/ko/3.13/library/stdtypes.html#common-sequence-operations)


````python
sales = (120000, 150000, 90000, 150000)

# count(): 지정한 값의 개수 확인
target_count = sales.count(150000)

print(target_count)  # 2


# index(): 지정한 값이 처음 나타나는 위치 확인
first_index = sales.index(150000)

print(first_index)   # 1

# 튜플에 없는 값을 index()로 찾으면 ValueError가 발생한다.
# sales.index(200000)
````

````python
# [실습] 튜플 메서드
# 문제: 검사 결과 튜플 results = ("양품", "양품", "불량", "양품", "불량", "양품", "양품") 에서 count()로 불량 횟수,
#  index()로 첫 불량 위치를 출력하고, 양품 수 ÷ 전체 수를 백분율로 "양품률: 71.4%" 형식으로 출력한다.


results = ("양품", "양품", "불량", "양품", "불량", "양품", "양품") 
find_fail = results.index("불량")
print(find_fail)
print(f"양품률: {results.count("양품")/len(results)*100:.1f}%")
````

# **2. 딕셔너리(Dictionary)**

리스트는 `sales[0]`, `sales[1]`처럼 위치 번호(인덱스)를 사용하여 요소에 접근한다. 하지만 데이터가 많아지면 **몇 번 위치에 어떤 데이터가 있는지** 기억하기 어려울 수 있다.

딕셔너리는 **키(Key)와 값(Value)을 쌍으로 저장하는 자료형**이다.

- **키(Key)**: 데이터를 찾기 위한 이름이나 식별자
- **값(Value)**: 키에 연결된 실제 데이터

딕셔너리는 **키를 사용하여 원하는 값에 접근한다.**

예를 들어 지역별 매출 데이터에서 `"서울"`이라는 키를 사용하면 서울 지역의 매출을 가져올 수 있다.

## **딕셔너리 생성**

딕셔너리는 중괄호 `{}` 안에 **`키: 값` 형태로 데이터를 작성**한다. 여러 개의 키-값 쌍은 쉼표(`,`)로 구분한다.

```python id="del6gz"
region_sales = {
    "서울": 350000,
    "부산": 280000
}
```

| 키(Key) | 값(Value) |
|---|---:|
| `"서울"` | 350000 |
| `"부산"` | 280000 |

키를 지정하면 해당 키에 연결된 값을 가져올 수 있다.

```python id="m1n6aa"
region_sales = {"서울": 350000, "부산": 280000}

print(region_sales["서울"])  # 350000
print(region_sales["부산"])  # 280000

print(type(region_sales))  # <class 'dict'>
```

딕셔너리의 키는 중복될 수 없으며, 값은 중복될 수 있다. 또한 딕셔너리는 **가변(Mutable) 객체**이므로 생성한 후에도 키-값 쌍을 추가하거나 값을 수정·삭제할 수 있다.

**공식 문서:** [Python 3.13 — 딕셔너리](https://docs.python.org/ko/3.13/tutorial/datastructures.html#dictionaries)

## **값 조회·추가·수정**

딕셔너리는 **키(Key)를 사용하여 값을 조회하거나 추가·수정할 수 있다.**

- `딕셔너리[키]`: 지정한 키에 연결된 값을 조회한다.
- `딕셔너리[키] = 값`: 해당 키가 없으면 새로운 키-값 쌍을 추가한다.
- `딕셔너리[키] = 값`: 해당 키가 이미 있으면 기존 값을 수정한다.


````python
# 1. 딕셔너리 생성 및 조회
region_sales = {
    "서울": 350000,
    "부산": 280000,
    "대구": 210000
}

print(region_sales)           # 전체 딕셔너리 출력
print(region_sales["서울"])   # 350000 ("서울" 키에 연결된 값 조회)

# 2. 값 추가 및 수정
region_sales["광주"] = 180000 # '광주' 키 추가
region_sales["서울"] = 400000 # '서울' 키의 기존 값 수정

print(region_sales)           # {'서울': 400000, '부산': 280000, '대구': 210000, '광주': 180000}
````

## **키와 값 전체 꺼내기 — 딕셔너리 뷰(Dictionary View)**

딕셔너리는 `keys()`, `values()`, `items()` 메서드를 사용하여 키와 값을 확인할 수 있다.

이 메서드들은 리스트가 아니라 **딕셔너리 뷰 객체(Dictionary View Object)**를 반환한다.

| 메서드 | 기능 | 반환 객체의 자료형 |
|---|---|---|
| `keys()` | 딕셔너리의 모든 키를 보여준다 | `dict_keys` |
| `values()` | 딕셔너리의 모든 값을 보여준다 | `dict_values` |
| `items()` | 딕셔너리의 모든 키-값 쌍을 보여준다 | `dict_items` |

### **1. 딕셔너리 뷰 객체란?**

뷰(View)는 **원본 딕셔너리의 데이터를 동적으로 보여주는 객체**이다.

- 원본 딕셔너리가 변경되면 뷰에도 변경 내용이 반영된다.
- 반복문에서 사용할 수 있다.
- 리스트와 달리 위치 번호를 이용한 인덱싱(`[0]`)은 사용할 수 없다.
- `list()`를 사용하면 뷰에 포함된 요소들을 새로운 리스트로 만들 수 있다.

### **2. keys(), values(), items() 사용하기**

```python
region_sales = {"서울": 350000, "부산": 280000}

print(region_sales.keys())
# dict_keys(['서울', '부산'])

print(region_sales.values())
# dict_values([350000, 280000])

print(region_sales.items())
# dict_items([('서울', 350000), ('부산', 280000)])
```

`items()`는 각 키와 값을 `(키, 값)` 형태의 튜플로 제공한다.

### **3. 원본 딕셔너리 변경과 뷰 객체**

뷰 객체는 원본 딕셔너리와 연결되어 있으므로 딕셔너리의 내용이 변경되면 변경된 데이터를 보여준다.

```python
region_sales = {"서울": 350000, "부산": 280000}

keys_view = region_sales.keys()

print(keys_view)
# dict_keys(['서울', '부산'])

# 원본 딕셔너리에 새로운 항목 추가
region_sales["인천"] = 210000

print(keys_view)
# dict_keys(['서울', '부산', '인천'])
```

`keys_view`를 다시 생성하지 않아도 추가된 `"인천"`이 반영된다.

### **4. 뷰 객체를 리스트로 변환하기**

딕셔너리 뷰 객체는 리스트처럼 인덱싱할 수 없지만, `list()`로 변환하면 인덱싱이 가능하다.

```python
region_sales = {"서울": 350000, "부산": 280000}

keys_list = list(region_sales.keys())

print(keys_list)     # ['서울', '부산']
print(keys_list[0])  # 서울
```



**공식 문서:** [Python 3.13 — 딕셔너리 뷰 객체](https://docs.python.org/ko/3.13/library/stdtypes.html#dictionary-view-objects)

````python


region_sales = {
    "서울": 350000,
    "부산": 280000,
    "대구": 210000
}

# 1. 딕셔너리 뷰 확인
print(region_sales.keys())
# dict_keys(['서울', '부산', '대구'])

print(region_sales.values())
# dict_values([350000, 280000, 210000])

print(region_sales.items())
# dict_items([('서울', 350000), ('부산', 280000), ('대구', 210000)])


# 2. 자료형 확인
print(type(region_sales.keys()))    # <class 'dict_keys'>
print(type(region_sales.values()))  # <class 'dict_values'>
print(type(region_sales.items()))   # <class 'dict_items'>
````

````python


# 3. list()로 변환하면 일반 리스트가 된다.
print(list(region_sales.keys()))
# ['서울', '부산', '대구']

print(list(region_sales.values()))
# [350000, 280000, 210000]

print(list(region_sales.items()))
# [('서울', 350000), ('부산', 280000), ('대구', 210000)]

print(type(list(region_sales.keys())))
# <class 'list'>

````

````python


# 4. 뷰와 리스트의 차이: 원본 데이터 변경
keys_view = region_sales.keys()        # 뷰 객체
keys_list = list(region_sales.keys())  # 리스트로 변환

# 원본 딕셔너리에 새로운 데이터 추가
region_sales["인천"] = 190000

print(keys_view)
# dict_keys(['서울', '부산', '대구', '인천'])
# → 뷰는 원본의 변경 사항이 반영된다.

print(keys_list)
# ['서울', '부산', '대구']
# → 리스트는 변환 당시의 데이터를 가지고 있다.


# 5. 인덱싱 차이
# print(keys_view[0])  # 오류! dict_keys는 인덱싱 불가능

print(keys_list[0])     # 서울
# → 리스트는 인덱싱이 가능하다.

````

````python
# [실습] 딕셔너리의 키와 값 전체 꺼내기
# 문제: 설비별 생산량을 저장한 딕셔너리에서
#       keys(), values(), items()를 사용하여 데이터를 확인하시오.

# 1. 설비별 생산량(production)에 {"M01": 120, "M02": 95, "M03": 130}을 할당한다.

production = {
    "M01": 120,
    "M02": 95,
    "M03": 130
}
# 2. 모든 설비 이름(키)을 출력한다.
# 출력: dict_keys(['M01', 'M02', 'M03'])
print(production.keys())

# 3. 모든 생산량(값)을 출력한다.
# 출력: dict_values([120, 95, 130])
print(production.values())

# 4. 모든 설비 이름과 생산량(키-값 쌍)을 출력한다.
# 출력: dict_items([('M01', 120), ('M02', 95), ('M03', 130)])
print(production.items())
print(production)
````

## **`get()`으로 안전하게 조회**

없는 키를 대괄호(dict["제주"])로 조회하면 프로그램이 멈추면서 KeyError가 발생한다.

반면 get() 메서드를 사용하면 없는 키라도 에러 대신 None을 반환하거나 기본값을 지정해 출력할 수 있어 매우 안전하다.

```python
딕셔너리.get(키)
딕셔너리.get(키, 없을때값)
```

````python
region_sales = {
    "서울": 350000,
    "부산": 280000
}

# 대괄호 조회: 없는 키 조회 시 KeyError 발생!
# print(region_sales["제주"]) # KeyError: '제주'

# get() 조회: 에러 없이 안전하게 조회
print(region_sales.get("서울"))     # 350000 (존재하는 키)
print(region_sales.get("제주"))     # None   (없는 키지만 에러 없이 None 반환)
print(region_sales.get("제주", 0))  # 0      (없는 키일 경우 기본값 0 반환)
````

````python
# [실습] get()으로 안전하게 조회
# 문제: 거래처별 미수금 딕셔너리(receivable)에서 get() 메서드를 사용하여
#       존재하는 키와 존재하지 않는 키를 조회하고 결과를 비교하시오.

# 1. 거래처별 미수금(receivable)에 {"A전자": 350000, "B식품": 0}을 할당한다.
receivable = {
    "A전자": 35000,
    "B식품": 0
}

# 2. "A전자"의 미수금을 조회하고 결과를 출력한다.
# 출력: 350000
print(receivable.get('A전자'))


# 3. 존재하지 않는 "C물산"의 미수금을 조회하고 결과를 출력한다.
# 출력: None
print(receivable.get('C물산'))
    
# 4. "C물산"의 미수금을 조회하되, 키가 없으면 기본값 0을 반환하도록 설정하고 출력한다.

print(receivable.get('C물산',90))
# 출력: 0을                 
````

## **딕셔너리 값 추가·수정·삭제**

딕셔너리는 **가변(Mutable) 객체**이므로 생성한 후에도 키-값 쌍을 추가하거나 기존 값을 수정·삭제할 수 있다.

딕셔너리에서 제공하는 `update()`, `pop()`, `popitem()`, `clear()` 메서드를 사용하면 여러 항목을 수정하거나 삭제할 수 있다.

| 메서드 | 기능 | 반환값 |
|---|---|---|
| `update()` | 여러 키-값 쌍을 추가하거나 수정 | `None` |
| `pop(키)` | 지정한 키와 해당 값을 삭제 | 삭제한 값 |
| `popitem()` | 마지막으로 추가된 키-값 쌍 삭제 | 삭제한 `(키, 값)` 튜플 |
| `clear()` | 모든 키-값 쌍 삭제 | `None` |

### **1. 여러 값 추가·수정 — update()**

`update()`는 다른 딕셔너리의 키와 값을 기존 딕셔너리에 한 번에 반영하는 메서드이다.

- 기존에 없는 키는 새로운 키-값 쌍으로 추가한다.
- 이미 존재하는 키는 새로운 값으로 수정한다.
- 기존 딕셔너리를 직접 변경하며, 반환값은 `None`이다.

**기본 문법**

```python
딕셔너리.update({키1: 값1, 키2: 값2})
```

```python
region_sales = {"서울": 350000, "부산": 280000}

result = region_sales.update({"서울": 400000, "인천": 210000})

print(region_sales)
# {'서울': 400000, '부산': 280000, '인천': 210000}

print(result)  # None
```

`"서울"`은 기존 키이므로 값이 수정되고, `"인천"`은 새로운 키이므로 추가된다.

### **2. 지정한 키 삭제 — pop()**

`pop()`은 지정한 키와 해당 값을 딕셔너리에서 삭제하고, **삭제한 값을 반환하는 메서드**이다.

**기본 문법**

```python
딕셔너리.pop(키)          # 지정한 키를 삭제하고 값 반환
딕셔너리.pop(키, 기본값)  # 키가 없으면 기본값 반환
```

- 키가 존재하면 해당 항목을 삭제하고 값을 반환한다.
- 키가 없고 기본값을 지정하지 않으면 `KeyError`가 발생한다.
- 키가 없지만 기본값을 지정하면 오류 없이 기본값을 반환하며, 딕셔너리는 변경되지 않는다.

```python
region_sales = {"서울": 350000, "부산": 280000}

# 존재하는 키 삭제
removed = region_sales.pop("서울")

print(removed)       # 350000
print(region_sales)  # {'부산': 280000}

# 존재하지 않는 키에 기본값 지정
removed = region_sales.pop("인천", 0)

print(removed)       # 0
print(region_sales)  # {'부산': 280000}
```

여기서 `0`은 삭제한 값이 아니라 **키가 존재하지 않을 때 대신 반환하는 기본값**이다.

**[참고] 리스트와 딕셔너리의 pop() 비교**

| 구분 | 리스트 `pop()` | 딕셔너리 `pop()` |
|---|---|---|
| 삭제 기준 | 인덱스 | 키 |
| 기본 문법 | `리스트.pop(인덱스)` | `딕셔너리.pop(키, 기본값)` |
| 인자 생략 | 마지막 요소 삭제 | 키 생략 불가능 |
| 반환값 | 삭제한 요소 | 삭제한 키의 값 |
| 대상이 없을 때 | `IndexError` | `KeyError` |
| 기본값 지정 | 불가능 | 가능 |

리스트의 `pop(0)`에서 `0`은 삭제할 요소의 인덱스이며, 딕셔너리의 `pop("인천", 0)`에서 `0`은 키가 없을 때 반환할 기본값이다.

### **3. 마지막 항목 삭제 — popitem()**

`popitem()`은 딕셔너리에 **마지막으로 추가된 키-값 쌍을 삭제하고 튜플로 반환하는 메서드**이다.

**기본 문법**

```python
딕셔너리.popitem()
```

```python
region_sales = {"서울": 350000, "부산": 280000, "인천": 210000}

removed = region_sales.popitem()

print(removed)
# ('인천', 210000)

print(region_sales)
# {'서울': 350000, '부산': 280000}
```

반환값은 `(키, 값)` 형태의 튜플이며, 딕셔너리가 비어 있으면 `KeyError`가 발생한다.

### **4. 모든 항목 삭제 — clear()**

`clear()`는 딕셔너리에 저장된 **모든 키-값 쌍을 삭제하는 메서드**이다.

기존 딕셔너리 객체를 직접 변경하며, 반환값은 `None`이다.

**기본 문법**

```python
딕셔너리.clear()
```

```python
region_sales = {"서울": 350000, "부산": 280000}

result = region_sales.clear()

print(region_sales)  # {}
print(result)        # None
```

`clear()`를 실행하면 딕셔너리 객체 자체가 삭제되는 것이 아니라 내부 항목이 모두 제거되어 빈 딕셔너리 `{}`가 된다.

### **[참고] del을 이용한 삭제**

`del`은 딕셔너리 메서드가 아닌 **파이썬의 삭제문(del statement)**이다. 지정한 키와 해당 값을 삭제할 수 있다.

```python
region_sales = {"서울": 350000, "부산": 280000}

del region_sales["부산"]

print(region_sales)  # {'서울': 350000}
```

존재하지 않는 키를 삭제하면 `KeyError`가 발생한다.

**공식 문서**

- [Python 3.13 — 딕셔너리 메서드](https://docs.python.org/ko/3.13/library/stdtypes.html#dict)
- [Python 3.13 — del 문](https://docs.python.org/ko/3.13/reference/simple_stmts.html#the-del-statement)

````python
region_sales = {
    "서울": 350000,
    "부산": 280000
}

# 1. update(): 기존 키는 수정하고 새로운 키는 추가
update_result = region_sales.update({
    "부산": 300000,  # 기존 키의 값 수정
    "대구": 210000   # 새로운 키와 값 추가
})

print(region_sales)
# {'서울': 350000, '부산': 300000, '대구': 210000}

# update()의 반환값은 None
print(update_result)  # None


# 2. pop(): 지정한 키와 값을 삭제하고 삭제한 값 반환
removed = region_sales.pop("부산")

print(removed)       # 300000
print(region_sales)  # {'서울': 350000, '대구': 210000}


# 3. pop(): 키가 없으면 지정한 기본값 반환
removed = region_sales.pop("인천", 0)

print(removed)       # 0
print(region_sales)  # {'서울': 350000, '대구': 210000}


# 4. popitem(): 마지막으로 추가된 키-값 쌍 삭제 후 튜플 반환
removed_item = region_sales.popitem()

print(removed_item)  # ('대구', 210000)
print(region_sales)  # {'서울': 350000}


# 5. clear(): 모든 키-값 쌍 삭제
clear_result = region_sales.clear()

print(region_sales)  # {}
print(clear_result)  # None


# 6. del: 지정한 키와 값을 삭제
region_sales = {"서울": 350000, "부산": 280000}

del region_sales["서울"]

print(region_sales)  # {'부산': 280000}


# [참고] 오류 발생 사례

# pop()에서 없는 키를 기본값 없이 지정하면 KeyError가 발생한다.
# region_sales.pop("제주")

# popitem()을 빈 딕셔너리에서 실행하면 KeyError가 발생한다.
# {}.popitem()

# 딕셔너리에 없는 키를 del로 삭제하면 KeyError가 발생한다.
# del region_sales["제주"]
````

````python
# [실습] 딕셔너리 값 추가·수정·삭제
# 문제: 재고 딕셔너리(stock)의 기존 수량을 수정하고 새로운 제품을 추가한 뒤,
#       지정한 제품을 삭제하고 남은 제품의 전체 재고 수량을 계산하시오.

# 1. 재고(stock)에 {"노트북": 10, "마우스": 50}을 할당한다.

stock = {
    "노트북": 10, 
    "마우스": 50
    }
# 2. update()를 사용하여 "마우스"의 재고를 45로 수정하고,
#    새로운 제품 "키보드"의 재고 30을 추가한다.
stock.update({
    "마우스": 45,
    "키보드": 30
    }
)

# 3. 변경된 전체 재고를 출력한다.
# 출력: {'노트북': 10, '마우스': 45, '키보드': 30}
print(stock)

# 4. del을 사용하여 "노트북" 항목을 삭제한다.
del stock["노트북"]

# 5. 삭제 후 남은 재고를 출력한다.
# 출력: {'마우스': 45, '키보드': 30}
print(stock)

# 6. 남은 "마우스"와 "키보드"의 재고 수량을 더하여 전체 재고 수량을 출력한다.
# 출력: 전체 재고: 75개
count = 0
for key in stock:
    count += stock[key]

print(f"전체 재고: {count}개")
````

# **3. 집합(Set)**

집합(`set`)은 **중복되지 않는 값들을 모아 저장하는 자료형**이다.

일반적으로 중괄호 `{}` 안에 여러 값을 쉼표(`,`)로 구분하여 작성한다.

집합은 리스트나 튜플과 달리 **요소의 순서가 정해져 있지 않으며, 중복된 값을 허용하지 않는다.**

### **집합의 주요 특징**

- **중복 불가:** 같은 값이 여러 번 들어가도 하나만 저장된다.
- **순서 없음:** 요소의 위치가 정해져 있지 않으므로 인덱싱과 슬라이싱을 사용할 수 없다.
- **가변(Mutable) 객체:** 생성한 후에도 요소를 추가하거나 삭제할 수 있다.
- **요소 추가·삭제:** `add()`로 요소를 추가하고 `remove()`, `discard()`로 삭제할 수 있다.
- **출력 순서:** 요소가 출력되는 순서는 일정하다고 보장되지 않는다.
- **빈 집합:** `set()`으로 생성한다. `{}`는 빈 딕셔너리를 생성한다.

### **집합 생성과 중복 제거**

```python
regions = {"서울", "부산", "서울", "대구"}

print(regions)      # 중복된 "서울"은 하나만 남음 (출력 순서는 달라질 수 있음)
print(len(regions)) # 3

print(type(regions))  # <class 'set'>
```

집합에 `"서울"`이 두 번 작성되어 있지만 실제로는 하나만 저장되므로 요소의 개수는 3개이다.

### **빈 집합 생성**

```python
empty_set = set()
empty_dict = {}

print(type(empty_set))   # <class 'set'>
print(type(empty_dict))  # <class 'dict'>
```
단, 빈 중괄호 {}는 딕셔너리를 생성하므로 빈 집합은 반드시 set()으로 생성해야 한다.
집합은 데이터에서 **중복된 값을 제거하거나 서로 다른 값의 종류를 확인할 때** 활용할 수 있다.

**공식 문서:** [Python 3.13 — 집합형(set, frozenset)](https://docs.python.org/ko/3.13/library/stdtypes.html#set-types-set-frozenset)

````python
# [예제] 집합 생성과 기본 메서드

# 1. 중괄호 {}를 사용하여 집합 생성
categories = {"전자", "식품", "전자", "의류"}

print(categories)        # 중복된 "전자"는 하나만 남는다
print(type(categories))  # <class 'set'>


# 2. set()을 사용하여 집합 생성
categories = set(["전자", "식품", "전자", "의류"])

print(categories)        # 중복 제거
print(type(categories))  # <class 'set'>


# 3. add(): 새로운 요소 추가
categories.add("생활")

print(categories)


# 4. remove(): 지정한 요소 삭제
categories.remove("의류")

print(categories)


# 5. discard(): 지정한 요소 삭제
categories.discard("식품")

print(categories)


# 6. discard(): 존재하지 않는 요소를 삭제해도 오류가 발생하지 않음
categories.discard("가구")

print(categories)


# [참고] remove()와 discard()의 차이

# remove()는 존재하지 않는 요소를 삭제하면 KeyError 발생
# categories.remove("가구")

# discard()는 존재하지 않는 요소를 삭제해도 오류가 발생하지 않음
# categories.discard("가구")
````

## **집합 연산: 교집합, 합집합, 차집합**

집합은 두 데이터 집합을 비교하여 **공통 항목, 전체 항목, 한쪽에만 있는 항목**을 구할 수 있다.

예를 들어 지난달과 이번 달의 거래처 목록을 비교하면 신규 거래처, 이탈 거래처, 계속 거래 중인 거래처를 확인할 수 있다.

| 연산 | 기호 | 기호 이름(발음) | 메서드 | 결과 |
|---|---|---|---|---|
| 교집합 | `a & b` | 앰퍼샌드(Ampersand) | `a.intersection(b)` | 양쪽 모두에 있는 요소 |
| 합집합 | `a \| b` | 파이프(Pipe) | `a.union(b)` | 어느 한쪽에라도 있는 요소 |
| 차집합 | `a - b` | 마이너스(Minus) | `a.difference(b)` | a에는 있고 b에는 없는 요소 |

### **1. 교집합(Intersection)**

교집합은 두 집합에 **공통으로 포함된 요소**를 구하는 연산이다.

```python
last_month = {"A전자", "B식품", "C물산"}
this_month = {"B식품", "C물산", "D상사"}

print(last_month & this_month)
# {'B식품', 'C물산'}  # 출력 순서는 달라질 수 있음

print(last_month.intersection(this_month))
# {'B식품', 'C물산'}  # 출력 순서는 달라질 수 있음
```

두 달 모두 거래한 거래처는 `"B식품"`, `"C물산"`이다.

### **2. 합집합(Union)**

합집합은 두 집합에 포함된 **모든 요소를 중복 없이** 구하는 연산이다.

```python
print(last_month | this_month)
# {'A전자', 'B식품', 'C물산', 'D상사'}  # 출력 순서는 달라질 수 있음

print(last_month.union(this_month))
# {'A전자', 'B식품', 'C물산', 'D상사'}  # 출력 순서는 달라질 수 있음
```

지난달 또는 이번 달에 거래한 거래처는 총 4곳이다.

### **3. 차집합(Difference)**

차집합은 첫 번째 집합에는 있지만 두 번째 집합에는 없는 요소를 구하는 연산이다.

**① 신규 거래처**

```python
print(this_month - last_month)  # {'D상사'}

print(this_month.difference(last_month))
# {'D상사'}
```

이번 달에만 존재하는 `"D상사"`가 신규 거래처이다.

**② 이탈 거래처**

```python
print(last_month - this_month)  # {'A전자'}

print(last_month.difference(this_month))
# {'A전자'}
```

지난달에는 있었지만 이번 달에는 없는 `"A전자"`가 이탈 거래처이다.

### **[참고] 집합 연산의 특징**

- 집합 연산 결과는 집합(`set`) 객체이다.
- 집합은 중복된 요소를 허용하지 않는다.
- 출력되는 요소의 순서는 보장되지 않는다.
- `a - b`와 `b - a`는 서로 다른 결과가 나올 수 있으므로 차집합은 연산 순서에 주의해야 한다.
- `&`, `|`, `-` 연산으로 구한 결과는 새로운 집합이며, 원본 집합은 변경되지 않는다.

**공식 문서:** [Python 3.13 — 집합 연산](https://docs.python.org/ko/3.13/library/stdtypes.html#set-types-set-frozenset)

````python
# [실습] 집합 연산: 교집합·합집합·차집합
# 문제: 지난달과 이번 달의 거래처 목록을 비교하여
#       신규 거래처, 이탈 거래처, 계속 거래 중인 거래처와 전체 거래처 수를 출력하시오.

# 1. 지난달과 이번 달의 거래처를 각각 저장한다.
last_month = {"A전자", "B식품", "C물산"}
this_month = {"B식품", "C물산", "D상사"}

# 2. 신규 거래처를 출력한다.
print("신규 거래처:", this_month - last_month)
# 출력: 신규 거래처: {'D상사'}

# 3. 이탈 거래처를 출력한다.
print("이탈 거래처:", last_month - this_month)
# 출력: 이탈 거래처: {'A전자'}

# 4. 계속 거래 중인 거래처를 출력한다.
print("계속 거래:", last_month & this_month)
# 출력: 계속 거래: {'B식품', 'C물산'}

# 5. 전체 거래처 수를 중복 없이 출력한다.
print("전체 거래처 수:", len(last_month | this_month))
# 출력: 전체 거래처 수: 4

# 참고: 집합의 요소 출력 순서는 달라질 수 있다.
````

### **1. 리스트와 튜플의 가장 큰 차이는 무엇인가?**

리스트는 **가변(Mutable) 객체**이므로 생성한 후에도 요소를 추가·수정·삭제할 수 있다.

반면 튜플은 **불변(Immutable) 객체**이므로 생성한 후 요소를 직접 추가·수정·삭제할 수 없다.

```python
sales_list = [100000, 200000]
sales_list[0] = 150000

print(sales_list)  # [150000, 200000]

sales_tuple = (100000, 200000)

# sales_tuple[0] = 150000  # TypeError 발생
```

### **2. 튜플 패킹과 언패킹은 무엇이며, 일반적인 언패킹에서 주의할 점은 무엇인가?**

**패킹(Packing)**은 여러 값을 쉼표로 구분하여 하나의 튜플로 묶는 것이다.

**언패킹(Unpacking)**은 튜플이나 리스트 등 반복 가능한 객체의 요소를 여러 변수에 나누어 할당하는 것이다.

```python
# 패킹
size = 30, 50

print(size)  # (30, 50)

# 언패킹
width, height = size

print(width)   # 30
print(height)  # 50
```

일반적인 언패킹에서는 왼쪽 변수 개수와 오른쪽 요소 개수가 일치해야 하며, 개수가 다르면 `ValueError`가 발생한다.

단, `*`를 사용하는 확장 언패킹에서는 여러 요소를 하나의 리스트로 받을 수 있다.

### **3. 딕셔너리의 get()을 사용하면 어떤 장점이 있는가?**

`get()`은 존재하지 않는 키를 조회해도 `KeyError`가 발생하지 않는다.

기본값을 지정하지 않으면 `None`을 반환하며, 기본값을 지정하면 키가 없을 때 해당 값을 반환한다.

```python
region_sales = {"서울": 350000, "부산": 280000}

print(region_sales.get("서울"))     # 350000
print(region_sales.get("제주"))     # None
print(region_sales.get("제주", 0))  # 0

# print(region_sales["제주"])  # KeyError 발생
```

`get()`으로 조회하는 것만으로는 딕셔너리에 새로운 키가 추가되지 않는다.

### **4. 딕셔너리의 keys(), values(), items()는 각각 무엇을 반환하는가?**

- `keys()`: 모든 키를 보여 주는 `dict_keys` 뷰 객체
- `values()`: 모든 값을 보여 주는 `dict_values` 뷰 객체
- `items()`: 모든 `(키, 값)` 쌍을 보여 주는 `dict_items` 뷰 객체

이들은 일반 리스트가 아닌 **딕셔너리 뷰 객체(Dictionary View Object)**이므로 원본 딕셔너리의 변경 사항이 반영된다.

```python
region_sales = {"서울": 350000, "부산": 280000}

print(region_sales.keys())
# dict_keys(['서울', '부산'])

print(region_sales.values())
# dict_values([350000, 280000])

print(region_sales.items())
# dict_items([('서울', 350000), ('부산', 280000)])
```

뷰 객체는 리스트와 달리 위치 번호를 이용한 인덱싱을 지원하지 않는다.

### **5. 딕셔너리의 update(), pop(), popitem(), clear()는 각각 어떤 기능을 수행하는가?**

| 메서드 | 기능 | 반환값 |
|---|---|---|
| `update()` | 여러 항목을 추가하거나 기존 값을 수정 | `None` |
| `pop(키)` | 지정한 키와 값을 삭제 | 삭제한 값 |
| `popitem()` | 마지막으로 추가된 키-값 쌍을 삭제 | `(키, 값)` 튜플 |
| `clear()` | 모든 항목을 삭제 | `None` |

```python
stock = {"노트북": 10, "마우스": 50}

# 추가 및 수정
stock.update({"마우스": 45, "키보드": 30})
print(stock)
# {'노트북': 10, '마우스': 45, '키보드': 30}

# 지정한 키 삭제
removed = stock.pop("노트북")
print(removed)  # 10

# 마지막 항목 삭제
removed_item = stock.popitem()
print(removed_item)  # ('키보드', 30)

# 모든 항목 삭제
stock.clear()
print(stock)  # {}
```

딕셔너리의 `pop()`은 리스트의 `pop()`과 달리 키를 기준으로 항목을 삭제하며, 키가 없을 때 반환할 기본값을 지정할 수 있다.

### **6. 집합의 주요 특징은 무엇이며, 교집합·합집합·차집합은 어떻게 다른가?**

집합(`set`)은 **중복을 허용하지 않고 요소의 순서가 정해져 있지 않은 자료형**이다.

집합은 인덱싱과 슬라이싱을 지원하지 않으며, `set`은 가변 객체이므로 요소를 추가하거나 삭제할 수 있다.

| 연산 | 기호 | 기능 |
|---|---|---|
| 교집합 | `&` | 두 집합에 공통으로 있는 요소 |
| 합집합 | `\|` | 두 집합의 모든 요소를 중복 없이 결합 |
| 차집합 | `-` | 앞쪽 집합에만 있는 요소 |

```python
last_month = {"A전자", "B식품", "C물산"}
this_month = {"B식품", "C물산", "D상사"}

# 교집합: 계속 거래
print(last_month & this_month)
# {'B식품', 'C물산'}

# 합집합: 전체 거래처
print(last_month | this_month)
# {'A전자', 'B식품', 'C물산', 'D상사'}

# 차집합: 신규 거래처
print(this_month - last_month)
# {'D상사'}

# 차집합: 이탈 거래처
print(last_month - this_month)
# {'A전자'}
```

집합은 요소의 출력 순서가 보장되지 않는다. 또한 차집합은 연산 순서에 따라 결과가 달라질 수 있다.
