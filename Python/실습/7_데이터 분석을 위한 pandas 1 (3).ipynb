# **1. Series와 DataFrame**

### **DataFrame의 기본 구조**

공식 문서: [pandas Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html) · [pandas DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html)

## **1) Series**

`Series`는 pandas에서 사용하는 **1차원 자료구조**이다.

여러 개의 값이 한 방향으로 나열되어 있으며, 각 값에는 해당 값을 구분하는 **인덱스**가 붙는다. `Series` 전체를 설명하는 이름도 지정할 수 있다.

```text
인덱스     값
  0      120
  1      150
  2       90

Series 이름: 매출
```

### **Series의 구성**

| 구성 | 설명 | 예 |
|---|---|---|
| 값(values) | Series에 저장된 실제 데이터 | `120`, `150`, `90` |
| 인덱스(index) | 각 값의 위치나 이름을 나타내는 표시 | `0`, `1`, `2` |
| 이름(name) | Series 전체가 무엇을 나타내는지 설명하는 이름 | `매출` |
| 자료형(dtype) | Series에 저장된 값의 자료형 | `int64` |

Series를 만들 때 인덱스를 따로 지정하지 않으면 pandas가 `0`부터 시작하는 기본 인덱스를 자동으로 만든다.

```python
pd.Series([120, 150, 90], name="매출")
```


### **Series의 차원과 크기**

Series는 값이 한 방향으로 나열된 1차원 구조이다.

| 속성 | 의미 | 결과 예 |
|---|---|---|
| `ndim` | 데이터가 몇 차원인지 확인 | `1` |
| `shape` | 각 방향의 데이터 개수를 튜플로 확인 | `(3,)` |
| `size` | 전체 값의 개수 확인 | `3` |
| `dtype` | 저장된 값의 자료형 확인 | `int64` |



공식 문서: [pandas Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html)

````python
# [예제] Series 만들기
import pandas as pd

sales_series = pd.Series([120, 150, 90], name="매출")

print(sales_series)
print("차원:", sales_series.ndim)      # 1
print("크기:", sales_series.shape)     # (3,) → 값이 3개
````

````python
import pandas as pd


# [1. Series 객체 생성]
# 데이터 값 [120, 150, 90]에 기본 인덱스(0, 1, 2)와 시리즈 이름("매출")을 부여
sales_series = pd.Series([120, 150, 90], name="매출")

# [2. Series 출력 및 구조 확인]
print(sales_series)


# [3. .ndim 속성: 차원 확인]
print("차원:", sales_series.ndim)

# [4. .shape 속성: 형태 및 데이터 개수 확인]
print("크기:", sales_series.shape)
````

## **2) DataFrame**

`DataFrame`은 pandas에서 사용하는 **2차원 표 형태의 자료구조**이다.

행과 열로 구성되며, 각 행에는 인덱스가 있고 각 열에는 열 이름이 있다. 하나의 DataFrame에는 숫자, 문자열, 날짜·시간 등 서로 다른 자료형의 열이 함께 들어갈 수 있다.

```text
       지역    제품군    매출
0      서울     전자     120
1      부산     가구     150
2      대전     전자      90
↑       ↑       ↑       ↑
인덱스   열      열       열
```

### **DataFrame의 구성**

| 구성 | 설명 | 예 |
|---|---|---|
| 행(row) | 관측 대상이나 기록 한 건 | 서울 지역의 전자제품 매출 기록 |
| 열(column) | 같은 항목의 값이 세로로 모인 것 | 지역, 제품군, 매출 |
| 열 이름(column name) | 각 열에 저장된 값의 의미를 나타내는 이름 | `"지역"`, `"제품군"`, `"매출"` |
| 인덱스(index) | 각 행을 구분하는 번호 또는 이름 | `0`, `1`, `2` |
| 값(value) | 행과 열이 만나는 각 칸에 저장된 데이터 | `"서울"`, `"전자"`, `120` |

DataFrame을 만들 때 인덱스를 지정하지 않으면 pandas가 `0`부터 시작하는 기본 인덱스를 자동으로 만든다.

### **딕셔너리로 DataFrame 만들기**

파이썬 딕셔너리를 `pd.DataFrame()`에 전달하여 표를 만들 수 있다.

```python
pd.DataFrame({
    "지역": ["서울", "부산", "대전"],
    "제품군": ["전자", "가구", "전자"],
    "매출": [120, 150, 90]
})
```

딕셔너리를 이용할 때는 다음과 같이 표가 만들어진다.

- 딕셔너리의 키는 열 이름이 된다.
- 딕셔너리의 값인 리스트는 해당 열에 저장된다.
- 각 리스트의 같은 위치에 있는 값들이 하나의 행을 만든다.
- 모든 리스트의 길이는 같아야 한다.

```text
"지역"   → ["서울", "부산", "대전"]
"제품군" → ["전자", "가구", "전자"]
"매출"   → [120, 150, 90]
```

### **DataFrame의 차원과 크기**

DataFrame은 행과 열로 구성된 2차원 구조이다.

| 코드 | 의미 | 결과 예 |
|---|---|---|
| `df.ndim` | 데이터의 차원 수 | `2` |
| `df.shape` | 행과 열의 개수 | `(3, 3)` |
| `len(df)` | 행의 개수 | `3` |
| `len(df.columns)` | 열의 개수 | `3` |
| `df.size` | 전체 데이터 칸의 개수 | `9` |

`shape`의 첫 번째 값은 행의 개수이고 두 번째 값은 열의 개수이다.

```text
shape → (행의 개수, 열의 개수)
         3           3
```

따라서 `shape`이 `(3, 3)`이면 3행 3열인 DataFrame이다.

`size`는 행의 개수와 열의 개수를 곱한 전체 칸의 개수이다.

```text
3행 × 3열 = 9칸
```

> `len(df)`는 DataFrame의 행만 센다. 열의 개수까지 확인하려면 `df.shape` 또는 `len(df.columns)`를 사용한다.

데이터를 불러온 뒤에는 `shape`을 이용하여 행과 열의 개수를 확인한다. 이는 데이터의 전체적인 크기를 파악하는 기본 점검 과정이다.

공식 문서: [pandas DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html) · [DataFrame 생성](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html)

````python
# [예제] DataFrame 만들기
sample_df = pd.DataFrame({
    "지역": ["서울", "부산", "대전"],
    "제품군": ["전자", "가구", "전자"],
    "매출": [120, 150, 90]
})

print("차원:", sample_df.ndim)                # 2
print("크기:", sample_df.shape)               # (3, 3) → 행 3개, 열 3개
print("행의 개수:", len(sample_df))            # 3 (행만 알려 준다)
print("열의 개수:", len(sample_df.columns))    # 3
print("전체 칸 수:", sample_df.size)           # 9

sample_df
````

````python
# [실습] DataFrame
# 문제: 설비별 생산 기록을 딕셔너리로 DataFrame으로 만들고 크기·열 이름을 확인한 뒤, 열 하나를 골라 Series가 되는지 본다.

# 1. 딕셔너리 {"설비": ["M01", "M02", "M03"], "생산량": [120, 98, 130], "불량": [2, 6, 3]} 으로 DataFrame production_df 를 만든다.
production_df = pd.DataFrame({
    "설비": ["M01", "M02", "M03"],
    "생산량": [120, 98, 130],
    "불량": [2, 6, 3]
})

# 2. shape 과 열 이름 목록(list(production_df.columns))을 출력한다.
# 출력: (3, 3)
# 출력: ['설비', '생산량', '불량']
print(production_df.shape)
print(list[production_df.columns])

# 3. "생산량" 열 하나를 대괄호로 골라 자료형(type)을 출력한다. Series가 나온다.
# 출력: <class 'pandas.core.series.Series'>   (pandas 버전에 따라 <class 'pandas.Series'>)
print(production_df["생산량"])

# 4. "생산량" 열의 합계를 출력한다.
# 출력: 348
print(sum(production_df["생산량"]))
print(production_df["생산량"].sum())

# 5. 표 형태로 확인한다.
print(production_df)
````

## **Series와 DataFrame 비교**

| 구분 | `Series` | `DataFrame` |
|---|---|---|
| 차원 | 1차원 | 2차원 |

DataFrame에서 열 하나를 선택하면 일반적으로 Series가 반환되고, 여러 열을 선택하면 DataFrame이 반환된다.

### **pandas 사칙연산 메서드**

연산자(`+`, `-`)와 같은 일을 하는 메서드가 따로 있다. 보통은 연산자를 쓰고, 다른 메서드와 이어 쓸 때 메서드 형태를 쓴다.



| 연산   | 연산자  | 메서드           | 메서드 발음 | 원어           | 원어 발음    |
| ---- | ---- | ------------- | ------ | ------------ | -------- |
| 더하기  | `+`  | `.add()`      | 애드     | add          | 애드       |
| 빼기   | `-`  | `.sub()`      | 서브     | subtract     | 섭트랙트     |
| 곱하기  | `*`  | `.mul()`      | 멀      | multiply     | 멀티플라이    |
| 나누기  | `/`  | `.div()`      | 디브     | divide       | 디바이드     |
| 나누기  | `/`  | `.truediv()`  | 트루디브   | true divide  | 트루 디바이드  |
| 몫    | `//` | `.floordiv()` | 플로어디브  | floor divide | 플로어 디바이드 |
| 나머지  | `%`  | `.mod()`      | 모드     | modulo       | 모듈로      |
| 거듭제곱 | `**` | `.pow()`      | 파우     | power        | 파워       |


````python
# [예제] 연산자와 메서드는 같은 결과를 낸다
sales = pd.Series([100, 200, 300])

print(sales + 10)
print(sales.add(10))
print("------------------")
print(sales - 10)
print(sales.sub(10))
print("------------------")
print(sales * 1.1)
print(sales.mul(1.1))
print("------------------")
print(sales / 100)
print(sales.div(100))
````

````python
# [실습] Series 연산
# 문제: 생산량 Series와 불량 Series로 양품 수량과 불량률을 구하고, 연산자와 메서드 결과가 같은지 확인한다.

# 1. production = pd.Series([120, 98, 130]), defects = pd.Series([2, 6, 3]) 을 만든다.

production = pd.Series([120, 98, 130])
defects = pd.Series([2, 6, 3])
# 2. 양품 수량 = 생산량 - 불량 출력한다.
# 출력: 0    118
# 출력: 1     92
# 출력: 2    127
# 출력: dtype: int64
print(production.sub(defects))
# 3. 불량률(%) = 불량 / 생산량 * 100 을 소수점 둘째 자리까지 반올림해 출력한다. (round(2) 사용)
# 출력: 0    1.67
# 출력: 1    6.12
# 출력: 2    2.31
# 출력: dtype: float64
print(round(defects.div(production)*100,2))
````

## **3) Series 두 개를 붙여 DataFrame 만들기 (`concat`)**

DataFrame은 같은 인덱스를 가진 Series 여러 개를 열로 묶은 표다. 이미 만들어 둔 Series 두 개가 있으면 `pd.concat()`으로 옆으로 붙여 DataFrame을 만들 수 있다.

```python
pd.concat([Series1, Series2], axis=1)   # axis=1: 옆으로(열로) 붙인다
```

- 두 Series의 인덱스가 같은 행끼리 짝지어 붙는다.
- 각 Series의 이름(`name`)이 열 이름이 된다.
- 붙인 표에서 열 하나를 꺼내면 다시 Series다.


공식 문서: [pandas.concat()](https://pandas.pydata.org/docs/reference/api/pandas.concat.html)

````python
# [예제] Series 두 개를 옆으로 붙여 DataFrame 만들기
output = pd.Series([240, 300, 180], index=["A라인", "B라인", "C라인"], name="생산량")   # 라인별 생산량
defect = pd.Series([5, 9, 3], index=["A라인", "B라인", "C라인"], name="불량수")       # 라인별 불량 수

line_df = pd.concat([output, defect], axis=1)   # 같은 인덱스(라인)끼리 짝지어 옆으로 붙인다
display(line_df)                                # Series 의 name 이 열 이름이 된다

print(type(line_df["생산량"]))                  # 열 하나를 꺼내면 다시 Series
````

### **CSV 파일로 저장하기**

`to_csv()`는 DataFrame을 CSV 형식의 텍스트 파일로 저장한다.

CSV(Comma-Separated Values)는 각 값을 쉼표 등의 구분자로 나누어 저장하는 파일 형식이다.

```python
sales_data.to_csv(
    "sales.csv",
    index=False,
    encoding="utf-8-sig"
)
```
| 인코딩           | 설명                                         | 특징                                       |
| ------------- | ------------------------------------------ | ---------------------------------------- |
| **UTF-8**     | 유니코드 문자를 바이트로 표현하는 대표적인 인코딩 방식             | 대부분의 운영체제와 웹에서 사용, 모든 유니코드 문자 표현 가능      |
| **UTF-8-SIG** | UTF-8 앞에 **BOM(Byte Order Mark)** 을 추가한 방식 | 일부 Windows 프로그램에서 UTF-8 파일임을 쉽게 인식하도록 도움 |
| **CP949**     | 한글 Windows에서 많이 사용된 문자 인코딩 방식              | 한글 표현에 특화되어 있고, UTF-8보다 오래된 방식           |



| 설정 | 의미 |
|---|---|
| `"sales.csv"` | 저장할 파일의 이름 |
| `index=False` | DataFrame의 인덱스를 파일에 별도 열로 저장하지 않음 |
| `encoding="utf-8-sig"` | 한글이 올바르게 표시 |


> 파일을 읽을 때는 파일을 저장할 때 사용한 인코딩과 호환되는 인코딩을 사용해야 한다.

---

### **Excel 파일로 저장하기**

각 옵션의 의미는 다음과 같다.

---

### **`index=False`를 사용하는 이유**

pandas의 DataFrame에는 각 행을 구분하는 인덱스가 있다.

```text
      지역    매출
0     서울     120
1     부산     150
2     대전      90
↑
인덱스
```

인덱스를 별도로 지정하지 않았다면 pandas가 `0`, `1`, `2`와 같은 기본 인덱스를 자동으로 만든다.

기본 인덱스가 필요하지 않은 경우 `index=False`를 지정하면 데이터 열만 파일에 저장된다.

```python
sales_data.to_csv(
    "sales.csv",
    index=False
)
```

#### `index=True`로 저장한 경우

```text
,지역,매출
0,서울,120
1,부산,150
2,대전,90
```

#### `index=False`로 저장한 경우

```text
지역,매출
서울,120
부산,150
대전,90
```

`index=False`가 항상 필요한 것은 아니다. 날짜나 제품번호처럼 인덱스 자체가 의미 있는 정보라면 인덱스를 파일에 함께 저장할 수 있다.

### **CSV와 Excel 저장 비교**

| 구분 | CSV | Excel |
|---|---|---|
| 확장자 | `.csv` | `.xlsx` |
| 저장 메서드 | `to_csv()` | `to_excel()` |
| 파일 구조 | 구분자로 값을 나눈 텍스트 | Excel 통합 문서 형식 |
| 여러 시트 | 지원하지 않음 | 지원 가능 |

| 인코딩 지정 | `encoding` 옵션 사용 가능 | 일반적으로 직접 지정하지 않음 |
| 주요 활용 | 데이터 전달 및 분석 | 사람이 표를 확인하거나 업무 문서로 활용 |

데이터 분석에서는 구조가 단순하고 여러 프로그램에서 사용할 수 있는 CSV 파일을 자주 사용한다. Excel 파일은 여러 시트나 업무용 표가 필요한 경우에 사용할 수 있다.
```text
utf-8: 일반적인 UTF-8 인코딩
utf-8-sig: 파일 앞에 UTF-8임을 알리는 표시가 포함된 형식
일부 Excel 환경에서는 이 표시를 이용하여 UTF-8 CSV임을 더 쉽게 인식함
```

# **3. 데이터 구조 확인**


## **데이터를 미리 보는 세 가지 방법**

| 메서드 | 하는 일 | 주로 확인하는 것 |
|---|---|---|
| `head(n)` | 처음 n개 행을 보여 준다. 숫자를 생략하면 5개 | 열 구성, 한글 깨짐 여부, 값의 형태 |
| `tail(n)` | 마지막 n개 행을 보여 준다. 숫자를 생략하면 5개 | 전체 행 수, 파일 아래쪽의 합계 행이나 빈 줄 |
| `sample(n)` | 무작위로 n개 행을 뽑아 보여 준다 | 특정 부분에 치우치지 않은 실제 값 |



`sample()`은 실행할 때마다 결과가 달라진다. 같은 결과를 다시 보고 싶으면 `random_state`에 아무 숫자나 지정한다.  
  
## **random_state (난수 고정)**
random_state는 무작위(랜덤) 결과를 만들 때 사용하는 시작값이다.

## **3) 결측 비율 `isna().mean()`과 중복 확인 `duplicated()`**

결측치가 몇 개인지보다 **전체의 몇 %인지**가 더 중요할 때가 많다. 1,000행 중 3개가 비어 있는 것과 10행 중 3개가 비어 있는 것은 다르게 처리해야 하기 때문이다.

| 코드 | 하는 일 | 결과 읽기 |
|---|---|---|
| `df.isna().sum()` | 열별 빈 칸 개수 | `True`(1)를 더한다 |
| `df.isna().mean() * 100` | 열별 빈 칸 비율(%) | `True`(1)·`False`(0)의 평균 = 비율 |
| `df.duplicated().sum()` | 완전히 같은 행이 몇 개 더 있는지 | 0이면 중복 없음 |
| `df.drop_duplicates()` | 중복 행을 지우고 처음 나온 행만 남긴다 | 원본은 그대로. 다시 대입한다 |
| `df.drop_duplicates("열", keep="last")` | 해당 열 값이 같으면 마지막 행만 남긴다 | 현재 행 순서에서 마지막에 나온 행을 유지 |
| `df["열"].nunique()` | 해당 열의 서로 다른 값의 가짓수 | 행 수와 비교해 식별자가 고유한지 확인 |


.

공식 문서: [duplicated()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.duplicated.html) · [drop_duplicates()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop_duplicates.html) · [nunique()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.nunique.html)

````python
# [예제] 결측 비율과 중복 행 확인
work = pd.DataFrame({
    "작업번호": ["W001", "W002", "W003", "W004", "W002"],   # 마지막 행은 W002 를 한 번 더 입력한 것
    "라인": ["A라인", "B라인", np.nan, "C라인", "B라인"],
    "수량": [4, np.nan, 12, 3, np.nan],
    "금액": [1800000, 650000, 240000, np.nan, 650000],
})
display(work)

print(work.isna().sum())                        # 열별 빈 칸 개수
print((work.isna().mean() * 100).round(1))      # 열별 빈 칸 비율(%)

print(work.isnull().sum() )   # 위와 동일한 결과 열별 빈칸 개수
print("전체 결측 수:", work.isnull().sum().sum())	# DataFrame 전체 결측값 개수


print("중복 행 수:", work.duplicated().sum())    # 앞에 똑같은 행이 있는 행의 개수 → 1
print("작업번호 종류:", work["작업번호"].nunique(), ", 전체 행 수:", len(work))   # 4 / 5 → 겹치는 번호가 있다

work_unique = work.drop_duplicates()            # 중복 행을 지운 새 표 (원본 work 는 그대로)
print("중복 제거 후 크기:", work_unique.shape)
````

## **4) 결측 처리: `fillna()` · `dropna()`**

결측값을 찾았으면 채우거나(`fillna`) 지운다(`dropna`). 어느 쪽을 쓸지는 왜 비어 있는지 상황에 따라 정한다. 작업이 없어서 0이 맞다면 0으로 채우고, 값이 있어야 하는데 빠졌다면 중앙값 같은 대표값으로 채운다.

| 기능 | 코드 | 결과 (아래 예시 데이터) |
|---|---|---|
| 값 하나로 채우기 | `df["수량"].fillna(0)` | W002 수량 NaN → 0.0 |
| 중앙값으로 채우기 | `df["수량"].fillna(df["수량"].median())` | W002 수량 NaN → 4.0 (중앙값) |
| 열마다 다르게 채우기 | `df.fillna({"수량": 0, "라인": "미입력"})` | 수량 NaN → 0.0 · 라인 NaN → "미입력" |
| 결측 있는 행 지우기 | `df.dropna()` | 하나라도 NaN이면 삭제 → W001만 남음 |
| 특정 열 기준으로 지우기 | `df.dropna(subset=["금액"])` | 금액이 NaN인 W004만 삭제 |
| 모두 비어 있는 행만 | `df.dropna(how="all")` | 모든 값이 NaN인 행만 삭제 → 4행 그대로 |
| 값이 있는 행만 고르기 | `df[df["금액"].notna()]` | 금액이 있는 W001~W003만 남음 |

예시 데이터 (결측이 있는 4행)

| 작업번호 | 라인 | 수량 | 금액 |
|---|---|---|---|
| W001 | A라인 | 4.0 | 1800000.0 |
| W002 | B라인 | NaN | 650000.0 |
| W003 | NaN | 12.0 | 240000.0 |
| W004 | C라인 | 3.0 | NaN |

두 메서드 모두 바뀐 결과를 새로 돌려줄 뿐 원본은 그대로다. 결과를 쓰려면 다시 대입한다.

```python
df["수량"] = df["수량"].fillna(0)
df = df.dropna(subset=["금액"])
```

### 참고: 결측값 처리 시 알아둘 점

**① `fillna()` 사용 시 재대입**

특정 열의 결측값을 변경할 때는 결과를 원래 열에 다시 저장하는 방식을 사용한다.

```python
df["수량"] = df["수량"].fillna(0)
```

**② `isna()`와 `notna()`의 차이**

- `isna()`: 결측값이면 `True`
- `notna()`: 값이 존재하면 `True`

```python
df[df["수량"].notna()]  # 수량이 비어 있지 않은 행 선택
```

**③ 결측값과 자료형**

일반적인 숫자 열에 `NaN`이 포함되면 정수가 `float64`로 표현될 수 있다.

결측값을 채운 뒤 정수로 변환하려면 `astype(int)`를 사용한다.

```python
df["수량"] = df["수량"].fillna(0).astype(int)
```

단, 정수로 변환하기 전에 결측값이 모두 처리되었는지 확인한다.

공식 문서: [fillna()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.fillna.html) · [dropna()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.dropna.html) · [notna()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.notna.html)

````python
import pandas as pd
import numpy as np

# [예제] 결측값 채우기와 지우기

# 1. 결측값(NaN)이 포함된 데이터 만들기
log = pd.DataFrame({
    "작업번호": ["W001", "W002", "W003", "W004"],
    "라인": ["A라인", "B라인", np.nan, "C라인"],
    "수량": [4, np.nan, 12, 3],
    "금액": [1800000, 650000, 240000, np.nan],
})

display(log)


# 2. fillna() : 결측값 채우기
# 수량이 비어 있으면 0으로 채운다.
print(log["수량"].fillna(0).tolist())
# [4.0, 0.0, 12.0, 3.0]

# 수량이 비어 있으면 중앙값으로 채운다.
# 중앙값 : 3, 4, 12 중 가운데 값인 4
print(log["수량"].fillna(log["수량"].median()).tolist())
# [4.0, 4.0, 12.0, 3.0]

# 열마다 다른 값으로 채우기
# 수량은 0, 라인은 "미입력"으로 채운다.
display(log.fillna({"수량": 0, "라인": "미입력"}))


# 3. dropna() : 결측값이 있는 행 삭제
# 하나라도 비어 있으면 해당 행 삭제
display(log.dropna())
# W001만 남는다.

# 금액이 비어 있는 행만 삭제
display(log.dropna(subset=["금액"]))
# W004만 삭제된다.


# 4. notna() : 결측값이 아닌 데이터 확인
# 값이 있으면 True, 비어 있으면 False
print(log["금액"].notna().tolist())
# [True, True, True, False]

# 금액이 비어 있지 않은 행만 선택
display(log[log["금액"].notna()])
# W001, W002, W003만 남는다.


# 5. 결측값을 실제로 변경하고 정수로 변환하기
# 원본 log는 유지하고 복사본을 사용한다.
clean_log = log.copy()

# 수량의 결측값을 0으로 채운 뒤 정수로 변환
clean_log["수량"] = clean_log["수량"].fillna(0).astype(int)

display(clean_log)
# 수량: 4, 0, 12, 3

print(clean_log["수량"].dtype)
# int64
````

## **5) 숫자 열 요약 `describe()`**

`describe()`는 DataFrame의 숫자 열에 대해 **개수·평균·표준편차·최솟값·사분위수·최댓값**을 한 번에 보여 주는 메서드이다.

데이터를 처음 불러왔을 때 숫자의 범위가 적절한지, 이상한 값이 포함되어 있지는 않은지 확인할 때 사용한다.

```python
df.describe()
```

### **describe() 결과 이해하기**

| 행 | 뜻 |
|---|---|
| `count` | 결측값을 제외한 값의 개수 |
| `mean` | 평균 |
| `std` | 표준편차 (데이터가 평균에서 얼마나 흩어져 있는지) |
| `min` | 최솟값 |
| `25%` | 제1사분위수: 데이터의 25% 지점 |
| `50%` | 제2사분위수: 중앙값 |
| `75%` | 제3사분위수: 데이터의 75% 지점 |
| `max` | 최댓값 |

**사분위수**는 데이터를 작은 값부터 정렬했을 때 전체 데이터를 네 부분으로 나누는 기준값이다.

실제 계산에서는 데이터 사이의 값을 이용할 수 있으므로, 사분위수가 원본 데이터에 존재하지 않는 값으로 나올 수도 있다.

---

### **문자열 열요약**

숫자 열이 있는 DataFrame에서 `describe()`를 실행하면 기본적으로 숫자 열을 요약한다.

`include` 옵션으로 요약할 자료형을 선택할 수 있다.

| 사용 방법 | 설명 |
|---|---|
| `df.describe()` | 기본적으로 숫자 열 요약 |
| `df.describe(include="object")` | `object` 자료형 열 요약 |
| `df.describe(include="all")` | 모든 자료형 열 요약 |
| `df["지역"].describe()` | 특정 열만 요약 |

```python
df.describe()                   # 숫자 열 요약
df.describe(include="object")   # object 자료형 열 요약
df.describe(include="all")      # 전체 열 요약
```

문자열 열을 요약하면 다음과 같은 항목을 확인할 수 있다.

| 항목 | 뜻 |
|---|---|
| `count` | 결측값을 제외한 값의 개수 |
| `unique` | 서로 다른 값의 개수 |
| `top` | 가장 자주 등장하는 값 |
| `freq` | 가장 자주 등장하는 값의 등장 횟수 |

## **6) 자주 쓰는 요약 통계 메서드**

`describe()`는 여러 값을 한 번에 보여 주고, 값 하나만 필요할 때는 메서드를 따로 쓴다. 숫자 열과 글자(범주) 열에서 쓰는 메서드가 다르다.

| 항목 | 수치형 (int · float) | 범주형 (문자 · category) |
|---|---|---|
| 개수 | `count()` 비어 있지 않은 수 | `nunique()` 종류 수 · `unique()` 종류 목록 |
| 대표값 | `mean()` 평균 · `median()` 중앙값 | `mode()[0]` 가장 많은 값 |
| 값별 개수 | — | `value_counts()` 개수 · `value_counts(normalize=True)` 비율 |
| 최댓값 · 최솟값 | `max()` · `min()` | `value_counts().max()` 가장 많이 등장한 횟수 |
| 위치(인덱스) | `idxmax()` · `idxmin()` 최댓값·최솟값의 인덱스 | `value_counts().idxmax()` 가장 많이 등장한 값 |
| 조건에 맞는 개수 | `(df["매출"] >= 1000000).sum()` True의 개수 | `(df["지역"] == "서울").sum()` |

### **사분위수 `quantile()`**

`quantile(0.25)`는 작은 값부터 늘어놓았을 때 1/4 지점(1사분위수), `quantile(0.75)`는 3/4 지점(3사분위수)의 값이다. `describe()`의 `25%`, `75%`와 같은 값이다. `quantile(0.5)`는 중앙값 `median()`과 같다.

### **차이의 크기: 절댓값 `abs()`**

"목표와의 차이", "두 평균의 차이"처럼 크기만 중요하고 방향(+, -)은 상관없을 때 절댓값을 쓴다. 절댓값은 음수는 양수로 바꾸고 양수는 그대로 둔다.

| 코드 | 적용 대상 | 예 |
|---|---|---|
| `abs(x)` | 값 하나 (파이썬 내장 함수) | `abs(-300000)` → `300000` |
| `df["열"].abs()` | 열 전체 (pandas 메서드) | `(df["매출"] - 1000000).abs()` → 행마다 목표와의 차이 |

### **`mode()` 뒤에 `[0]`을 붙이는 이유**

가장 많이 나온 값이 여러 개일 수 있으므로 `mode()`는 값 하나가 아니라 Series를 돌려준다. 첫 번째 값만 쓰려면 `[0]`을 붙인다.

### **개수 세기: `len()` · `count()` · `sum()`의 차이**

세 가지 모두 "몇 개"를 구하지만 세는 대상이 다르다.

| 코드 | 세는 것 | 결측값 |
|---|---|---|
| `len(df)` | 전체 행 수 (`df.shape[0]`과 같다) | 포함 |
| `df["열"].count()` | 비어 있지 않은 값의 수 | 제외 |
| `(조건식).sum()` | 조건이 `True`인 행의 수 |  제외|
| `len(df[조건식])` | 조건으로 골라낸 행의 수 | `(조건식).sum()`과 같다 |



공식 문서: [Series 요약 통계 메서드](https://pandas.pydata.org/docs/reference/series.html#computations-descriptive-stats) · [quantile()](https://pandas.pydata.org/docs/reference/api/pandas.Series.quantile.html) · [mode()](https://pandas.pydata.org/docs/reference/api/pandas.Series.mode.html) · [value_counts()](https://pandas.pydata.org/docs/reference/api/pandas.Series.value_counts.html) · [Series.abs()](https://pandas.pydata.org/docs/reference/api/pandas.Series.abs.html) · [abs() 내장 함수](https://docs.python.org/ko/3.13/library/functions.html#abs)

````python
# [예제] 수치형·범주형 요약 통계

# 1. 수치형 요약 통계
print("매출 개수:", df["매출"].count())                   # 결측값 제외 개수
print("매출 중앙값:", df["매출"].median())                # 중앙값
print("매출 1사분위수:", df["매출"].quantile(0.25))       # describe()의 25%
print("매출 3사분위수:", df["매출"].quantile(0.75))       # describe()의 75%
print("매출이 가장 큰 행의 인덱스:", df["매출"].idxmax())   # 최댓값의 인덱스 라벨
print("매출 평균:", df["매출"].mean())                   # 평균

print("-" * 40)
````

````python

# 2. 범주형 요약 통계
print("지역 종류 수:", df["지역"].nunique())              # 서로 다른 지역 개수
print("지역 목록:", df["지역"].unique())                  # 중복을 제외한 지역 목록

print("지역별 주문 건수:")
print(df["지역"].value_counts())                         # 지역별 개수

print("지역별 주문 비율:")
print(df["지역"].value_counts(normalize=True))          # 지역별 비율 (0~1)

print("-" * 40)
````

````python

# 3. 조건에 맞는 주문 수
print("매출 100만 이상 주문 수:", (df["매출"] >= 1000000).sum())

print("-" * 40)

# 4. 결측값 처리에 따른 합계
print("매출 합계:", df["매출"].sum())                      # 결측값 제외
print("매출 합계(결측값 포함):", df["매출"].sum(skipna=False))
# 결측값이 있으면 skipna=False의 결과는 일반적으로 NaN

print("-" * 40)

# 5. 절댓값
print("평균과 중앙값의 차이:",
      round(abs(df["매출"].mean() - df["매출"].median())))

````

## **4) 숫자형으로 변환하기**

CSV나 Excel에서 불러온 데이터가 문자열로 저장되어 있으면 숫자 계산을 위해 자료형을 변환해야 한다.

### **① `astype()`**

`astype()`은 지정한 자료형으로 변환한 결과를 반환한다.

| 코드 | 의미 |
|---|---|
| `.astype(int)` | 정수형 변환 |
| `.astype(float)` | 실수형 변환 |
| `.astype(str)` | 문자열형 변환 |

```python
df["수량"] = df["수량"].astype(int)
```

### **② `pd.to_numeric()`**

`pd.to_numeric()`은 숫자로 변환할 때 사용하는 pandas 함수이다.

```python
pd.to_numeric(df["수량"])
```

숫자로 변환할 수 없는 값이 있으면 기본적으로 오류가 발생한다.

이런 값을 결측값으로 처리하려면 `errors="coerce"`를 지정한다.

```python
pd.to_numeric(df["수량"], errors="coerce")
```

변환에 실패한 값은 결측값으로 표시되므로 `isna()`로 확인할 수 있다.

| 구분 | `astype()` | `pd.to_numeric()` |
|---|---|---|
| 목적 | 지정한 자료형으로 변환 | 숫자형으로 변환 |
| 사용 예 | `.astype(int)` | `pd.to_numeric(s)` |
| 잘못된 숫자 | 오류 발생 가능 | `errors="coerce"`로 결측 처리 가능 |

공식 문서: [astype()](https://pandas.pydata.org/docs/reference/api/pandas.Series.astype.html) · [to_numeric()](https://pandas.pydata.org/docs/reference/api/pandas.to_numeric.html)

---


````python
import pandas as pd

# [예제 1] astype()으로 자료형 바꾸기

s = pd.Series(["10", "20", "30"])

print("변환 전:", s.dtype)

numbers = s.astype(int)

print("변환 후:", numbers.dtype)
print(numbers.tolist())  # [10, 20, 30]
print("합계:", numbers.sum())  # 60
````

````python

# [예제 2] pd.to_numeric()으로 숫자 변환

data = pd.Series(["100", "200", "입력오류", "300"])

# 숫자로 변환할 수 없는 값은 NaN 처리
result = pd.to_numeric(data, errors="coerce")

print("\n숫자 변환 결과:")
print(result)
# 0    100.0
# 1    200.0
# 2      NaN
# 3    300.0
# dtype: float64

print("결측값 개수:", result.isna().sum())  # 1
print("합계:", result.sum())  # 600.0
````

이런 값은 바로 `astype(int)`로 변환할 수 없으므로, 먼저 문자열을 정리해야 한다.

**처리 순서**

1. 현재 자료형 확인
2. `.str.replace()`로 쉼표나 단위 제거
3. `astype()` 또는 `pd.to_numeric()`으로 숫자형 변환
4. 변환 결과와 자료형 확인

```python
df["가격"] = df["가격"].str.replace(",", "", regex=False).astype(int)
```

`.str.replace()`의 `regex=False`는 정규표현식이 아니라 입력한 문자를 그대로 찾아 바꾸겠다는 뜻이다.

** 예제**

- `"1,200,000"` → `1200000`
- `"3,500원"` → `3500`

변환할 수 없는 값이 남아 있다면 임의로 숫자로 바꾸기보다 원본을 확인하거나 결측값으로 처리한다.

공식 문서: [Series.str.replace()](https://pandas.pydata.org/docs/reference/api/pandas.Series.str.replace.html)


````python


# [예제 2] 문자열에 쉼표가 있을 때

price_df = pd.DataFrame({
    "제품": ["노트북", "책상", "커피"],
    "가격": ["1,200,000", "300,000", "15,000"]
})

print("\n변환 전:")
print(price_df.dtypes)

# 쉼표 제거 후 정수형 변환
price_df["가격"] = (
    price_df["가격"]
    .str.replace(",", "")
    .astype(int)
)

print("\n변환 후:")
print(price_df.dtypes)

print("가격 합계:", price_df["가격"].sum())
# 1515000

display(price_df)

````

## **6) 날짜형으로 변환하기: `pd.to_datetime()`**

CSV에서 불러온 날짜는 문자열로 읽힐 수 있다.

날짜 범위 조회, 연도·월 추출, 기간 계산을 하려면 날짜형으로 변환해야 한다.

### **① 날짜형 변환**

`pd.to_datetime()`은 문자열 등의 데이터를 날짜·시간 자료형으로 변환하는 함수이다.

```python
df["주문일"] = pd.to_datetime(df["주문일"])
```

변환한 뒤에는 `dtype`으로 자료형을 확인할 수 있다.

### **② 변환할 수 없는 날짜 처리**

잘못된 날짜가 포함되어 있다면 `errors="coerce"`를 사용할 수 있다.

```python
df["주문일"] = pd.to_datetime(df["주문일"], errors="coerce")
```

날짜로 변환하지 못한 값은 **`NaT`(Not a Time)**로 표시된다.

| 구분 | `NaN` | `NaT` |
|---|---|---|
| 의미 | Not a Number | Not a Time |
| 용도 | 주로 숫자 데이터의 결측값 | 날짜·시간 데이터의 결측값 |
| 예 | `np.nan` | `pd.NaT` |
| 자료형 예시 | `float64` | `datetime64[ns]` |

`NaT`는 새로운 자료형이 아니라 날짜·시간 데이터에서 결측값을 나타내는 특수한 값이다. `isna()`로 확인할 수 있다.

공식 문서: [pandas.to_datetime()](https://pandas.pydata.org/docs/reference/api/pandas.to_datetime.html)

---


````python
import pandas as pd

# [예제 1] 문자열 날짜를 날짜형으로 변환

date_df = pd.DataFrame({
    "주문번호": [101, 102, 103],
    "주문일": ["2026-01-05", "2026-02-10", "2026-03-15"]
})

print("변환 전:", date_df["주문일"].dtype)

date_df["주문일"] = pd.to_datetime(date_df["주문일"])

print("변환 후:", date_df["주문일"].dtype)

display(date_df)
````

## **7) 날짜 정보 추출과 계산: `.dt`**

날짜형으로 변환한 Series에는 `.dt` 접근자를 사용하여 **연도·월·일·요일 등의 날짜 정보를 추출**할 수 있다.

또한 날짜를 비교하거나, 날짜끼리 빼서 기간을 계산하고, 특정 날짜에 일정 기간을 더할 수도 있다.

### **자주 사용하는 날짜 기능**

| 기능 | 코드 | 활용 |
|---|---|---|
| 연도 | `.dt.year` | 연도별 매출 |
| 월 | `.dt.month` | 월별 주문 |
| 일 | `.dt.day` | 일별 기록 |
| 요일 번호 | `.dt.dayofweek` | 요일별 주문 |
| 요일 이름 | `.dt.day_name()` | 요일 이름 표시 |
| 시·분 | `.dt.hour`, `.dt.minute` | 작업 시작 시각 |
| 월 단위 기간 | `.dt.to_period("M")` | 월별 데이터 구분 |
| 날짜 비교 | `df["주문일"] >= "2026-04-01"` | 특정 날짜 이후 주문 조회 |
| 날짜 차이 | `배송일 - 주문일` | 배송 기간 계산 |
| 기간의 일수 | `.dt.days` | 배송 소요일 |
| 날짜 더하기 | `주문일 + pd.Timedelta(days=7)` | 7일 후 날짜 계산 |
| 날짜 표시 형식 | `.dt.strftime("%Y-%m-%d")` | 보고서 날짜 형식 |

### **① 연도·월·일·요일 추출**

`.dt.year`, `.dt.month`, `.dt.day`를 사용하면 날짜에서 필요한 정보를 추출할 수 있다.

```python
df["주문일"].dt.year       # 연도
df["주문일"].dt.month      # 월
df["주문일"].dt.day        # 일
df["주문일"].dt.dayofweek  # 요일 번호
```

요일 번호는 **월요일이 `0`, 일요일이 `6`**이다.

| 월 | 화 | 수 | 목 | 금 | 토 | 일 |
|---|---|---|---|---|---|---|
| 0 | 1 | 2 | 3 | 4 | 5 | 6 |

따라서 요일 번호가 `5` 이상이면 토요일 또는 일요일이다.

요일 이름이 필요하면 `.dt.day_name()`을 사용한다.

```python
df["주문일"].dt.day_name()
```

`.dt.year`, `.dt.month`, `.dt.day`, `.dt.dayofweek`는 괄호 없이 사용하는 속성이고, `.dt.day_name()`은 괄호를 붙여 호출하는 메서드이다.

---

### **② 월 단위로 변환하기: `to_period()`**

`.dt.to_period("M")`은 날짜를 **연도와 월 단위의 기간**으로 변환한다.

```python
df["주문일"].dt.to_period("M")
```

예를 들어 `2026-01-05`, `2026-01-12`는 모두 `2026-01`로 표현된다.

`"M"`은 월 단위를 뜻한다.

이후 `value_counts()`로 월별 주문 건수를 세거나, `groupby()`의 기준으로 사용하여 월별 매출을 집계할 수 있다.

### **③ 날짜 비교하기**

날짜형으로 변환한 열은 날짜와 비교하여 조건에 맞는 데이터를 조회할 수 있다.

```python
df[df["주문일"] >= "2026-04-01"]
```

`2026-04-01` 이상인 주문일을 가진 행만 선택한다.

여기서 `"2026-04-01"`은 문자열이지만, pandas는 날짜형 Series와 비교할 때 유효한 날짜 문자열을 날짜로 해석하여 비교할 수 있다.

### **④ 날짜 차이 계산: `Timedelta`**

`Timedelta`는 **두 날짜 또는 시간 사이의 차이(기간)를 나타내는 자료형**이다.

날짜형 데이터끼리 빼면 그 차이가 `Timedelta`로 계산된다.

```python
df["주문일"] - df["주문일"].min()
```

예를 들어 첫 주문일이 `2026-01-05`이고 다른 주문일이 `2026-01-12`라면 결과는 `7 days`이다.

| 구분 | 의미 | 예 |
|---|---|---|
| `Timestamp` | 특정 날짜와 시간 | 2026-01-05 |
| `Timedelta` | 날짜·시간 사이의 기간 | 7 days |

기간에서 일수 부분을 추출할 때는 `.dt.days`를 사용한다.

```python
gap = df["주문일"] - df["주문일"].min()

gap.dt.days
```

배송일과 주문일의 차이를 계산할 때도 같은 방법을 사용한다.

```python
df["배송일"] - df["주문일"]
```

단, 위 코드는 `df`에 날짜형 `"배송일"` 열이 있을 때 사용할 수 있다.

---

### **⑤ 날짜 더하기: `pd.Timedelta()`**

특정 날짜에 일정 기간을 더하려면 `pd.Timedelta()`를 사용한다.

```python
df["주문일"] + pd.Timedelta(days=7)
```

각 주문일에서 7일 후의 날짜를 계산한다.

| 코드 | 의미 |
|---|---|
| `pd.Timedelta(days=7)` | 7일 |
| `pd.Timedelta(weeks=1)` | 1주 |
| `pd.Timedelta(hours=2)` | 2시간 |
| `pd.Timedelta(minutes=30)` | 30분 |

**참고: 'N일 후'와 'N일째'의 차이**

시작일을 1일째로 세는 경우, 7일째는 시작일에서 6일을 더한 날짜이다.

```text
시작일: 2026-01-05 (1일째)

+ pd.Timedelta(days=6) → 2026-01-11 (7일째)
+ pd.Timedelta(days=7) → 2026-01-12 (7일 후)
```

---

### **⑥ 날짜 표시 형식 바꾸기: `strftime()`**

`.dt.strftime()`은 날짜를 원하는 형식의 **문자열로 변환**하는 메서드이다.

```python
df["주문일"].dt.strftime("%Y-%m-%d")
```

날짜를 `2026-01-05`와 같은 문자열 형식으로 표시한다.

| 기호 | 의미 | 예 |
|---|---|---|
| `%Y` | 네 자리 연도 | 2026 |
| `%m` | 두 자리 월 | 01 |
| `%d` | 두 자리 일 | 05 |
| `%H` | 시(24시간) | 13 |
| `%M` | 분 | 30 |
| `%S` | 초 | 45 |

예를 들어 다음처럼 사용할 수 있다.

```python
df["주문일"].dt.strftime("%Y/%m/%d")
```

**주의:** `%m`은 월, `%M`은 분이므로 대소문자를 구분해야 한다.

또한 `strftime()`의 결과는 날짜형이 아니라 문자열이다. 날짜 계산은 원래 날짜형 데이터를 사용한다.

### **⑦ 심화 참고: `.dt.seconds`와 `.dt.total_seconds()`의 차이**

작업 시간이나 배송 기간을 초·분·시간 단위로 계산할 때 사용할 수 있다.

| 기능 | 의미 |
|---|---|
| `.dt.days` | 기간의 일수 부분 |
| `.dt.seconds` | 일수 부분을 제외한 나머지 초 |
| `.dt.total_seconds()` | 전체 기간을 초 단위로 계산 |

예를 들어 작업 시간이 **1일 1시간 30분**이라면 다음과 같다.

```text
기간: 1일 1시간 30분

.dt.days                 → 1
.dt.seconds              → 5400
.dt.total_seconds()      → 91800.0
.dt.total_seconds() / 60 → 1530.0
```

`.dt.seconds`는 전체 초가 아니라 일수 부분을 제외한 나머지 초이므로, 기간이 하루 이상일 때 주의해야 한다.

**전체 작업 시간을 분이나 시간으로 변환하려면 `.dt.total_seconds()`를 사용한다.**

```python
job["작업시간"].dt.total_seconds() / 60    # 전체 분
job["작업시간"].dt.total_seconds() / 3600  # 전체 시간
```

### **10. 날짜 차이와 전체 시간 계산**
