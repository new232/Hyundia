# **`1 range()`: 정해진 횟수만큼 반복하기**

리스트와 같은 데이터가 없더라도 "지정한 횟수만큼 코드를 되풀이"하고 싶을 때 `range()` 함수를 사용한다.

`range(5)`는 `0`부터 시작하여 `5` 전까지, 즉 `0, 1, 2, 3, 4` 총 **5개의 정수를 연속으로 생성**한다. **(지정한 끝 숫자인 5는 포함되지 않는다.)**

---

###  **`range()`의 3가지 활용 패턴**

| 구분 | 형태 | 생성되는 숫자 목록 | 설명 |
|---|---|---|---|
| **1. 횟수 지정** | `range(5)` | `0, 1, 2, 3, 4` | **0부터 4까지** 총 5번 반복 |
| **2. 범위 지정** | `range(1, 6)` | `1, 2, 3, 4, 5` | **1부터 5까지** 1씩 증가 |
| **3. 간격 지정** | `range(1, 10, 2)` | `1, 3, 5, 7, 9` | **1부터 9까지** 2씩 건너뜀 (홀수만) |

>끝에 적은 숫자(5, 6, 10)는 포함되지 않는다(끝 숫자 - 1까지 생성)는 점만 주의하면 된다.

공식 문서: [range() 함수 개요](https://docs.python.org/ko/3.13/library/stdtypes.html#range)

````python
# 1. range(끝): 0부터 시작하여 지정한 횟수만큼 반복 (0 ~ 4)
print("--- 1. range(5) ---")
for i in range(5):
    print(i, end=" ")
print("\n")

# 2. range(시작, 끝): 시작 숫자부터 (끝 - 1)까지 반복 (1 ~ 5)
print("--- 2. range(1, 6) ---")
for i in range(1, 6):
    print(i, end=" ")
print("\n")

# 3. range(시작, 끝, 증가폭): 간격을 두고 숫자 생성 (1 ~ 9 홀수만)
print("--- 3. range(1, 10, 2) ---")
for i in range(1, 10, 2):
    print(i, end=" ")
````

````python
# [실습] range()
# 문제: range(3)으로 "1회차 점검"~"3회차 점검" 을 출력하고, range(1, 6)으로 1~5의 합계를 total에 누적해 출력하고, range(0, 60, 15)로 "0분 측정"~"45분 측정" 을 출력한다.
for i in range(3):
    print(f"{i+1}회차 점검")
total = 0
for j in range(1,6):
    total += j
print(total)
````

# **2 리스트 컴프리헨션: 반복문을 한 줄로 줄여 쓰기**

위에서 문자열을 숫자로 바꿀 때 "빈 리스트 만들기 → `for`로 하나씩 꺼내기 → `int()`로 바꾸기 → `append()`로 담기"의 4단계를 거쳤다. 이 네 줄을 대괄호 `[]` 안에 한 줄로 적는 방법이 **리스트 컴프리헨션**(list comprehension)이다.

```text
 반복문 방식                          한 줄 방식
 sales_numbers = []                  sales_numbers = [int(v) for v in sales]
 for v in sales:                                      │      └─ 1. sales에서 v를 하나씩 꺼내
     sales_numbers.append(int(v))                     └─ 2. int(v)로 바꿔 새 리스트에 담는다
```

읽는 순서는 오른쪽의 `for v in sales`가 먼저(하나씩 꺼내기), 왼쪽의 `int(v)`가 나중(바꿔서 담기)이다. 뒤에 `if 조건`을 붙이면 조건에 맞는 값만 담는다.

```python
[int(v) for v in sales]                      # 모두 숫자로
[int(v) for v in sales if int(v) >= 150000]  # 15만 원 이상만 숫자로
```

결과는 반복문 방식과 같다.

| 종류 | 형태 | 예 |
|---|---|---|
| 리스트 컴프리헨션 | `[식 for 변수 in 대상]` | `[int(v) for v in sales]` |
| 딕셔너리 컴프리헨션 | `{키: 값 for 변수 in 대상}` | `{v: int(v) for v in sales}` |
| 집합 컴프리헨션 | `{식 for 변수 in 대상}` | `{int(v) for v in sales}` |

컴프리헨션은 대괄호 []를 사용하면 리스트, 중괄호 {} 안에 표현식을 작성하면 집합, 키: 값 형태로 작성하면 딕셔너리를 생성한다.



공식 문서: [리스트 컴프리헨션](https://docs.python.org/ko/3.13/tutorial/datastructures.html#list-comprehensions) · [집합](https://docs.python.org/ko/3.13/tutorial/datastructures.html#sets) · [딕셔너리](https://docs.python.org/ko/3.13/tutorial/datastructures.html#dictionaries)


````python
# [참고] 리스트 컴프리헨션
# 위 셀에서 문자열 형태의 숫자들을 정수로 바꾸기 위해 **빈 리스트 생성 ➔ `for` 문 반복 ➔ `int()` 변환 ➔ `append()`**의 4단계를 거친다
# 이 과정을 대괄호 `[]` 안에서 한 줄로 작성하는 '리스트 컴프리헨션' 문법을 제공
# 핵심 구조: [ 변환할_표현식  for  변수  in  리스트 ]


# [리스트 컴프리헨션 구조]  리스트의 문자열 요소(v)를 하나씩 꺼내 정수(int)로 변환 후 새로운 리스트에 담음
# [   int(v)   for v in sales   ]
#    └─ 2. 정수로 변환   └─ 1. sales에서 문자열 v를 하나씩 꺼냄
sales_numbers = [int(v) for v in sales]

print("변환된 숫자 리스트:", sales_numbers)  # [100000, 150000, 90000] (정수형 리스트)



#  조건문(if) 필터링까지 한 줄로 결합
# 15만 원 이상인 매출만 골라내어 정수로 변환
high_sales = [int(v) for v in sales if int(v) >= 150000]

print("15만 원 이상 매출만 추출:", high_sales) # [150000]
````

````python
# [예제] 괄호만 바꾸면 리스트·집합·딕셔너리 컴프리헨션
scores = [70, 85, 90, 85]

print([s for s in scores if s >= 80])         # [85, 90, 85]  리스트: 중복이 그대로 남는다
print({s for s in scores if s >= 80})         # 85, 90 두 개  집합: 중복이 사라진다 (출력 순서는 다를 수 있다)
print({s: s + 5 for s in scores if s >= 80})  # {85: 90, 90: 95}  딕셔너리: 같은 키는 하나로 합쳐진다
````

````python
# [실습] 집합 컴프리헨션
# 문제: 숫자 목록 numbers = [10, 20, 10, 30, 20]에서
#       집합 컴프리헨션을 사용하여 중복을 제거한 집합을 생성하시오.
#       생성한 집합을 result에 저장하고,
#       sorted()를 사용하여 오름차순으로 정렬한 결과를 출력하시오.

# [실행 결과]
# [10, 20, 30]

numbers = [10, 20, 10, 30, 20]
result = {v for v in numbers}
print(sorted(result))
````

````python
# [실습] 집합 컴프리헨션과 if 조건
# 문제: 숫자 목록 numbers = [10, 20, 30, 20, 40, 10]에서
#       집합 컴프리헨션과 if 조건을 사용하여
#       20 이상인 숫자만 중복 없이 모아 result에 저장하시오.
#       결과는 sorted()를 사용하여 오름차순으로 출력하시오.

# [실행 결과]
# [20, 30, 40]

numbers = [10, 20, 30, 20, 40, 10]
result = {v for v in numbers if v>= 20 }
print(sorted(result))
````

````python
# [실습] 집합 컴프리헨션
# 문제: 대소문자와 공백이 섞인 설비 코드 목록을 공백 제거·대문자로 정리해 중복 없는 코드 집합을 만들고, 정렬한 목록과 종류 수를 출력한다.
#  설비 코드 codes = ["m01", "M01 ", " m02", "M03", "m02"]
codes = ["m01", "M01 ", " m02", "M03", "m02"]

codes_set = {v.strip().upper() for v in codes }
print(sorted(codes_set))
````

````python
# [실습] 집합 컴프리헨션
# 문제: 대소문자와 공백이 섞인 설비 코드 목록을 공백 제거·대문자로 정리해 중복 없는 코드 집합을 만들고, 정렬한 목록과 종류 수를 출력한다.
#  설비 코드 codes = ["m01", "M01 ", " m02", "M03", "m02"]
codes = ["m01", "M01 ", " m02", "M03", "m02"]

codes_set = set(map(lambda x:x.strip().upper(),codes))
print(sorted(codes_set))
````

# **3 딕셔너리 반복하기 (`items()`)**

딕셔너리는 단순 리스트와 달리 **키(Key)와 값(Value)이 한 쌍**으로 이루어져 있다.

딕셔너리의 `.items()` 메서드를 `for` 문과 함께 사용하면 **키(Key)와 값(Value)을 한 번에 하나씩 짝지어 꺼내올 수 있다.**

---

### **`items()` 반복문의 구조와 동작 원리**

```text
 for region, sales in region_sales.items():
      │        │       └─ 1. (키, 값) 튜플 짝을 생성
      │        └─ 3. 매출 금액(Value)이 담김
      └─ 2. 지역 이름(Key)이 담김

```
---

### **[참고] 딕셔너리 반복 순회 3가지 패턴 비교**

| 구문 | 꺼내오는 대상 | 예시 출력 (`region_sales`) |
|---|---|---|
| `for k in region_sales:` | **Key만** | `"서울"`, `"부산"`, `"대구"` |
| `for v in region_sales.values():` | **Value만** | `350000`, `280000`, `210000` |
| `for k, v in region_sales.items():` | **Key, Value 둘 다** | `"서울" 350000` ... |

---

````python
# [실습 1] items()를 활용한 키-값 동시 출력
region_sales = {
    "서울": 350000,
    "부산": 280000,
    "대구": 210000
}

# region에는 Key("서울"), sales에는 Value(350000)가 순서대로 담김
for region, sales in region_sales.items():
    print(region, sales)
    print(region_sales[region])
    print('\n')

````

````python
# [실습] 딕셔너리 반복하기
# 문제: 제품별 재고 딕셔너리(stock)에서
#       for문과 if문을 사용하여 재고가 0인 제품만
#       "키보드 품절" 형식으로 출력하시오.
#       단, items()를 사용하여 제품명과 재고 수량을 함께 꺼내시오.

stock = {"노트북": 10, "마우스": 45, "키보드": 0}

# [실행 결과]
# 키보드 품절
for key, value in stock.items():
    if value == 0:
        print(f"{key} 품절")
````

# **4 딕셔너리로 개수·합계 세기: `get(키, 0)`**

반복문으로 데이터를 하나씩 처리하면서 지역별 건수, 제품별 판매량, 제품별 매출 합계처럼 같은 키에 해당하는 값을 누적할 때 딕셔너리를 사용할 수 있다.

처음 등장한 키는 딕셔너리에 존재하지 않으므로 대괄호 []로 조회하면 KeyError가 발생한다.

이때 get(키, 0)을 사용하면 키가 없을 때 0을 반환하도록 설정할 수 있다.

```python
counts[키] = counts.get(키, 0) + 1     # 개수 세기
totals[키] = totals.get(키, 0) + 값    # 합계 내기
```

```text
 regions = ["서울", "부산", "서울"]

 1번째 "서울"   counts.get("서울", 0) → 0   0 + 1 = 1   {'서울': 1}
 2번째 "부산"   counts.get("부산", 0) → 0   0 + 1 = 1   {'서울': 1, '부산': 1}
 3번째 "서울"   counts.get("서울", 0) → 1   1 + 1 = 2   {'서울': 2, '부산': 1}
```

키가 처음 나오면 0에서 시작하고, 이미 있으면 기존 값에 더한다. 뒤에서 배울 pandas의 `value_counts()`, `groupby()`가 이 일을 한 줄로 해 준다.

공식 문서: [dict.get()](https://docs.python.org/ko/3.13/library/stdtypes.html#dict.get)


````python
# [예제] 지역별 주문 건수 세기
regions = ["서울", "부산", "서울", "대구", "서울"]

counts = {}                                     # 빈 딕셔너리에서 시작한다
for region in regions:
    counts[region] = counts.get(region, 0) + 1  # 없는 키는 0에서 시작해 1을 더한다

print(counts)   # {'서울': 3, '부산': 1, '대구': 1}

# 참고: counts[region] = counts[region] + 1 로 쓰면 처음 나온 키에서 KeyError가 난다
````

````python
# [실습] 딕셔너리로 제품별 판매 건수 세기
# 문제: 판매된 제품 목록 products = ["노트북", "마우스", "노트북", "키보드", "마우스", "노트북"]에서
#       for문과 get(키, 0)을 사용하여 제품별 판매 건수를 계산하시오.
#       빈 딕셔너리 counts에서 시작하여 제품별 건수를 누적하고 결과를 출력하시오.

# [실행 결과]
counts = {}
products = ["노트북", "마우스", "노트북", "키보드", "마우스", "노트북"]
for key in products:
    counts[key] = counts.get(key, 0) + 1
print(counts)
    
# {'노트북': 3, '마우스': 2, '키보드': 1}
````

````python
# [실습] 딕셔너리로 제품별 매출 합계 계산
# 문제: 제품별 판매 기록을 확인하여 같은 제품의 매출을 합산하시오.
#       for문과 get(키, 0)을 사용하여 제품별 매출 합계를 totals에 저장하고 출력하시오.

products = ["노트북", "마우스", "노트북", "키보드"]
sales = [100000, 20000, 150000, 50000]
totals = {}
for i,v  in zip(products,sales):
    totals[i]= totals.get(i,0) + v
print(totals)
# {'노트북': 250000, '마우스': 20000, '키보드': 50000}
````

# **`5 enumerate()`: 순번(인덱스)과 값을 함께 꺼내기**

반복문(`for`)을 돌릴 때 리스트의 데이터뿐만 아니라 "몇 번째 데이터인지 순번(인덱스 번호)"이 함께 필요할 때 `enumerate()` 함수를 사용한다.

파이썬의 인덱스는 기본적으로 **`0`부터 시작**한다.

---

### **`enumerate()`의 구조와 동작 원리**

```text
 for index, value in enumerate(sales):
      │       │                └─ 1. 순번과 값을 짝지어 생성
      │       └─ 3. 데이터 값(Value)이 담김
      └─ 2. 위치 번호(Index, 0부터 시작)가 담김
```
---

###  **[참고] 시작 번호 변경 옵션 (`start=1`)**

| 구문 | 인덱스 시작값 | 출력 예시 (`sales = [120000, ...]`) |
|---|---|---|
| `enumerate(sales)` | **0부터 시작** | `0 120000`, `1 150000` ... |
| `enumerate(sales, start=1)` | **1부터 시작** | `1 120000`, `2 150000` ... |

---

````python
# [실습 1] 기본 enumerate() 사용 (0부터 시작하는 인덱스)
sales = [120000, 150000, 90000]

for index, value in enumerate(sales):
    print(index, value)

print("-" * 40)

# [실습 2] start=1 옵션을 활용하여 보기 쉬운 순번 출력
for i, value in enumerate(sales, start=1):
    print(f"{i}번째 건 매출액: {value:,}원")
````

````python
# [실습] enumerate()를 활용한 제품 번호 출력
# 문제: 제품 목록 products = ["노트북", "마우스", "키보드", "모니터"]에서
#       enumerate()를 사용하여 순번을 1부터 시작하고,
#       각 제품의 순번과 제품명을 출력하시오.

# [실행 결과]
# 1번 제품: 노트북
# 2번 제품: 마우스
# 3번 제품: 키보드
# 4번 제품: 모니터


products = ["노트북", "마우스", "키보드", "모니터"]

for i,v in enumerate(products,1):
    print(f"{i}번 제품 {v}")
````

````python
# [실습] enumerate()
# 문제: 일별 생산량 production = [120, 135, 98, 142]에서
#       enumerate()와 if문을 사용하여 생산량이 100 미만인 날의
#       순번(1부터 시작)과 생산량을 출력하시오.

# [실행 결과]
# 3일차 생산량 부족: 98
production = [120, 135, 98, 142]

for i,v in enumerate(production,1):
    if v<100:
        print(f"{i}  {v}")
````

# **6 `zip()`: 여러 리스트를 같은 위치끼리 묶어서 처리하기**

서로 연관된 데이터를 담고 있는 **두 개 이상의 리스트를 같은 인덱스(위치)끼리 1:1로 짝지어 동시에 순회**할 때 `zip()` 함수를 사용한다.

---

###  **`zip()`의 구조와 동작 원리**
for 변수1, 변수2 in zip(리스트1, 리스트2): ➔ 리스트의 개수만큼 변수를 쉼표로 구별하여 지정한다.
```text
 regions = ["서울", "부산", "대구"]
 sales   = [350000, 280000, 210000]

 for region, value in zip(regions, sales):
      │        │           └─ 1. 같은 인덱스의 요소끼리 (서울, 350000) 짝지어 전달
      │        └─ 3. sales 리스트의 값이 담김 (350000)
      └─ 2. regions 리스트의 값이 담김 ("서울")
```
---

zip()은 여러 반복 가능한 객체의 같은 위치에 있는 요소들을 튜플로 묶어 차례대로 제공하는 반복자(iterator)를 반환한다. for문에서는 각 튜플을 언패킹하여 여러 변수로 받을 수 있다.
###  **[참고] 두 리스트의 길이가 다를 때 주의점**

| 상황 | 동작 방식 | 예시 |
|---|---|---|
| **길이가 같을 때** | 모든 요소가 1:1로 정확히 맞물림 | 3개 & 3개 ➔ 3회 반복 |
| **길이가 다를 때** | **가장 짧은 리스트의 길이에 맞춰** 남는 데이터는 버려짐 | 3개 & 5개 ➔ 3회만 반복 |

---

````python
# [실습 1] 기본 zip() 활용: 두 리스트 병렬 순회

regions = ["서울", "부산", "대구"]
sales = [350000, 280000, 210000]



# 1. zip() 결과 확인
# zip()은 반복자이므로 객체 자체를 출력하면 데이터가 직접 표시되지 않는다.
# 모든 튜플을 리스트 형태로 확인하려면 list()로 변환한다.

print(zip(regions, sales))
# <zip object at 0x...>  ← 객체 정보 출력 (주소는 실행마다 다를 수 있음)

print(list(zip(regions, sales)))
# [('서울', 350000), ('부산', 280000), ('대구', 210000)]

print("-" * 40)


# 2. for문으로 두 리스트를 동시에 순회
# zip()이 반환한 튜플을 region과 value에 각각 나누어 저장한다.

for region, value in zip(regions, sales):
    print(region, value)

print("-" * 40)


# [실습 2] zip()을 활용한 딕셔너리 변환
# zip()으로 만든 (키, 값) 쌍을 dict()에 전달하면 딕셔너리가 생성된다.

region_sales_dict = dict(zip(regions, sales))

print("딕셔너리 변환 결과:", region_sales_dict)
# {'서울': 350000, '부산': 280000, '대구': 210000}
````

````python
# [실습] zip()을 활용한 제품별 주문 금액 계산
# 문제: 제품명, 주문 수량, 단가가 각각 리스트로 주어져 있다.
#       zip()을 사용하여 제품별 주문 금액을 계산하고,
#       전체 주문 금액을 출력하시오.
#       또한 제품명과 주문 수량을 딕셔너리로 만들어 출력하시오.

products = ["노트북", "마우스", "키보드"]
quantities = [2, 10, 5]
prices = [1200000, 25000, 45000]

# [실행 결과]
# 노트북: 2개 × 1,200,000원 = 2,400,000원
# 마우스: 10개 × 25,000원 = 250,000원
# 키보드: 5개 × 45,000원 = 225,000원
# 전체 주문 금액: 2,875,000원
# {'노트북': 2, '마우스': 10, '키보드': 5}
total= 0
for i,j,z in zip(products,quantities,prices):
    print(f"{i}: {j}개 x {z:,.0f}원 = {j*z:,.0f}원")
    total += j*z

print(f"전체 주문 금액: {total:,.0f}원")
````

### **1. range()는 어떤 역할을 하며, 시작값과 끝값을 지정할 때 주의할 점은 무엇인가?**

`range()`는 일정한 규칙에 따라 정수의 시퀀스를 나타내는 `range` 객체를 생성하는 내장 함수이다.

반복문에서 정해진 횟수만큼 반복하거나 일정 범위의 숫자를 사용할 때 활용한다.

```python
for number in range(1, 5):
    print(number)

# 1
# 2
# 3
# 4
```

`range(시작, 끝)`에서 시작값은 포함하지만 **끝값은 포함하지 않는다.** 증가 간격을 생략하면 기본값은 1이다.

### **2. 리스트 컴프리헨션은 무엇이며, 일반적인 for문과 비교하여 어떤 특징이 있는가?**

리스트 컴프리헨션(List Comprehension)은 반복문을 사용하여 **새로운 리스트를 간결하게 생성하는 문법**이다.

일반적인 `for`문과 `append()`를 이용한 리스트 생성 과정을 한 줄로 작성할 수 있다.

```python
sales = ["100000", "150000", "90000"]

# 일반적인 for문
sales_numbers = []

for value in sales:
    sales_numbers.append(int(value))

print(sales_numbers)
# [100000, 150000, 90000]

# 리스트 컴프리헨션
sales_numbers = [int(value) for value in sales]

print(sales_numbers)
# [100000, 150000, 90000]
```

리스트 컴프리헨션은 `if` 조건을 추가하여 조건에 맞는 요소만 선택할 수도 있다.

---

### **3. 딕셔너리의 items()를 for문과 함께 사용하면 어떤 장점이 있는가?**

`items()`는 딕셔너리의 **키와 값을 `(키, 값)` 형태로 제공하는 뷰 객체**를 반환한다.

`for`문에서 언패킹을 사용하면 키와 값을 동시에 꺼내 처리할 수 있다.

```python
region_sales = {"서울": 350000, "부산": 280000}

for region, sales in region_sales.items():
    print(f"{region}: {sales:,}원")

# 서울: 350,000원
# 부산: 280,000원
```

이미 키와 값을 각각의 변수로 받았으므로 키를 사용하여 값을 다시 조회할 필요가 없다.

---

### **4. 딕셔너리의 get(키, 0)을 사용하여 데이터의 개수나 합계를 누적하는 원리는 무엇인가?**

`get(키, 0)`은 딕셔너리에 해당 키가 없으면 `0`을 반환하고, 키가 존재하면 기존 값을 반환한다.

이를 반복문과 함께 사용하면 같은 키에 해당하는 데이터의 개수나 합계를 누적할 수 있다.

```python
regions = ["서울", "부산", "서울", "대구", "서울"]

counts = {}

for region in regions:
    counts[region] = counts.get(region, 0) + 1

print(counts)
# {'서울': 3, '부산': 1, '대구': 1}
```

처음 등장한 지역은 0에서 시작해 1을 더하고, 이미 등장한 지역은 기존 개수에 1을 더한다.

합계를 계산할 때는 `+ 1` 대신 누적할 값을 더한다.

---

### **5. enumerate()는 언제 사용하며, 순번의 시작값은 어떻게 변경하는가?**

`enumerate()`는 반복 가능한 객체에서 **순번과 요소를 함께 가져올 때** 사용하는 내장 함수이다.

기본 순번은 `0`부터 시작하며, `start` 인자를 사용하면 시작값을 변경할 수 있다.

```python
products = ["노트북", "마우스", "키보드"]

for number, product in enumerate(products, start=1):
    print(f"{number}번 제품: {product}")

# 1번 제품: 노트북
# 2번 제품: 마우스
# 3번 제품: 키보드
```

`enumerate()`는 순번과 요소를 담은 튜플을 차례대로 제공하는 반복자 객체를 반환한다.

---

### **6. zip()은 어떤 객체를 반환하며, 여러 리스트의 길이가 다르면 어떻게 동작하는가?**

`zip()`은 여러 반복 가능한 객체의 같은 위치에 있는 요소들을 튜플로 묶어 차례대로 제공하는 **반복자(Iterator) 객체**를 반환한다.

```python
products = ["노트북", "마우스", "키보드"]
quantities = [2, 10, 5]

for product, quantity in zip(products, quantities):
    print(f"{product}: {quantity}개")

# 노트북: 2개
# 마우스: 10개
# 키보드: 5개
```

여러 리스트의 길이가 다르면 기본적으로 **가장 짧은 리스트의 길이까지만** 처리하며 나머지 요소는 묶이지 않는다.

또한 `dict(zip())`을 활용하면 키 목록과 값 목록을 결합하여 딕셔너리를 생성할 수 있다.

```python
product_dict = dict(zip(products, quantities))

print(product_dict)
# {'노트북': 2, '마우스': 10, '키보드': 5}
```
