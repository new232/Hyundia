## **변수 이름을 짓는 규칙**

변수 이름은 아무렇게나 지을 수 없다. 아래 규칙을 어기면 오류가 난다.

- 문자와 숫자, 밑줄(`_`)을 쓸 수 있다
- 숫자로 시작할 수 없다 (`1st_sales` 불가, `sales1`은 가능)
- 공백과 특수문자는 쓸 수 없다 (`my sales`, `a@b` 불가)
- 대문자와 소문자를 서로 다른 것으로 구분한다 (`sales`, `Sales`, `SALES`는 각각 다른 변수)
- 키워드는 이름으로 쓸 수 없다

파이썬 에서는 `총매출 = 100`처럼 한글로도 변수 이름을 지을 수 있다. 다만 영문으로 작성하는 것을 권장하므로, 이 과정에서도 영문 이름을 사용한다.

키워드는 파이썬이 문법으로 이미 정해 놓은 단어다. `if`, `for`, `True`, `False`, `None`, `and`, `or`, `import` 등이 있다. 참과 거짓을 뜻하는 `True`, `False`는 **첫 글자가 대문자**다. 소문자 `true`, `false`는 키워드가 아니어서 변수 이름으로 쓸 수는 있지만, 혼동을 피하기 위해 쓰지 않는다.

```text
 sales = 100      # 가능
 1st_sales = 100  # 오류: 숫자로 시작
 my sales = 100   # 오류: 공백 포함
 if = 100         # 오류: 키워드 사용
```

공식 문서: https://docs.python.org/ko/3.13/reference/lexical_analysis.html#keywords

````python
sales = 120000
order_count = 5
average_sales = sales / order_count

print(sales)
print(order_count)
print(average_sales)
````

## **f-string**

f-string은 문자열 안에 변수 값을 넣어 출력하는 방법이다. 따옴표 앞에 `f`를 붙이고, 값을 넣을 자리를 중괄호 `{}`로 표시한다.



```python
f"문자열 {변수명} 문자열"
```

중괄호 안에는 변수뿐 아니라 계산식도 넣을 수 있다.

```python
region = "서울"
sales = 1234567

print(f"{region} 지역 매출은 {sales}원이다.")
print(f"주문당 평균은 {sales / 5}원이다.")
```

공식 문서: [f-string](https://docs.python.org/ko/3.13/reference/lexical_analysis.html#f-strings)

### 서식 지정자로 숫자 정리하기

통계 결과는 자릿수가 길거나 소수점이 많아 그대로 출력하면 읽기 어렵다. 중괄호 안에서 변수 이름 뒤에 콜론(`:`)을 붙이면 출력할 모양을 지정할 수 있다.

```python
f"{변수명:서식}"
```

콜론은 여기부터 출력 모양을 지정한다는 표시다.

**천 단위 쉼표**

```python
sales = 1234567

print(f"{sales}")     # 1234567
print(f"{sales:,}")   # 1,234,567
```

**소수점 자릿수**

`f`는 실수 형태로 출력한다는 뜻이고, 앞의 숫자는 표시할 소수점 자릿수다. 지정한 자리 아래는 반올림된다.

```python
average_sales = 123456.789

print(f"{average_sales:.2f}")   # 123456.79
print(f"{average_sales:.0f}")   # 123457
```

**두 서식을 함께 쓰기**

천 단위 쉼표를 먼저 쓰고 소수점 지정을 뒤에 쓴다.

```python
print(f"{average_sales:,.2f}")   # 123,456.79
```

**백분율**

`%`를 쓰면 값에 100을 곱해 백분율로 표시한다.

```python
rate = 0.2567

print(f"{rate:.1%}")   # 25.7%
```

| 서식 | 의미 | 예시 값 | 출력 |
|---|---|---|---|
| `{v}` | 값을 그대로 출력 | `1234567` | `1234567` |
| `{v:,}` | 천 단위 쉼표 | `1234567` | `1,234,567` |
| `{v:.2f}` | 반올림하여 소수점 둘째 자리까지 | `123456.789` | `123456.79` |
| `{v:.0f}` | 반올림하여 정수로 | `123456.789` | `123457` |
| `{v:,.2f}` | 쉼표와 소수점 함께 | `123456.789` | `123,456.79` |
| `{v:.1%}` | 백분율 한 자리까지 | `0.2567` | `25.7%` |



````python
region = "서울"
sales = 1234567
average_sales = 123456.789

print(f"지역: {region}")
print(f"총매출: {sales:,}원")
print(f"평균 매출: {average_sales:,.2f}원")
````

````python
#  f-string
# 문자열 앞에 f를 붙이고, 중괄호 {} 안에 변수나 식을 넣으면 그 자리에 값이 들어간다.
# {값:,}  → 천 단위 쉼표      {값:.1f} → 소수 첫째 자리까지      {값:.0%} → 백분율
amount = 1234567.891
print(f"{region} 매출: {amount:,}원")     # 서울 매출: 1,234,567.891원
````

````python
print(f"반올림: {amount:,.0f}원")         # 반올림: 1,234,568원
print(f"소수 한 자리: {amount:.1f}")      # 소수 한 자리: 1234567.9
print(f"비율: {0.0347:.1%}")              # 비율: 3.5%
````

# **2. 객체와 자료형**

파이썬은 숫자, 문자열, 리스트 등의 데이터를 객체로 표현하며, 각 객체는 자료형을 가진다. 자료형에 따라 저장되는 데이터의 의미와 사용할 수 있는 연산 및 기능이 달라진다.


| 분류 | 자료형 | 의미 | 예시 |
|---|---|---|---|
| 숫자형 | `int` | 정수 | `100` |
| 숫자형 | `float` | 실수 | `3.14` |
| 문자열 | `str` | 문자로 이루어진 데이터 | `"서울"` |
| 불리언형 | `bool` | 참과 거짓 | `True`, `False` |
| 리스트 | `list` | 순서가 있고 변경 가능한 자료 모음 | `[10, 20, 30]` |
| 튜플 | `tuple` | 순서가 있고 변경할 수 없는 자료 모음 | `(10, 20, 30)` |
| 딕셔너리 | `dict` | 키와 값으로 구성된 자료 모음 | `{"서울": 100}` |
| 집합 | `set` | 중복을 허용하지 않는 자료 모음 | `{"서울", "부산"}` |
| 특수한 값 | `NoneType` | 값이 없음을 나타내는 값 | `None` |


`None`은 자료형의 이름이 아니라 값이 없음을 나타내는 특수한 값이다. `None`의 자료형은 `NoneType`이다.

공식 문서: [내장 자료형](https://docs.python.org/ko/3.13/library/stdtypes.html) · [표준형 계층](https://docs.python.org/ko/3.13/reference/datamodel.html#the-standard-type-hierarchy) · [`None`](https://docs.python.org/ko/3.13/library/constants.html#None)


````python
order_count = 10
average_sales = 125000.5
region = "서울"
is_target = True
manager_name = None

print(type(order_count))
print(type(average_sales))
print(type(region))
print(type(is_target))
print(type(manager_name))
````

````python
print(type(10))        # <class 'int'>   정수
print(type(3.5))       # <class 'float'> 실수
print(type("10"))      # <class 'str'>   문자열 (따옴표가 있으면 숫자처럼 보여도 문자열이다)
print(type(True))      # <class 'bool'>  참/거짓
print(type(None))      # <class 'NoneType'> 값 없음
````

## **`None`과 값 없음**

`None`, 빈 문자열 `""`, 숫자 `0`은 서로 다른 값이다.
데이터 분석에서는 값이 없다는 의미를 명확히 구분해야 한다.

None은 “유효한 값이 없거나 아직 정해지지 않았음”을 나타내는 특별한 단일 객체이며, 자료형은 NoneType이다. 빈 문자열이나 숫자 0과는 다르다.

| 값 | 의미 |
|---|---|
| `None` | 유효한 값이 없거나 아직 정해지지 않았음 |
| `""` | 비어 있는 문자열이 존재함 |
| `0` | 숫자 0이라는 값이 존재함 |

## **객체 동일성 비교: `is`, `is not`**

`==`와 `is`는 비교하는 기준이 다르다.

- `==` (동등성): 두 변수가 **가지고 있는 값**이 같은지 비교한다.
- `is` (동일성): 두 변수가 **메모리상에서 완전히 같은 객체**를 가리키는지 비교한다.

파이썬에서 변수는 값을 직접 담는 상자가 아니라 객체를 가리키는 이름이다. 따라서 값은 같더라도 메모리상에서 서로 다른 객체일 수 있으므로 두 연산자의 결과가 다를 수 있다.

> **`None`을 비교할 때 `is`를 쓰는 이유:**
> `None`은 파이썬 전체에서 메모리상에 단 하나만 존재하는 특수한 객체다. 따라서 값이 비어 있는지 확인할 때는 `== None` 대신 **`is None`** 또는 `is not None`을 쓰는 것이 공식 권장 사항이다.

```python
value is None
value is not None
```

일반적인 숫자나 문자열 값 비교에는 `==`, `!=`를 사용한다.

공식 문서: [값 비교](https://docs.python.org/ko/3.13/reference/expressions.html#comparisons)


````python
# 1. None 체크 (is 사용)
manager_name = None

print(manager_name is None)      # True  (manager_name이 None 객체를 가리킴)
print(manager_name is not None)  # False (None이 아니라는 조건은 거짓)

# 2. 일반 값 비교 (== 사용)
region = "서울"
print(region == "서울")          # True  (값이 "서울"과 같은지 비교)

# 3. 값은 같지만 다른 객체인 예시
list_a = [1, 2, 3]
list_b = [1, 2, 3]

print(list_a == list_b)          # True  (내용물/값이 같음)
print(list_a is list_b)          # False (메모리상 서로 다른 별개의 리스트 객체임)
````

## **함수와 메서드의 차이**

| ---------- | -------------------------- | ------------------------------------ |

- 일반적인 사용 코드에서는 객체 뒤에 점(`.`)을 붙여 호출하는지가 가장 쉽게 확인할 수 있는 차이다.

## **산술연산자**

연산자는 값끼리 **계산하거나 비교**할 때 쓰는 기호다. 매출의 합을 구하거나, 어떤 값이 기준보다 큰지 판단하는 일이 모두 연산자로 이루어진다.

- `+`: 더하기
- `-`: 빼기
- `*`: 곱하기
- `/`: 나누기
  - 결과는 항상 실수(`float` 형)가 된다. `10 / 5`는 정수 `2`가 아니라 실수 `2.0`이다
  - 0으로 나누면 `ZeroDivisionError`가 발생한다
- `//`: 몫
  - 나눈 결과에서 소수점 아래를 내려 정수로 만든다
  - `7 // 2`는 `3`, `-7 // 2`는 `-4`가 된다 (작은 쪽 정수로 내림)
- `%`: 나머지
  - 나눈 뒤 남는 값을 구한다. `7 % 2`는 `1`이다
  - 짝수와 홀수를 구분하거나(`x % 2 == 0`) 배수를 판별할 때 사용한다
- `**`: 거듭제곱연산자이다.
   -  `2 ** 3`은 `8`이다

### **계산 순서**

수학과 같이 곱하기와 나누기를 먼저 계산하고 더하기와 빼기를 나중에 계산한다. 순서를 바꾸려면 괄호를 사용한다.

```python
print(10 + 5 * 2)     # 20  (곱하기를 먼저 계산)
print((10 + 5) * 2)   # 30  (괄호를 먼저 계산)
```

평균처럼 합계를 먼저 구한 뒤 나누어야 하는 계산에서는 괄호를 정확히 써야 한다.

```python
print((120000 + 150000 + 90000) / 3)
```

| 연산자 | 의미 | 예시 | 결과 |
|---|---|---|---|
| `+` | 더하기 | `10 + 3` | `13` |
| `-` | 빼기 | `10 - 3` | `7` |
| `*` | 곱하기 | `10 * 3` | `30` |
| `/` | 나누기 | `10 / 3` | `3.333...` |
| `//` | 몫 | `10 // 3` | `3` |
| `%` | 나머지 | `10 % 3` | `1` |
| `**` | 거듭제곱 | `10 ** 3` | `1000` |

공식 문서: [숫자 자료형과 연산](https://docs.python.org/ko/3.13/library/stdtypes.html#numeric-types-int-float-complex)



공식 문서: [숫자 자료형](https://docs.python.org/ko/3.13/library/stdtypes.html#numeric-types-int-float-complex) · [부동 소수점 산술](https://docs.python.org/ko/3.13/tutorial/floatingpoint.html)

````python
a = 10
b = 3

print(a + b)   # 13                 : 더하기
print(a - b)   # 7                  : 빼기
print(a * b)   # 30                 : 곱하기
print(a / b)   # 3.3333333333333335 : 나누기 (결과는 항상 실수 float 형태로 반환)
print(a // b)  # 3                  : 몫 (10을 3으로 나눈 정수 몫만 남김, 소수점 아래 내림)
print(a % b)   # 1                  : 나머지 (10을 3으로 나눈 뒤 남은 값)
print(a ** b)  # 1000               : 거듭제곱 (10의 3제곱 = 10 * 10 * 10)
````

## **연산자 우선순위**

하나의 식에 여러 연산자가 함께 사용되면 연산자 우선순위에 따라 계산 순서가 결정된다. 우선순위가 높은 연산을 먼저 수행하며, 괄호를 사용하면 계산 순서를 명확하게 지정할 수 있다.

| 우선순위 | 연산 또는 문법 | 연산자·표현 | 예 |
|---:|---|---|---|
| 1 | 괄호로 묶은 식 | `(...)` | `(10 + 5) * 2` |
| 2 | 거듭제곱 | `**` | `2 ** 3` |
| 3 | 단항 부호 | `+x`, `-x` | `-10`, `+10` |
| 4 | 곱셈·나눗셈·몫·나머지 | `*`, `/`, `//`, `%` | `10 * 2`, `10 // 3` |
| 5 | 덧셈·뺄셈 | `+`, `-` | `10 + 3`, `10 - 3` |
| 6 | 비교·동일성·멤버십 | `==`, `!=`, `>`, `>=`, `<`, `<=`, `is`, `is not`, `in`, `not in` | `sales >= 100000` |
| 7 | 논리 부정 | `not` | `not is_target` |
| 8 | 논리곱 | `and` | `condition1 and condition2` |
| 9 | 논리합 | `or` | `condition1 or condition2` |

> 참고: 괄호 `()`는 연산자가 아니라 표현식을 묶어 계산 순서를 지정하는 문법 기호다. 함수 이름 뒤의 괄호는 함수를 호출할 때 사용한다.

### **Python에는 `++`, `--` 증감 연산자가 없다**

파이썬에는 C나 Java의 `count++`, `++count`와 같은 전위·후위 증감 연산자가 없다. 값을 증가하거나 감소시키려면 `+=`, `-=`를 사용한다. `++count`는 값을 증가시키지 않는다. 단항 `+`를 두 번 적용한 것으로 해석되므로 원래 값이 그대로 나온다. `count++`와 `count--`는 문법 오류가 발생한다.

공식 문서: [Python 연산자 우선순위](https://docs.python.org/ko/3.13/reference/expressions.html#operator-precedence)


````python
# 1. 우선순위: 곱하기가 더하기보다 먼저다
print(10 + 5 * 2)      # 20
print((10 + 5) * 2)    # 30  괄호가 가장 먼저

# 2. 평균 계산에서 괄호를 빠뜨리면 결과가 달라진다
print(120000 + 150000 + 90000 / 3)     # 300000.0  (90000 / 3 만 먼저 계산된다)
print((120000 + 150000 + 90000) / 3)   # 120000.0  (합계를 먼저 구한 뒤 나눈다)

# 3. ++ 는 증가가 아니다
count = 10
print(+count)    # 10   양수 부호, 그대로
print(-count)    # -10  음수 부호, 부호가 뒤집힌다
print(++count)   # 10   +(+10)
print(--count)   # 10   -(-10), 두 번 뒤집혀 원래대로
print(-+count)   # -10  -(+10)
````

## **`round()` (소수점 반올림)**

숫자를 지정한 소수점 자릿수로 반올림한다. 자릿수를 생략하면 가장 가까운 정수로 반올림한다.

```python
round(숫자)                  # 가장 가까운 정수로 반올림
round(숫자, 소수점자릿수)    # 지정한 소수점 자릿수까지 반올림
```
 **파이썬 round()의 주의점 (Round half to even):**
짝수 반올림 (Round to Even):

반올림할 숫자가 두 결과의 정확한 중간값인 경우, 마지막 자리가 짝수인 값을 선택한다.

round(0.5) → 0, round(-0.5) → 0

round(2.5) → 2, round(3.5) → 4

**부동소수점 오차로 인한 결과 차이:**

실수(float)는 2진수로 저장되므로 round(2.675, 2)의 결과가 2.68이 아니라 **2.67**로 출력될 수 있다. 이는 버그가 아닌 2진 소수의 한계이다.

 **결론:** 계산 결과 자체를 반올림해 다시 사용하려면 round()를 사용하고, 원래 값은 유지하면서 화면에 표시할 자릿수만 조정하려면 f-string 서식을 사용한다.

공식 문서: https://docs.python.org/ko/3.13/library/functions.html#round

````python
# 1. 기본 사용 및 정수 반환
print(round(123456.789))       # 123457 (int 형으로 반환)
print(round(123456.789, 2))    # 123456.79 (float 형으로 반환)

# 2. 짝수 반올림 (Round to even)
print(round(0.5))              # 0  (가장 가까운 짝수인 0)
print(round(2.5))              # 2  (가장 가까운 짝수인 2)
print(round(3.5))              # 4  (가장 가까운 짝수인 4)

# 3. 부동소수점 오차 주의 예시
print(round(2.675, 2))         # 2.67 (2.68이 아님에 주의!)

# 4. 출력 목적일 때 권장하는 f-string
average_sales = 123456.789
print(f"{average_sales:,.2f}")  # 123,456.79 (반올림하여 문자열로 표시)


print(f"{2.675:,.2f}")   # 2.67
````

`**`는 거듭제곱이다. `x ** 2`는 x를 두 번 곱한 값(제곱)이고, `x ** 0.5`는 두 번 곱하면 x가 되는 수(제곱근)다.

## **`divmod()`와 `math.ceil()`**


| 함수 | 하는 일 | 예시 | 결과 |
|---|---|---|---|
| `divmod(a, b)` | 몫과 나머지를 튜플로 함께 반환 | `divmod(47, 12)` | `(3, 11)` |
| `math.ceil(x)` | 소수점을 올려 정수로 (올림) | `math.ceil(3.2)` | `4` |
| `math.floor(x)` | 소수점을 버려 정수로 (내림) | `math.floor(3.8)` | `3` |


공식 문서: [divmod()](https://docs.python.org/ko/3.13/library/functions.html#divmod) · [math.ceil()](https://docs.python.org/ko/3.13/library/math.html#math.ceil)


````python
#47개를 최대 12개를 담는 박스에 담으려면 3.9… 박스가 아니라 4박스가 필요하므로 올림을 써야 한다.

import math

total_quantity = 47
box_size = 12

# 1. divmod(): 몫과 나머지를 한 번에
full_boxes, remainder = divmod(total_quantity, box_size)
print(f"가득 찬 박스: {full_boxes}개, 남는 수량: {remainder}개")   # 가득 찬 박스: 3개, 남는 수량: 11개

# 2. math.ceil(): 남는 수량까지 담으려면 박스가 몇 개 필요한가
needed_boxes = math.ceil(total_quantity / box_size)
print(f"필요한 박스: {needed_boxes}개")                            # 필요한 박스: 4개

# 3. round()와의 차이
print(round(47 / 12))        # 4  (3.92를 반올림)
print(round(37 / 12))        # 3  (3.08을 반올림 → 박스가 모자란다)
print(math.ceil(37 / 12))    # 4  (올림 → 모두 담을 수 있다)
````

## **단락 평가 (Short-Circuit)**

논리식의 결과가 이미 결정되면 뒤의 표현식은 실행하지 않는다.

- `and`: 앞의 값이 거짓이면 뒤를 확인하지 않는다.
- `or`: 앞의 값이 참이면 뒤를 확인하지 않는다.

이 성질은 오류를 피할 때 쓸 수 있다. 예를 들어 `count > 0 and total / count > 100`처럼 적으면, `count`가 0일 때 앞이 거짓이 되어 뒤의 나눗셈을 실행하지 않으므로 `ZeroDivisionError`가 나지 않는다.

공식 문서: [불 연산](https://docs.python.org/ko/3.13/library/stdtypes.html#boolean-operations-and-or-not)


````python
# Short-Circuit(단락 평가)
# 논리식의 결과가 이미 결정되면 뒤의 표현식은 실행하지 않는다.

# and: 앞이 거짓이면 뒤를 실행하지 않는다.
# 숫자 0은 거짓으로 취급하며, 결과를 정한 쪽의 값이 그대로 나온다.
print(0 and 100)      # 0

# and: 앞이 참이면 뒤까지 실행
print(1 and 2)        # 2

# or: 앞이 참이면 뒤를 실행하지 않는다.
print("a" or "b")     # a

#  예: 건수가 0이면 나눗셈을 하지 않는다
count = 0
total = 0
print(count > 0 and total / count > 100)   # False  (ZeroDivisionError가 나지 않는다)
````

# 인덱싱(Indexing)과 슬라이싱(Slicing)

문자열은 여러 글자가 순서대로 나열된 자료형이다. 파이썬에서는 **인덱싱(Indexing)**과 **슬라이싱(Slicing)**을 사용하여 문자열의 특정 글자나 원하는 범위의 글자를 가져올 수 있다.

## 1. 인덱싱(Indexing)이란?

**인덱싱은 문자열에서 특정 위치의 글자 하나를 가져오는 것**이다.

문자열의 각 글자에는 위치를 나타내는 번호가 있으며, 이 번호를 **인덱스(Index)**라고 한다.

- 앞에서부터 세는 인덱스는 `0`부터 시작한다.
- 뒤에서부터 세는 인덱스는 `-1`부터 시작한다.
- 인덱싱할 때는 대괄호 `[ ]` 안에 위치 번호를 작성한다.

### 인덱스 번호 확인

```text
문자열:    F   O   O   D   -   2   0   2   6
양수 인덱스 0   1   2   3   4   5   6   7   8
음수 인덱스 -9 -8  -7  -6  -5  -4  -3  -2  -1
```

### 기본 문법

```python
문자열[위치]  # 특정 위치의 글자 하나 가져오기
```

### 예제

```python
code = "FOOD-2026"

print(code[0])    # F
print(code[1])    # O
print(code[4])    # -
print(code[-1])   # 6
print(code[-2])   # 2
```

`code[0]`은 첫 번째 글자인 `F`를 가져오고, `code[-1]`은 마지막 글자인 `6`을 가져온다.

음수 인덱스를 사용하면 문자열의 전체 길이를 몰라도 마지막 글자나 끝에서 두 번째 글자를 가져올 수 있다.

## 2. 슬라이싱(Slicing)이란?

**슬라이싱은 문자열에서 원하는 범위의 글자를 잘라서 가져오는 것**이다.

인덱싱이 글자 하나를 가져오는 방식이라면, 슬라이싱은 여러 글자를 범위로 지정하여 가져오는 방식이다.

### 기본 문법

```python
문자열[시작:끝]
```

- `시작`: 가져올 범위의 시작 위치이며 결과에 포함된다.
- `끝`: 가져올 범위의 끝 위치이며 결과에 포함되지 않는다.

**슬라이싱은 시작 위치는 포함하고 끝 위치는 포함하지 않는다.**

### 예제

```python
code = "FOOD-2026"

print(code[0:4])   # FOOD
print(code[5:9])   # 2026
```

`code[0:4]`는 0번부터 4번 직전인 3번까지 가져오므로 `FOOD`가 출력된다.

```text
문자열:  F   O   O   D   -   2   0   2   6
인덱스:  0   1   2   3   4   5   6   7   8
선택:   [F   O   O   D]
결과:    FOOD
```

슬라이싱은 기본적으로 인덱스를 1씩 증가시키면서 글자를 가져온다. 증가 간격을 생략한 일반적인 [시작:끝] 형태에서 시작과 끝이 유효하고 시작이 끝보다 작거나 같으면, 가져오는 글자 수는 끝 번호 - 시작 번호이다.

예를 들어 [0:4]는 0번부터 3번까지 가져오므로 4 - 0 = 4개의 글자를 가져온다.

## 3. 슬라이싱에서 시작과 끝 생략하기

슬라이싱은 시작 위치나 끝 위치를 생략할 수 있다.

```python
code = "FOOD-2026"

print(code[:4])    # FOOD
print(code[5:])    # 2026
print(code[:])     # FOOD-2026
```

- `code[:4]`: 처음부터 4번 직전까지 가져온다.
- `code[5:]`: 5번부터 마지막까지 가져온다.
- `code[:]`: 문자열 전체를 가져온다.

## 4. 인덱싱과 슬라이싱 비교

| 구분 | 인덱싱(Indexing) | 슬라이싱(Slicing) |
|---|---|---|
| 목적 | 특정 위치의 글자 가져오기 | 원하는 범위의 글자 가져오기 |
| 문법 | `문자열[위치]` | `문자열[시작:끝]` |
| 예시 | `code[0]` | `code[0:4]` |
| 결과 | `F` | `FOOD` |
| 특징 | 하나의 위치 지정 | 시작과 끝 범위 지정 |

## 5. 전체 예제

```python
code = "FOOD-2026"

# 인덱싱: 특정 위치의 글자 하나 가져오기
print(code[0])      # F
print(code[-1])     # 6

# 슬라이싱: 원하는 범위의 글자 가져오기
print(code[0:4])    # FOOD
print(code[:4])     # FOOD
print(code[5:])     # 2026
```

### **참고: len()**

글자 수를 셀 때 쓰는 `len()`은 위 메서드들과 사용 형태가 다르다. `len()`은 특정 자료형에 속하지 않는 독립된 함수이므로, 점을 찍지 않고 괄호 안에 값을 넣어 사용한다.

```python
region = "서울특별시"

print(region.upper())   # 메서드: 값 뒤에 점을 찍는다
print(len(region))      # 함수: 괄호 안에 값을 넣는다
```

`region.len()`처럼 메서드 형태로 쓰면 문자열에 `len`이라는 기능이 없어 `AttributeError`가 발생한다.

공식 문서: [len()](https://docs.python.org/ko/3.13/library/functions.html#len)

## **리스트에서 값 꺼내기 (인덱싱과 슬라이싱)**

문자열과 동일한 위치 번호(인덱스) 규칙을 사용한다. 앞에서부터 세면 0번부터, 뒤에서부터 세면 -1번부터 시작한다.

```python
리스트[위치]         # 인덱싱: 특정 위치의 값 하나 꺼내기
리스트[시작:끝]      # 슬라이싱: 범위로 여러 값 잘라내기
```
* 음수 인덱스: 전체 길이를 몰라도 마지막 값을 바로 꺼낼 수 있다. (sales[-1]은 마지막 항목)

* 끝 위치 미포함: 슬라이싱에서 끝 위치의 항목은 결과에 포함되지 않는다. (잘라낸 개수 = 끝 번호 - 시작 번호)

* 생략 표기: 시작이나 끝 인덱스는 생략할 수 있다.

- `sales[0]`: 첫 번째 값
- `sales[-1]`: 맨 마지막 값 (음수는 뒤에서부터)
- `sales[1:4]`: 1번부터 3번까지 (**끝 번호 4는 포함하지 않는다**)

````python
sales = [120000, 150000, 90000, 200000, 170000]

print(sales[0])        # 120000 (첫 번째 항목)
print(sales[-1])       # 170000 (마지막 항목)
print(sales[1:4])      # [150000, 90000, 200000] (1번부터 3번 인덱스까지)
print(sales[-2:])      # [200000, 170000] (마지막 2개 항목 잘라내기)
````

## **값 추가·수정·삭제**

리스트는 문자열과 달리 **가변(Mutable) 객체**이므로 생성한 후에도 내부 요소를 추가, 수정, 삭제할 수 있다.

리스트의 요소를 변경할 때는 인덱스를 이용하거나 리스트에서 제공하는 메서드를 사용할 수 있다.

- `리스트.append(값)`: 리스트의 맨 뒤에 요소를 추가한다.
- `리스트[인덱스] = 값`: 지정한 위치의 요소를 수정한다.
- `리스트.pop()`: 리스트의 마지막 요소를 삭제하고 삭제한 값을 반환한다.
- `리스트.remove(값)`: 지정한 값과 일치하는 첫 번째 요소를 삭제한다.

### **1. 값 추가 — append()**

`append()`는 리스트의 맨 뒤에 새로운 요소를 추가하는 메서드이다.

```python
sales = [10, 20, 30]

sales.append(40)

print(sales)  # [10, 20, 30, 40]
```

`append()`는 새로운 리스트를 반환하는 것이 아니라 **기존 리스트 객체를 직접 변경한다.** 따라서 반환값은 `None`이다.

### **2. 값 수정 — 인덱싱**

리스트는 인덱스를 사용하여 특정 위치의 요소를 수정할 수 있다.

```python
sales = [10, 20, 30]

sales[1] = 25

print(sales)  # [10, 25, 30]
```

인덱스 `1`에 있는 값 `20`이 `25`로 변경된다. 이때 새로운 리스트를 만드는 것이 아니라 기존 리스트 객체의 요소를 변경한다.

### **3. 값 삭제 — pop(), remove()**

리스트에서 요소를 삭제할 때는 `pop()`과 `remove()` 메서드를 사용할 수 있다. 두 메서드는 삭제 기준과 반환값이 다르다.

**① pop() — 위치를 기준으로 삭제**

`pop()`은 지정한 위치의 요소를 삭제하고, 삭제한 값을 반환한다. 위치를 생략하면 마지막 요소를 삭제한다.

```python
sales = [10, 20, 30]

removed = sales.pop()

print(removed)  # 30
print(sales)    # [10, 20]
```

**② remove() — 값을 기준으로 삭제**

`remove()`는 지정한 값과 일치하는 첫 번째 요소를 삭제한다.

```python
sales = [10, 20, 30]

sales.remove(20)

print(sales)  # [10, 30]
```

### **[참고] pop()과 remove() 비교**

| 방법 | 삭제 기준 | 반환값 |
|---|---|---|
| `pop(인덱스)` | 지정한 위치의 요소 삭제 | 삭제한 값 |
| `remove(값)` | 지정한 값과 일치하는 첫 번째 요소 삭제 | `None` |

- `pop()`은 삭제한 값을 반환하므로 변수에 저장할 수 있다.
- `pop()`에서 인덱스를 생략하면 마지막 요소를 삭제한다.
- `pop()`은 빈 리스트에서 실행하거나 존재하지 않는 인덱스를 지정하면 `IndexError`가 발생한다.
- `remove()`는 인덱스가 아니라 값을 지정한다. 같은 값이 여러 개 있으면 첫 번째 요소만 삭제하며, 해당 값이 없으면 `ValueError`가 발생한다.


**공식 문서:** [Python 3.13 — 리스트 자료형](https://docs.python.org/ko/3.13/tutorial/datastructures.html#more-on-lists)

````python
# [예제] 리스트 값 추가·수정·삭제

sales = [120000, 150000, 90000, 180000]

# 1. append(): 리스트 맨 뒤에 값 추가
append_result = sales.append(200000)

print(sales)          # [120000, 150000, 90000, 180000, 200000]
print(append_result)  # None


# 2. 인덱싱: 지정한 위치의 값 수정
sales[0] = 130000

print(sales)          # [130000, 150000, 90000, 180000, 200000]


# 3. remove(): 지정한 값을 찾아 삭제
remove_result = sales.remove(90000)

print(sales)          # [130000, 150000, 180000, 200000]
print(remove_result)  # None


# 4. pop(): 지정한 위치의 값을 삭제하고 반환
removed_sales = sales.pop(1)

print(sales)          # [130000, 180000, 200000]
print(removed_sales)  # 150000


# 5. del: 지정한 위치의 값을 삭제
del sales[0]

print(sales)          # [180000, 200000]


# [참고] 오류 발생 사례

# remove()에서 리스트에 없는 값을 지정하면 ValueError가 발생한다.
# sales.remove(99999)

# pop()에서 범위를 벗어난 위치를 지정하면 IndexError가 발생한다.
# sales.pop(10)

# 인덱싱으로 수정할 때 범위를 벗어난 위치를 지정하면 IndexError가 발생한다.
# sales[10] = 100000
````

## **`sort()`와 `sorted()`**

파이썬에서는 `sort()` 메서드와 `sorted()` 내장 함수를 사용하여 데이터를 정렬할 수 있다.

두 방법 모두 기본적으로 **오름차순(작은 값부터 큰 값 순서)**으로 정렬하지만, 원본 데이터의 변경 여부와 반환값에 차이가 있다.

| 구분 | `sort()` | `sorted()` |
|---|---|---|
| 종류 | 리스트 메서드 | 내장 함수 |
| 사용 형태 | `리스트.sort()` | `sorted(리스트)` |
| 원본 데이터 | 직접 변경 | 변경하지 않음 |
| 반환값 | `None` | 정렬된 새 리스트 |
| 기본 정렬 | 오름차순 | 오름차순 |
| 내림차순 | `reverse=True` | `reverse=True` |

`reverse=True`를 지정하면 내림차순(큰 값부터 작은 값 순서)으로 정렬한다.

### **1. sorted() — 새로운 리스트 반환**

`sorted()`는 원본 데이터를 변경하지 않고 **정렬된 새로운 리스트를 반환하는 내장 함수**이다.

```python
sales = [120000, 150000, 90000]

print(sorted(sales))                # [90000, 120000, 150000]
print(sorted(sales, reverse=True))  # [150000, 120000, 90000]
print(sales)                        # [120000, 150000, 90000]
```

### **2. sort() — 기존 리스트 직접 변경**

`sort()`는 리스트의 요소를 직접 정렬하는 메서드이다. 정렬된 리스트를 새로 반환하지 않으며, 반환값은 `None`이다.

```python
sales = [120000, 150000, 90000]

result = sales.sort()

print(sales)   # [90000, 120000, 150000]
print(result)  # None

sales.sort(reverse=True)

print(sales)   # [150000, 120000, 90000]
```

**주의:** `sort()`는 기존 리스트를 직접 변경하므로 `sales = sales.sort()`처럼 작성하면 `sales`에 `None`이 할당된다.

공식 문서:

- [Python 3.13 — list.sort()](https://docs.python.org/ko/3.13/library/stdtypes.html#list.sort)

## **문자열과 숫자 리스트의 차이**
`"100,200,300"`은 겉보기엔 숫자가 들어있어 보이지만 통째로 **하나의 문자열**이고, `[100, 200, 300]`은 **숫자 3개가 각각 담긴 리스트**다.

문자열 자체는 사칙연산이나 합계(`sum`) 계산을 할 수 없지만, 리스트는 내부 값 하나하나에 접근하여 계산할 수 있다.  
따라서 텍스트로 된 데이터를 받아오면 **나누고(`split`) 숫자 타입으로 변환**하여 리스트로 다룬다.

공식 문서: [자료형 표준 타입](https://docs.python.org/ko/3.13/library/stdtypes.html)

````python
# 1. 문자열과 리스트 타입 비교
sales_text = "100,200,300"
sales_list = [100, 200, 300]

print(sales_text)       # 100,200,300
print(sales_list)       # [100, 200, 300]

print(type(sales_text)) # <class 'str'>  (문자열 하나)
print(type(sales_list)) # <class 'list'> (리스트)

# 2. 연산 및 통계 계산의 차이
print(sum(sales_list))  # 600 (숫자 리스트는 합계 계산 가능)
# sum(sales_text)       #  TypeError 발생! 문자열은 sum() 불가

# 3. [참고] 문자열을 쪼갰을 때의 주의점 (split 사용)
split_result = sales_text.split(",")
print(split_result)     # ['100', '200', '300']
                        #  split() 결과는 숫자가 아닌 '문자열 리스트'임에 주의!
````

````python
# [실습] 문자열과 숫자 리스트의 차이
# 문제: sales_text = "120,150,90" 을 split(",")으로 나눠 문자열 리스트를 출력하고, 세 항목을 int()로 각각 바꿔 sales = [120, 150, 90] 을 만들어 합계를 출력한다.
sales_text = "120,150,90"
sales_text = sales_text.split(",")
new_sales_text = []
for i in range(len(sales_text)):
   new_sales_text.append(int(sales_text[i]))
print(new_sales_text)
````

## **학습 점검 정답**

### **1. 문자열 `"100"`과 정수 `100`의 차이는 무엇인가?**

`"100"`은 문자로 이루어진 문자열(`str`)이고, `100`은 계산에 사용하는 정수(`int`)다.

```python
print(type("100"))  # <class 'str'>
print(type(100))    # <class 'int'>

print("100" + "20") # 10020
print(100 + 20)     # 120
```

문자열 `"100"`을 숫자 계산에 사용하려면 `int()`로 변환해야 한다.

```python
number = int("100")

print(number + 20)  # 120
```

### **2. `=`와 `==`의 차이는 무엇인가?**

`=`는 오른쪽 값을 왼쪽 변수에 대입할 때 사용하고, `==`는 두 값이 같은지 비교할 때 사용한다.

```python
score = 100         # 대입
print(score == 100) # 비교 결과: True
```

- `=`: 대입
- `==`: 값 비교
- `==`의 결과는 `True` 또는 `False`다

### **3. 리스트의 첫 번째 위치 번호는 무엇인가?**

리스트의 첫 번째 위치 번호는 `0`이다. 파이썬의 인덱스는 `0`부터 시작한다.

```python
regions = ["서울", "부산", "대구"]

print(regions[0])  # 서울
print(regions[1])  # 부산
```

### **4. 함수와 메서드의 사용 형태는 어떻게 다른가?**

함수는 일반적으로 함수 이름 뒤에 괄호를 붙여 호출하고, 처리할 값을 인자로 전달한다.

메서드는 특정 객체 뒤에 점(`.`)을 붙여 호출한다.

```python
text = "  Python  "
values = [10, 20]

print(len(text))    # 함수
print(text.strip()) # 문자열 메서드

values.append(30)   # 리스트 메서드
```

- 함수: `함수명(값)`
- 메서드: `객체.메서드명()`

---

### **5. `None`, 빈 문자열, 숫자 `0`은 같은 값인가?**

서로 다른 값이다.

| 값 | 의미 |
|---|---|
| `None` | 값이 없음을 나타내는 특수한 값 |
| `""` | 글자가 하나도 없는 문자열 |
| `0` | 정수 0 |

```python
value = None
text = ""
number = 0

print(value is None) # True
print(text == "")    # True
print(number == 0)   # True
```

`None`인지 확인할 때는 `is None`을 사용한다.

---

### **6. `==`와 `is`는 어떻게 다른가?**

`==`는 두 값이 같은지 비교하고, `is`는 두 변수가 같은 객체를 가리키는지 비교한다.

```python
a = [1, 2]
b = [1, 2]

print(a == b)   # True  (값이 같다)
print(a is b)   # False (서로 다른 리스트 객체다)
```

`None`인지 확인할 때는 `is None`을 쓴다. 일반적인 숫자·문자열 비교에는 `==`를 쓴다.


### **[참고] del 삭제문**

`del`은 함수나 메서드가 아닌 **파이썬의 삭제문(del statement)**이다.

리스트 요소뿐만 아니라 변수 이름의 바인딩, 딕셔너리 항목, 객체 속성 등을 삭제할 때도 사용한다.

**① 리스트 요소 삭제**

```python
sales = [10, 20, 30]

del sales[0]

print(sales)  # [20, 30]
```

**② 변수 이름 삭제**

```python
score = 90

del score

print(score)  # NameError 발생
```

변수 이름과 객체 사이의 참조 연결을 제거한다. 객체 자체가 반드시 삭제되는 것은 아니다.

**③ 딕셔너리 항목 삭제**

```python
product = {"name": "키보드", "price": 30000}

del product["price"]

print(product)  # {'name': '키보드'}
```

**④ 객체 속성 삭제**

```python
class Product:
    pass

item = Product()
item.price = 30000

del item.price

print(item.price)  # AttributeError 발생
```

| 대상 | 사용 예시 | 동작 |
|---|---|---|
| 리스트 요소 | `del sales[0]` | 지정한 요소 삭제 |
| 리스트 범위 | `del sales[1:3]` | 지정한 범위의 요소 삭제 |
| 변수 이름 | `del score` | 이름의 바인딩 제거 |
| 딕셔너리 항목 | `del product["price"]` | 지정한 키와 값 삭제 |
| 객체 속성 | `del item.price` | 해당 속성 삭제 |
