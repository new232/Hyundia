# **4. 조건문 (Conditional Statements)**

지금까지 작성한 코드는 위에서 아래로 모든 줄이 무조건 순서대로 실행됐다. 하지만 실제 데이터 분석이나 프로그램에서는 **조건이나 값에 따라 다르게 처리**해야 할 때가 많다.

예를 들어 매출이 목표치를 넘었는지에 따라 다른 메시지를 출력하거나 등급을 나누고 싶을 때 필요한 것이 바로 **조건문**이다.

우리도 일상에서 조건에 따라 행동을 결정한다.
- **만약 비가 오면** ➔ 우산을 챙긴다.
- **그렇지 않으면** ➔ 그냥 나간다.

`if`는 영어로 "만약"이라는 뜻이며, 파이썬 조건문은 이 "만약 ~라면"을 코드로 표현하는 도구다.



## **1) 코드의 흐름이 분기된다**

조건문은 지정한 조건이 **참(`True`)인지 거짓(`False`)인지**를 판단하여 실행할 코드 블록을 선택한다.

```text
                 ┌─ 조건이 참(True)이면  ➔ A 코드 실행
 매출을 확인한다 ─┤
                 └─ 조건이 거짓(False)이면 ➔ B 코드 실행
```


  
  
## **2)  if**

조건이 참일 때만 실행하는 가장 기본적인 구조다.

if 조건: ➔ "만약 이 조건이 참이라면"

들여쓴 줄 ➔ 조건이 참일 때만 실행되는 코드 영역



공식 문서: [if 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#if-statements)

````python
sales = 150000

if sales >= 100000:
    print("목표 달성")


#sales가 100,000 이상이므로 "목표 달성!"이 출력된다. 만약 sales = 50000이었다면 조건이 거짓이므로 아무것도 출력되지 않고 넘어간다.
````

````python
# [실습] if
# 문제: defect_rate = 0.05 일 때 0.03 이상이면 "재검사 필요" 를 출력하는 if 문을 작성하고, defect_rate를 0.01로 바꿔 다시 실행해 아무것도 출력되지 않는지 확인한다.
defect_rate = 0.05
if defect_rate >= 0.03:
    print("재 검사 필요")
defect_rate = 0.01
if defect_rate >= 0.03:
    print("재 검사 필요")
````

## **3) else: 그렇지 않으면**
if만 사용하면 조건이 거짓일 때 아무 일도 일어나지 않는다.

"조건이 거짓일 때 수행할 다른 코드"를 지정하고 싶다면 else("그렇지 않으면")를 붙인다.

```text
 if   조건이 참이면   →  A 실행
 else 조건이 거짓이면 →  B 실행
```


````python
sales = 80000

if sales >= 100000:
    print("목표 달성!")
else:
    print("목표 미달성 - 개선 필요")
````

````python
# [실습] if-else
# 문제: stock = 8, safety_stock = 10 일 때 재고가 안전 재고보다 적으면 "발주 필요", 아니면 "재고 충분" 을 if-else로 출력한다.
stock = 11
safety_stock = 10
if stock < safety_stock :
    print("발주 필요")
else:
    print("재고 충분")
````

## **4) elif(엘리프):  다중 조건**

`elif`는 **앞의 조건이 거짓(False)일 때 다른 조건을 추가로 검사하는 키워드**이다.

`elif`는 `else if`의 줄임말로, "그렇지 않고 만약"이라는 의미이다.

여러 조건을 판단할 때 `if` 다음에 `elif`를 사용하며, 필요한 경우 `else`를 마지막에 작성할 수 있다.

조건은 위에서부터 순서대로 검사하며, **처음으로 참(True)이 되는 조건의 코드 블록만 실행하고 나머지 조건은 검사하지 않는다.**

```python
sales = 150000

if sales >= 200000:
    print("우수")
elif sales >= 100000:
    print("보통")
else:
    print("미달")

# 출력: 보통
```

위 예제에서는 첫 번째 조건이 `False`이고 두 번째 조건이 `True`이므로 `"보통"`을 출력한다. 이후 `else` 블록은 실행하지 않는다.

`elif`는 여러 번 사용할 수 있으며, `else`는 모든 조건이 거짓일 때 실행된다.

## **5) 들여쓰기(Indentation)**
파이썬은 중괄호 {} 대신 들여쓰기로 코드의 포함 관계를 구분한다. 같은 블록은 같은 칸만큼 들여쓰며, 공식 튜토리얼은 공백 4칸을 쓰고 탭은 쓰지 않도록 권장한다.

들여쓰기가 없는 줄은 조건문의 결과와 상관없이 항상 실행된다.

```text
 if sales >= 100000:
     print("목표 달성")   # 들여쓰기 O → 조건이 참일 때만 실행
 print("확인 완료")       # 들여쓰기 X → 조건과 무관하게 항상 실행
```

## **조건문 작성 규칙**

- 조건문은 반드시 `if`로 시작한다
- `elif`는 필요한 만큼 여러 번 사용할 수 있다. (필요 없으면 생략 가능)
- `else`는 맨 마지막에 한 번만 쓴다. (필요 없으면 생략 가능)
- if, elif, else 줄 끝에는 콜론(:)을 반드시 붙여야 한다.

````python
# [예제] 매출액에 따른 매장 등급 및 안내 메시지 출력

sales = 120000
target = 100000

# 1. 다중 조건문 (if - elif - else)
if sales >= 200000:
    grade = "A등급 (우수)"
elif sales >= target:
    grade = "B등급 (보통)"
else:
    grade = "C등급 (개선 필요)"

print(f"현재 매출: {sales:,}원")
print(f"매장 평가: {grade}")

# 2. 들여쓰기 구분을 통한 실행 흐름 확인
if sales >= target:
    diff = sales - target
    print(f"초과 달성 금액: {diff:,}원") # 조건이 참일 때만 실행

print("--- 매출 평가 처리 종료 ---")     # 조건과 무관하게 항상 실행
````

````python
# [실습] if-elif-else
# 문제: 온도(temperature)가 78.5일 때, 온도가 80 이상이면 "경고",
#       70 이상이면 "주의", 그 외에는 "정상"을 상태(status)에 저장하시오.
#       if-elif-else 조건문을 작성하고, 조건문 밖에서 온도와 상태를 출력하시오.

# [실행 결과]
# 온도 78.5도: 주의
temperature = 78.5
status = ""
if temperature>= 80:
    status = "경고"
elif temperature>= 70:
    status = "주의"
else:
    status = "정상"
print(temperature,"도",status)
````

## **여러 조건을 함께 확인하기 (논리 연산자)**

조건문에서 **두 개 이상의 조건을 함께 검사할 때**는 앞서 배운 논리 연산자(`and`, `or`, `not`)를 사용한다.

- **`and` :** 모든 조건이 참(`True`)일 때 참이 된다.
- **`or` :** 하나 이상의 조건이 참(`True`)이면 참이 된다.
- **`not` :** 조건의 참과 거짓을 반대로 바꾼다.

```python
sales = 150000
region = "서울"

# 매출이 100000원 이상이고 지역이 서울인 경우
if sales >= 100000 and region == "서울":
    print("조건 충족")
else:
    print("조건 미충족")

# 출력: 조건 충족
```

논리 연산자를 사용하면 여러 조건을 하나의 조건식으로 작성할 수 있다.

**공식 문서:** [Python 3.13 — 논리 연산자(Boolean Operations)](https://docs.python.org/ko/3.13/reference/expressions.html#boolean-operations)

````python
# [실습 1] and 연산자: 두 조건이 모두 참일 때만 실행
sales = 180000
region = "서울"

if sales >= 150000 and region == "서울":
    print("서울 고매출 주문")   # 두 조건이 모두 참이므로 실행됨
else:
    print("일반 주문")

# [실습 2] or 연산자: 둘 중 하나만 참이어도 실행
sales = 80000
region = "서울"

# 매출이 10만 원 이상이거나, 지역이 '서울'이면 우대 매장 처리
if sales >= 100000 or region == "서울":
    print("우대 관리 매장")     # sales는 거짓이지만 region이 참이므로 실행됨
````

````python
# [실습] 조건문과 논리 연산자
# 문제: 회원 A 의 주문 금액(amount)은 320000원, 회원 여부(is_member)는 False이다.
#       주문 금액과 회원 여부에 따라 배송비와 할인 적용 여부를 판단하시오.
#       주문 금액이 300000원 이상이거나 회원이면 "무료 배송",
#       그렇지 않으면 "배송비 3,000원"을 출력한다.
#       또한 주문 금액이 500000원 이상이고 회원이면 "VIP 할인 적용",
#       그렇지 않으면 "일반 결제"를 출력한다.
A_amount = 320000
is_member = False
if A_amount>= 300000 or is_member == True:
    print("무료 배송")
else:
    print("배송비 3000원")

if A_amount>=500000 and is_member == True:
    print("VIP 할인 적용")
else:
    print("일반 결제")

````

## **빈 리스트 주의 (`ZeroDivisionError` 예방)**

데이터가 들어있지 않은 빈 리스트(`[]`)에 대해 `sum()`을 실행하면 `0`이 나오지만, **평균(`sum / len`)을 계산하려고 하면 0으로 나누게 되어 `ZeroDivisionError`가 발생하며 프로그램이 멈춘다.**

따라서 평균을 계산하거나 통계량을 낼 때는 반드시 **조건문으로 리스트에 데이터가 존재하는지 먼저 확인**하고 처리해야 한다.

---

###  리스트 포함 여부를 검사하는 두 가지 방식

1. `if len(sales) == 0:` ➔ 리스트의 길이를 직접 측정하여 비교하는 방식
2. `if not sales:` ➔ 빈 리스트가 거짓(`False`)으로 취급되는 성질을 이용한방식

공식 문서: [ZeroDivisionError 예외](https://docs.python.org/ko/3.13/library/exceptions.html#ZeroDivisionError)

````python
# 1. 빈 리스트 생성
sales = []

# 2. 데이터 유무에 따른 조건 처리 (ZeroDivisionError 방지)
if len(sales) == 0:
    average = None
else:
    average = sum(sales) / len(sales)

# 3. 결과 출력
print(average)  # None (에러 없이 안전하게 출력됨)

````

## **Truth Value: 참과 거짓으로 취급되는 값**

조건문(`if`)에는 일반적으로 `sales >= 100000`과 같은 **비교식**을 사용하지만, 값이나 변수만 작성해도 파이썬은 이를 참(`True`) 또는 거짓(`False`)으로 판단한다.

이처럼 객체를 참 또는 거짓으로 평가하는 것을 **진릿값 검사(Truth Value Testing)**라고 한다.

### **거짓(Falsy)으로 취급되는 대표적인 값**

다음 값들은 조건문에서 거짓(`False`)으로 취급된다.

| 구분 | 거짓으로 취급되는 값 |
|---|---|
| 불리언 | `False` |
| 숫자 | `0`, `0.0` |
| 문자열 | `""` |
| 리스트 | `[]` |
| 튜플 | `()` |
| 딕셔너리 | `{}` |
| 집합 | `set()` |
| 값이 없음을 나타내는 객체 | `None` |

반대로 **0이 아닌 숫자, 내용이 있는 문자열, 요소가 들어 있는 리스트** 등은 일반적으로 참(Truthy)으로 취급된다.

### **조건문에서 활용하기**

**리스트에 데이터가 하나라도 들어 있는지 확인**할 때 `if len(sales) > 0:` 대신 `if sales:`처럼 간결하게 조건식을 작성할 수 있다.

앞의 "빈 리스트 주의"에서 본 `if not sales:`도 바로 이 성질을 활용한 것이다.

**공식 문서:** [Python 3.13 — 진릿값 검사(Truth Value Testing)](https://docs.python.org/ko/3.13/library/stdtypes.html#truth-value-testing)

````python
# 1. 거짓으로 취급되는 값(0) 확인
if 0:
    print("참")
else:
    print("거짓 (0은 거짓으로 취급됨)")   # 이 줄이 실행됨

# 2. 데이터가 있는 리스트 (참으로 취급)
sales = [120000, 90000]

if sales:
    print("데이터가 존재함")             # 리스트에 값이 있으므로 참

# 3. 빈 리스트 처리 (거짓으로 취급)
empty_list = []

if not empty_list:
    print("경고: 처리할 데이터가 없습니다.") # 빈 리스트이므로 not False -> True
````

````python
# [실습] Truth Value
# 문제: 비교 연산자 없이 값의 참과 거짓(Truth Value)을 이용하여
#       다음 조건에 따라 결과를 출력하시오.
#
#       memo = ""일 때, 메모가 없으면 "메모 없음"을 출력한다.
#       returns = ["R001", "R002"]일 때, 반품 내역이 있으면 반품 건수를 출력한다.
#       receivable = 0일 때, 미수금이 없으면 "미수금 없음"을 출력한다.
#
#       단, 조건문에는 변수를 직접 사용하거나 not 연산자를 활용한다.
memo = ""
if len(memo) ==0:
    print("메모없음")

returns = ["R001", "R002"]
if returns:
    print(f"반품내역 {len(returns)}건 있음")

receivable = 0
if receivable == 0:
    print("미수금 없음")


# [실행 결과]
# 메모 없음
# 반품 2건 처리 필요
# 미수금 없음
````

# **5. 반복문**

리스트에 매출 100건이 있다고 하자. 각 값을 하나씩 출력하려고 `print()`를 100번 쓸 수는 없다.

**반복문**은 "같은 작업을 여러 값에 대해 자동으로 되풀이"하는 것이다.
  

---



## **for 문의 구조**

```text
        for   item   in    sales:
        │     │             └─ 1. 대상 리스트/순서형 데이터
        │     └─ 2. 하나씩 꺼내서 담을 변수
        └─ 3. 아래 들여쓴 코드를 데이터 개수만큼 반복 실행!

             ┌───────────────────────┐
             │   print(item)         │  <-- 반복 실행할 코드 (들여쓰기 필수)
             └───────────────────────┘
```

`for item in sales:` 는 "sales에서 값을 하나씩 꺼내 `item`에 담고, 아래 들여쓴 코드를 실행한다"는 뜻이다. 리스트의 값이 5개면 아래 코드가 5번 실행된다.

아래 코드는 매출을 하나씩 꺼내 출력한다.

공식 문서: [for 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#for-statements)

````python
sales = [120000, 150000, 90000]

for value in sales:
    print(value)
````

## **반복문 안에서 조건 확인하기 (`for` + `if` 결합)**

반복문(`for`)과 조건문(`if`)을 함께 사용하면 "여러 개의 데이터 중 특정 조건에 맞는 데이터만 필터링하여 처리"할 수 있다.

대량의 데이터에서 특정 기준을 만족하는 값만 골라내거나, 조건별로 통계를 낼 때 이 방식을 가장 기본적으로 사용한다.

---

###  **3단계 실행 흐름**

```text
 1. 리스트에서 값을 하나씩 꺼낸다 (for)
       │
       ▼
 2. 꺼낸 값이 조건에 맞는지 검사한다 (if)
       │
       ├─ 참(True)이면 ──➔ 3. 지정한 코드 실행 (print, 계산 등)
       └─ 거짓(False)이면 ➔ 무시하고 다음 값으로 이동

```


````python
# [실습 1] 기본 필터링: 15만 원 이상인 고매출 데이터만 출력

sales_list = [120000, 180000, 90000, 210000, 150000]

print("=== 15만 원 이상 매출 목록 ===")
for sales in sales_list:
    if sales >= 150000:
        print(f"고매출 발견: {sales:,}원")

# [실습 2] : 조건에 맞는 데이터 개수 세기 (카운팅)
high_sales_count = 0

for sales in sales_list:
    if sales >= 150000:
        high_sales_count += 1 # 조건에 맞을 때만 1씩 증가

print(f"\n 총 {high_sales_count}건")
````

````python
# [실습] for문과 if문
# 문제: 점수 목록 scores = [85, 60, 90, 70, 95]에서
#       for문과 if문을 사용하여 80점 이상인 점수만 출력하시오.

# [실행 결과]
# 85
# 90
# 95

scores = [85, 60, 90, 70, 95]
for item in scores:
    if(item>=80):
        print(item)
````

````python
# [실습 2] for문과 if문으로 개수 세기
# 문제: 점수 목록 scores = [85, 60, 90, 70, 95]에서
#       80점 이상인 점수의 개수를 세어 출력하시오.
#       단, 개수를 저장할 변수 count는 0으로 시작한다.

# [실행 결과]
# 80점 이상: 3개


scores = [85, 60, 90, 70, 95]
count = 0
for num in scores:
    if num >=80:
        count+=1
print(f"80점 이상 {count}개")
````

````python
# [실습 3] for문과 if문으로 리스트 만들기
# 문제: 숫자 목록 numbers = [10, 25, 30, 15, 40]에서
#       20 이상인 숫자만 골라 result 리스트에 추가하고 출력하시오.
#       단, result는 빈 리스트로 시작한다.



numbers = [10, 25, 30, 15, 40]
result = []
for i in numbers:
    if i >= 20:
        result.append(i)
print(result)
# [실행 결과]
# [25, 30, 40]
````

````python
# [실습] for문과 if문
# 문제: 온도 목록에서 80도 이상인 온도를 찾아 경고 메시지를 출력하고,
#       경고 건수를 계산한 뒤 정상 온도만 모은 리스트를 출력하시오.

#       temperatures = [72.1, 81.5, 68.9, 79.8, 85.2]
#       80도 이상이면 "경고: 81.5도" 형식으로 출력한다.
#       80도 이상인 온도의 개수를 warning_count에 저장한다.
#       80도 미만인 온도는 normal 리스트에 추가한다.

temperatures = [72.1, 81.5, 68.9, 79.8, 85.2]
Warning_count = 0
normal = []
for i in temperatures:
    if(i>=80):
        print(f"경고 {i}도")
        Warning_count += 1
        normal.append(i)
print(f"경고 건수: {Warning_count}건")
print(f"정상 온도 :{normal}")
# [실행 결과]
# 경고: 81.5도
# 경고: 85.2도
# 경고 건수: 2건
# 정상 온도: [72.1, 68.9, 79.8]
````

## **문자열 리스트를 숫자 리스트로 바꾸기**

파일이나 시스템에서 가져온 숫자 데이터는 `["100000", "150000", "90000"]`처럼 문자열로 저장된 경우가 많다.

이러한 문자열 데이터를 계산에 사용하려면 숫자로 변환해야 한다.

데이터가 여러 개라면 `for`문을 사용하여 하나씩 변환할 수 있다.

```text
1. 빈 리스트를 만든다.          sales_numbers = []
2. for문으로 값을 꺼낸다.       for value in sales:
3. int()로 정수로 변환한다.     number = int(value)
4. 새 리스트에 추가한다.        sales_numbers.append(number)
```

이 방법은 **문자열 리스트의 요소를 하나씩 정수로 변환하여 새로운 리스트에 저장하는 기본적인 방법**이다.

````python
# 1. 텍스트 데이터 및 split 처리
sales_text = "100000,150000,90000"
sales = sales_text.split(",")  # ['100000', '150000', '90000'] (문자열 리스트)

# 2. 숫자를 담을 빈 리스트 생성
sales_numbers = []

# 3. for 문을 돌며 정수(int)로 변환 후 append
for value in sales:
    sales_numbers.append(int(value))

# 4. 결과 출력 및 통계 계산
print("변환된 숫자 리스트:", sales_numbers) # [100000, 150000, 90000]
print("매출 합계:", sum(sales_numbers))    # 340000
````

## **`while` 반복문**

`for` 문이 지정한 리스트나 횟수만큼 반복한다면, `while` 문은 조건이 참(`True`)인 동안 코드를 계속해서 반복 실행한다.

조건이 계속 참으로 유지되면 프로그램이 끝나지 않는 무한 반복(무한 루프)에 빠질 수 있으므로, 반복을 멈추게 할 탈출 조건(값의 변경 등)을 코드 안에 반드시 작성해야 한다.

### **`while` 문의 구조와 동작 원리**
```text
 count = 0             ➔ 1. 초기 값 설정

 while count < 3:      ➔ 2. 조건 검사 (참이면 아래 코드 실행 / 거짓이면 종료)
     print(count)      ➔ 3. 실행할 코드 (들여쓰기 필수)
     count += 1        ➔ 4. 조건 변경 (반복을 끝내기 위한 핵심)

```
* while 조건식: ➔ 조건식의 결과가 True인 동안 아래 들여쓴 코드를 무한히 반복 실행한다.
* 조건식 검사 ➔ 코드 실행 ➔ 조건식 재검사의 과정이 계속 순환한다.
* 조건 변경 코드: 반복문 안에서 조건식에 영향을 주는 값(예: count += 1)을 바꿔주지 않으면 조건이 평생 True로 남아 무한 루프(Infinite Loop)에 빠진다.
---

###  **`for` 문 vs `while` 문 차이점**

| 구분 | `for` 문 | `while` 문 |
|---|---|---|
| **주요 용도** | 데이터 수나 **반복 횟수가 정해져 있을 때** | 특정 **조건이 만족되는 동안** 계속 반복할 때 |
| **예시** | 리스트 요소 5개 출력하기 | 목표 매출 달성할 때까지 반복하기 |

공식 문서: [while 문 개요](https://docs.python.org/ko/3.13/reference/compound_stmts.html#the-while-statement)

````python
# [실습 1] while 문 기본 활용: count가 3보다 작은 동안 반복
count = 0

while count < 3:
    print(f"현재 count: {count}")
    count += 1  # count를 1씩 증가시켜 언젠가 condition(count < 3)이 False가 되도록 함

print("반복문 종료!")

# [실행 순서]
# 1회차: count = 0 (0 < 3 참) ➔ print(0) ➔ count가 1이 됨
# 2회차: count = 1 (1 < 3 참) ➔ print(1) ➔ count가 2가 됨
# 3회차: count = 2 (2 < 3 참) ➔ print(2) ➔ count가 3이 됨
# 4회차: count = 3 (3 < 3 거짓) ➔ while 문 탈출
````

````python
# [실습] while 문
# 문제:  재고 stock = 50 에서 하루에 12개씩 출고한다. 재고가 12개 미만이 될 때까지 출고를 반복하며
#    매번 "출고 후 재고: 38" 형식으로 출력한다.
# 출력: 출고 후 재고: 38
# 출력: 출고 후 재고: 26
# 출력: 출고 후 재고: 14
# 출력: 출고 후 재고: 2
stock = 50
while stock>12:
    stock -= 12
    print(f"출고 후 재고: {stock}")


# 2. 반복이 끝난 뒤 남은 재고를 출력한다.
# 출력: 남은 재고: 2

print("남은 재고:",stock)
````

## **`break`와 `continue`: 반복 중단과 건너뛰기**

반복문 안에서 흐름을 바꾼다.

| 문장 | 하는 일 | 쓰는 경우 |
|---|---|---|
| `break` | 반복을 **즉시 끝낸다** | 찾던 값을 발견했을 때, 더 볼 필요가 없을 때 |
| `continue` | 이번 차례만 **건너뛰고** 다음 값으로 넘어간다 | 결측이나 잘못된 값을 빼고 처리할 때 |

```text
 for value in values:
     if 조건A:
         break        ➔ 반복문 전체를 빠져나간다
     if 조건B:
         continue     ➔ 아래 코드를 건너뛰고 다음 value로
     처리 코드
```

`break`는 `while` 문에서 탈출 조건을 안쪽에 둘 때도 쓴다. 둘 다 조건문(`if`)과 함께 쓰는 것이 보통이다.

공식 문서: [break 문과 continue 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#break-and-continue-statements)


````python
# 1. break: 첫 불량을 발견하면 검사를 멈춘다
results = ["양품", "양품", "불량", "양품", "불량"]

number = 0
for result in results:
    number += 1           # 몇 번째인지 센다
    if result == "불량":
        print(f"{number}번째에서 첫 불량 발견, 검사 중단")
        break
    print(f"{number}번째 양품")

# 2. continue: 결측값(None)은 건너뛰고 합계를 구한다
sales = [120000, None, 90000, None, 150000, 80000]
total = 0

for value in sales:
    if value is None:
        continue          # None이면 아래 줄을 건너뛴다
    total += value

print("결측 제외 합계:", total)



# 3. 매출 데이터에서 결측값(None)은 건너뛰고 합계를 구한다.
#    None이면 continue로 다음 데이터를 확인한다.
#    누적 매출이 350000원 이상이면 break로 반복문을 종료한다.

total = 0

for value in sales:
    if value is None:
        continue          # 결측값이면 건너뛰기

    total += value

    if total >= 350000:
        print("목표 매출 도달, 계산 중단")
        break             # 반복문 종료

print("결측 제외 합계:", total)
````

````python
# [실습] break와 continue

# 1. 일별 생산량 production = [120, 135, 0, 142, 0, 130] 에서 0(휴무일)은 continue로 건너뛰고
#    가동일의 생산량 합계와 가동일 수를 구해 출력한다.
# 출력: 가동일 4일, 합계 527
production = [120, 135, 0, 142, 0, 130]
product_total= 0
product_days= 0
for i in production:
    if i == 0:
        continue
    
    product_total += i
    product_days +=1
print(f"가동일{product_days}일, 합계 {product_total}")
# 2. temperatures = [72.1, 75.4, 83.0, 79.8, 85.2, 92.0, 76.5] 설비 온도를 순서대로 확인한다.
#    80도 미만 : "정상" 출력 후 continue로 다음 온도를 확인한다.
#    80도 이상 : "경고" 출력 후 다음 온도를 확인한다.
#    90도 이상 : "위험 경보, 점검 중단" 출력 후 break로 반복문을 종료한다.
temperatures = [72.1, 75.4, 83.0, 79.8, 85.2, 92.0, 76.5] 

for i in temperatures:
    if i<80:
        print("정상")
        continue
    if i>=80:
        print("경고")
    if i>= 90:
        print("위험 경보, 점검 중단")
        break
````

### **1. 조건문에서 들여쓰기가 필요한 이유는 무엇인가?**

파이썬은 **들여쓰기(Indentation)**를 사용하여 조건에 따라 실행할 코드 블록을 구분한다.

```python
sales = 150000

if sales >= 100000:
    print("기준 이상")  # 조건이 참일 때만 실행

print("확인 완료")      # 조건의 참/거짓과 관계없이 실행
```

`if` 아래에 포함되는 문장은 같은 깊이로 들여쓴다. 들여쓰기가 잘못되면 `IndentationError`가 발생하거나 의도와 다르게 실행될 수 있다.

### **2. if, elif, else는 각각 어떤 역할을 하며, 여러 조건을 어떤 순서로 검사하는가?**

- `if`: 첫 번째 조건을 검사한다.
- `elif`: 앞의 조건이 거짓일 때 다른 조건을 검사한다.
- `else`: 앞의 조건이 모두 거짓일 때 실행한다.

조건은 위에서부터 순서대로 검사하며, **처음으로 참(True)이 되는 조건의 코드 블록만 실행한다.**

```python
sales = 150000

if sales >= 200000:
    print("우수")
elif sales >= 100000:
    print("보통")
else:
    print("미달")

# 출력: 보통
```

첫 번째 조건은 거짓이고 두 번째 조건이 참이므로 `"보통"`을 출력한다. 이후의 `else` 블록은 실행하지 않는다.

---

### **3. 조건문에서 논리 연산자 and, or, not은 각각 어떻게 사용되는가?**

논리 연산자는 여러 조건을 연결하거나 조건의 참과 거짓을 반대로 바꿀 때 사용한다.

| 연산자 | 조건 |  

| `and` |  모든 조건이 참일 때 참 |  

| `or` | 하나 이상의 조건이 참이면 참 |

| `not` | 참과 거짓을 반대로 변경 |

```python
sales = 150000
region = "서울"

if sales >= 100000 and region == "서울":
    print("조건 충족")
else:
    print("조건 미충족")

# 출력: 조건 충족
```

`and`와 `or`는 **단락 평가(Short-circuit Evaluation)**를 수행하므로, 왼쪽 조건의 결과에 따라 오른쪽 조건을 평가하지 않을 수 있다.

---

### **4. Truth Value란 무엇이며, 빈 리스트를 조건문에서 확인하는 방법은 무엇인가?**

**진릿값 검사(Truth Value Testing)**는 객체를 참(`True`) 또는 거짓(`False`)으로 평가하는 것이다.

조건문에서는 비교식을 작성하지 않고 값이나 변수를 직접 조건으로 사용할 수 있다.

대표적으로 `0`, `0.0`, `""`, `[]`, `()`, `{}`, `set()`, `None`, `False`는 거짓으로 취급된다.

```python
sales = []

if not sales:
    print("데이터가 없습니다.")
else:
    print("데이터가 있습니다.")

# 출력: 데이터가 없습니다.
```

`if sales:`는 리스트에 요소가 있을 때 참이 되고, `if not sales:`는 리스트가 비어 있을 때 참이 된다.

### **5. for문과 while문의 차이는 무엇인가?**

`for`문은 리스트와 같은 반복 가능한 객체에서 요소를 하나씩 꺼내 처리할 때 사용한다.

`while`문은 **조건이 참(True)인 동안 코드 블록을 반복 실행**한다.

**① for문**

```python
machines = ["M01", "M02", "M03"]

for machine in machines:
    print(machine)

# M01
# M02
# M03
```

**② while문**

```python
count = 1

while count <= 3:
    print(count)
    count += 1

# 1
# 2
# 3
```

`while`문은 조건이 계속 참이면 반복이 종료되지 않을 수 있으므로, 종료 조건에 도달하도록 작성해야 한다.

### **6. break와 continue는 어떻게 다른가?**

- `break`: **반복문 자체를 즉시 종료**한다.
- `continue`: **현재 반복의 남은 코드를 건너뛰고 다음 반복으로 진행**한다.

**① break 사용**

```python
for number in [10, 20, 30, 40]:
    if number == 30:
        break
    print(number)

# 10
# 20
```

`30`을 만나면 반복문이 종료되므로 이후의 `40`은 처리하지 않는다.

**② continue 사용**

```python
for number in [10, 20, 30, 40]:
    if number == 30:
        continue
    print(number)

# 10
# 20
# 40
```

`30`을 만나면 해당 반복의 남은 코드를 건너뛰고 다음 값인 `40`을 처리한다.
