```python
# 1. 라이브러리 불러오기
import numpy as np
import pandas as pd

print("NumPy 버전:", np.__version__)
print("Pandas 버전:", pd.__version__)
```

### 2. NumPy: 수치 연산과 배열(Array) 다루기
대량의 숫자 데이터를 빠르게 계산하고 통계치를 구할 때 사용합니다.

```python
# 1부터 10까지 숫자 배열 생성
arr = np.arange(1, 11)
print("배열 생성:", arr)
print("평균:", arr.mean())
print("합계:", arr.sum())
print("각 원소에 10 곱하기:", arr * 10)
```

### 3. Pandas: 데이터프레임(표) 생성하기
엑셀 표와 같은 2차원 테이블 데이터를 다룰 때 사용합니다.

```python
# 딕셔너리로 샘플 데이터프레임 생성
data = {
    '차종': ['아반떼', '쏘나타', '그랜저', '투싼', '싼타페'],
    '가격(만원)': [2200, 2800, 3700, 2700, 3500],
    '판매량': [120, 95, 140, 80, 110]
}

df = pd.DataFrame(data)

# 주피터에서는 print(df) 대신 변수명만 적으면 예쁜 표로 렌더링됩니다.
df
```

### 4. 데이터 가공 및 필터링
새로운 계산 컬럼을 추가하고, 원하는 조건의 데이터만 추출해 봅니다.

```python
# '총매출(만원)' 컬럼 추가 (가격 x 판매량)
df['총매출(만원)'] = df['가격(만원)'] * df['판매량']

# 판매량이 100대 이상인 차종만 필터링
filtered_df = df[df['판매량'] >= 100]
filtered_df
```
